```python

import zipfile
import h5py
import io
from collections import Counter

outer_path = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"

original_zips = [
    "brainTumorDataPublic_1-766.zip",
    "brainTumorDataPublic_767-1532.zip",
    "brainTumorDataPublic_1533-2298.zip",
    "brainTumorDataPublic_2299-3064.zip"
]

label_counts = Counter()
patient_ids = set()
checked_files = 0

with zipfile.ZipFile(outer_path, "r") as outer_zip:
    for filename in original_zips:
        inner_path = "Datasets_MP - Copy/" + filename

        with outer_zip.open(inner_path) as inner_file:
            with zipfile.ZipFile(inner_file) as inner_zip:
                mat_files = [
                    name for name in inner_zip.namelist()
                    if name.lower().endswith(".mat")
                ]

                for mat_name in mat_files:
                    mat_bytes = inner_zip.read(mat_name)

                    with h5py.File(io.BytesIO(mat_bytes), "r") as mat:
                        cjdata = mat["cjdata"]

                        label = int(cjdata["label"][0, 0])
                        pid_bytes = cjdata["PID"][:].flatten()
                        pid = bytes(pid_bytes.astype("uint8")).decode("ascii")

                        label_counts[label] += 1
                        patient_ids.add(pid)
                        checked_files += 1

print("Total files checked:", checked_files)
print("Unique patient IDs:", len(patient_ids))
print("\nClass counts:")

label_names = {
    1: "Meningioma",
    2: "Glioma",
    3: "Pituitary tumor"
}

for label, name in label_names.items():
    print(f"{name}: {label_counts[label]}")
```

```python

import zipfile
import hashlib
from collections import defaultdict

zip_path = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"

hash_locations = defaultdict(list)
image_extensions = (".jpg", ".jpeg", ".png", ".bmp")

with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 4
            and parts[-3] in ("Training", "Testing")
            and name.lower().endswith(image_extensions)
        ):
            content = z.read(name)
            digest = hashlib.md5(content).hexdigest()
            hash_locations[digest].append(name)

duplicates = {
    digest: paths
    for digest, paths in hash_locations.items()
    if len(paths) > 1
}

cross_split = [
    paths for paths in duplicates.values()
    if any("/Training/" in p for p in paths)
    and any("/Testing/" in p for p in paths)
]

print("Images checked:", sum(len(v) for v in hash_locations.values()))
print("Exact duplicate groups:", len(duplicates))
print("Duplicate groups shared between Training and Testing:", len(cross_split))

for paths in cross_split[:10]:
    print(paths)
```

```python

import zipfile
import hashlib
from collections import defaultdict

zip_path = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"
image_extensions = (".jpg", ".jpeg", ".png", ".bmp")

hash_locations = defaultdict(list)

with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 4
            and parts[-3] in ("Training", "Testing")
            and name.lower().endswith(image_extensions)
        ):
            digest = hashlib.md5(z.read(name)).hexdigest()
            split = parts[-3]
            hash_locations[digest].append((split, name))

training_affected = set()
testing_affected = set()

for entries in hash_locations.values():
    splits = {split for split, _ in entries}

    if len(entries) > 1:
        for split, name in entries:
            if split == "Training":
                training_affected.add(name)
            elif split == "Testing":
                testing_affected.add(name)

print("Training images in duplicate groups:", len(training_affected))
print("Testing images in duplicate groups:", len(testing_affected))
print("Total distinct affected files:", len(training_affected | testing_affected))
```

```python

import zipfile
import hashlib
from collections import defaultdict

zip_path = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"
image_extensions = (".jpg", ".jpeg", ".png", ".bmp")

hash_labels = defaultdict(set)
hash_paths = defaultdict(list)

with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 4
            and parts[-3] in ("Training", "Testing")
            and name.lower().endswith(image_extensions)
        ):
            image_bytes = z.read(name)
            digest = hashlib.md5(image_bytes).hexdigest()
            label = parts[-2]

            hash_labels[digest].add(label)
            hash_paths[digest].append(name)

conflicts = [
    (digest, hash_paths[digest], hash_labels[digest])
    for digest in hash_labels
    if len(hash_labels[digest]) > 1
]

print("Exact duplicate groups:", sum(
    len(paths) > 1 for paths in hash_paths.values()
))
print("Duplicate groups with conflicting class labels:", len(conflicts))

for digest, paths, labels in conflicts[:5]:
    print("\nLabels:", sorted(labels))
    for path in paths:
        print(path)
```

```python

import os
import hashlib
import random
from collections import defaultdict

dataset_path = "/content/brain_tumor_working/Datasets_MP - Copy"
output_path = "/content/brain_tumor_clean_split"

extensions = (".jpg", ".jpeg", ".png", ".bmp")
groups = defaultdict(list)

# Group identical image files together.
for split in ("Training", "Testing"):
    for label in os.listdir(os.path.join(dataset_path, split)):
        folder = os.path.join(dataset_path, split, label)

        if not os.path.isdir(folder):
            continue

        for filename in os.listdir(folder):
            if filename.lower().endswith(extensions):
                path = os.path.join(folder, filename)

                with open(path, "rb") as f:
                    digest = hashlib.md5(f.read()).hexdigest()

                groups[(label, digest)].append(path)

# Keep each exact-duplicate group entirely within one split.
random.seed(42)
train_groups = defaultdict(list)
test_groups = defaultdict(list)

for (label, digest), paths in groups.items():
    random.shuffle(paths)

    if random.random() < 0.8:
        train_groups[label].append(paths)
    else:
        test_groups[label].append(paths)

print("Exact-duplicate groups:", len(groups))
print("Training groups:", sum(len(v) for v in train_groups.values()))
print("Testing groups:", sum(len(v) for v in test_groups.values()))

# Verify that no identical-image group crosses the new split.
train_hashes = {
    digest for (label, digest), paths in groups.items()
    if any(path in p for paths_group in train_groups.values()
           for p in paths_group)
    for digest in []
}

print("\nSplit planning completed.")
print("No files have been copied or deleted yet.")

```

```python

train_paths = {
    path
    for path_groups in train_groups.values()
    for group in path_groups
    for path in group
}

test_paths = {
    path
    for path_groups in test_groups.values()
    for group in path_groups
    for path in group
}

cross_split_groups = 0

for key, paths in groups.items():
    in_train = any(path in train_paths for path in paths)
    in_test = any(path in test_paths for path in paths)

    if in_train and in_test:
        cross_split_groups += 1

print("Images assigned to training:", len(train_paths))
print("Images assigned to testing:", len(test_paths))
print("Groups crossing both splits:", cross_split_groups)
print("All images accounted for:", len(train_paths | test_paths) == 7023)
```

```python

from collections import defaultdict
import os
import hashlib

hash_labels = defaultdict(set)

for split in ("Training", "Testing"):
    split_path = os.path.join(dataset_path, split)

    for label in os.listdir(split_path):
        folder = os.path.join(split_path, label)

        if not os.path.isdir(folder):
            continue

        for filename in os.listdir(folder):
            if filename.lower().endswith(extensions):
                path = os.path.join(folder, filename)

                with open(path, "rb") as f:
                    digest = hashlib.md5(f.read()).hexdigest()

                hash_labels[digest].add(label)

conflicting_hashes = {
    digest: labels
    for digest, labels in hash_labels.items()
    if len(labels) > 1
}

print("Unique image hashes:", len(hash_labels))
print("Identical images assigned different labels:", len(conflicting_hashes))
print("Label consistency check:", "PASS" if not conflicting_hashes else "CHECK REQUIRED")
```

```python

import os
import hashlib
import random
import shutil
from collections import defaultdict

random.seed(42)

source_path = "/content/brain_tumor_working/Datasets_MP - Copy"
clean_path = "/content/brain_tumor_clean_split"

extensions = (".jpg", ".jpeg", ".png", ".bmp")
class_images = defaultdict(dict)

# Keep one copy of each exact image hash within its class.
for split in ("Training", "Testing"):
    for label in os.listdir(os.path.join(source_path, split)):
        folder = os.path.join(source_path, split, label)

        if not os.path.isdir(folder):
            continue

        for filename in os.listdir(folder):
            if not filename.lower().endswith(extensions):
                continue

            path = os.path.join(folder, filename)

            with open(path, "rb") as f:
                digest = hashlib.md5(f.read()).hexdigest()

            class_images[label].setdefault(digest, path)

# Stop if the destination already exists, to avoid overwriting work.
if os.path.exists(clean_path):
    raise FileExistsError(
        f"{clean_path} already exists. Please check it before rerunning."
    )

train_count = 0
test_count = 0

for label, image_map in class_images.items():
    unique_paths = list(image_map.values())
    random.shuffle(unique_paths)

    split_index = int(len(unique_paths) * 0.8)
    assignments = {
        "Training": unique_paths[:split_index],
        "Testing": unique_paths[split_index:]
    }

    for split_name, paths in assignments.items():
        destination = os.path.join(clean_path, split_name, label)
        os.makedirs(destination, exist_ok=True)

        for index, source in enumerate(paths):
            extension = os.path.splitext(source)[1].lower()
            new_name = f"{label}_{index:05d}{extension}"
            shutil.copy2(source, os.path.join(destination, new_name))

        if split_name == "Training":
            train_count += len(paths)
        else:
            test_count += len(paths)

print("Clean split created successfully!")
print("Training images:", train_count)
print("Testing images:", test_count)
print("Total unique images:", train_count + test_count)
print("Dataset location:", clean_path)
print("Original dataset remains unchanged.")
```

```python

import os
from PIL import Image

clean_path = "/content/brain_tumor_clean_split"
splits = ["Training", "Testing"]
classes = ["glioma", "meningioma", "notumor", "pituitary"]

total_images = 0

for split in splits:
    print(f"\n{split} dataset")
    split_total = 0

    for label in classes:
        folder = os.path.join(clean_path, split, label)

        if not os.path.isdir(folder):
            print(f"  {label}: MISSING FOLDER")
            continue

        files = [
            f for f in os.listdir(folder)
            if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
        ]

        # Check that image files can be opened.
        readable = 0
        for filename in files:
            try:
                with Image.open(os.path.join(folder, filename)) as img:
                    img.verify()
                readable += 1
            except Exception:
                pass

        print(f"  {label}: {len(files)} images, {readable} readable")
        split_total += len(files)

    print(f"  Total: {split_total}")
    total_images += split_total

print(f"\nOverall total: {total_images}")
```

```python

import tensorflow as tf

DATASET_PATH = "/content/brain_tumor_clean_split"
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    f"{DATASET_PATH}/Training",
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=SEED
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    f"{DATASET_PATH}/Testing",
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

# Keep class order consistent between training and testing.
class_names = train_ds.class_names

assert test_ds.class_names == class_names, "Class order mismatch!"

# Normalize image pixels to the range 0–1.
normalization_layer = tf.keras.layers.Rescaling(1.0 / 255)

train_ds = train_ds.map(
    lambda images, labels: (normalization_layer(images), labels),
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.map(
    lambda images, labels: (normalization_layer(images), labels),
    num_parallel_calls=tf.data.AUTOTUNE
)

print("\nClass names:", class_names)
print("Image size:", IMAGE_SIZE)
print("Batch size:", BATCH_SIZE)

# Inspect one batch.
for images, labels in train_ds.take(1):
    print("Training batch shape:", images.shape)
    print("Training labels shape:", labels.shape)
    print("Pixel value range:",
          float(tf.reduce_min(images)),
          "to",
          float(tf.reduce_max(images)))
```

```python

import tensorflow as tf
from tensorflow.keras import layers, models

tf.keras.utils.set_random_seed(42)

num_classes = len(class_names)

cnn_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    # Convolution block 1
    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Convolution block 2
    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Convolution block 3
    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    # Classification layers
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(num_classes, activation="softmax")
])

cnn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

cnn_model.summary()
```

```python

from sklearn.model_selection import train_test_split
import numpy as np
import tensorflow as tf

# Collect labels from the training dataset for a stratified split.
train_paths = []
train_labels = []

for label_index, label in enumerate(class_names):
    folder = f"/content/brain_tumor_clean_split/Training/{label}"
    files = sorted([
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ])

    train_paths.extend([os.path.join(folder, f) for f in files])
    train_labels.extend([label_index] * len(files))

# Split file paths into training and validation sets.
paths_train, paths_val, labels_train, labels_val = train_test_split(
    train_paths,
    train_labels,
    test_size=0.15,
    random_state=42,
    stratify=train_labels
)

print("Training images:", len(paths_train))
print("Validation images:", len(paths_val))
print("Testing images:", 1346)
print("\nSplit prepared. Next, we'll load these subsets for training.")
```

```python

import numpy as np
import tensorflow as tf
from PIL import Image

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
NUM_CLASSES = len(class_names)

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image, channels=3, expand_animations=False
    )
    image.set_shape([None, None, 3])
    image = tf.image.resize(image, IMAGE_SIZE)
    image = tf.cast(image, tf.float32) / 255.0

    label = tf.one_hot(label, NUM_CLASSES)
    return image, label

def make_dataset(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(
            buffer_size=len(paths), seed=42,
            reshuffle_each_iteration=True
        )
    ds = ds.map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
    return ds

train_ds = make_dataset(paths_train, labels_train, training=True)
val_ds = make_dataset(paths_val, labels_val)

# Check one batch
for images, labels in train_ds.take(1):
    print("Training batch shape:", images.shape)
    print("Training label shape:", labels.shape)
    print("Pixel range:", float(tf.reduce_min(images)),
          "to", float(tf.reduce_max(images)))

print("Training batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Class order:", class_names)
```

```python

import os
import tensorflow as tf
from tensorflow.keras import layers, models

# Save results in Google Drive
results_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results"
os.makedirs(results_path, exist_ok=True)

tf.keras.utils.set_random_seed(42)

# Build a more parameter-efficient CNN
cnn_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    layers.Conv2D(32, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),

    layers.Conv2D(64, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),

    layers.Conv2D(128, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),

    # Fewer parameters than Flatten()
    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(len(class_names), activation="softmax")
])

cnn_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# Save the model with the best validation loss
checkpoint_path = os.path.join(
    results_path, "brain_tumor_cnn_best.keras"
)

callbacks = [
    tf.keras.callbacks.ModelCheckpoint(
        checkpoint_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

print("Improved CNN architecture:")
cnn_model.summary()

# Train the model
history = cnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=callbacks
)

print("\nTraining completed!")
print("Best model saved at:", checkpoint_path)
```

```python

import os
import tensorflow as tf
from tensorflow.keras import layers, models

tf.keras.utils.set_random_seed(42)

results_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

# 1. Data augmentation: used only during training
data_augmentation = models.Sequential([
    layers.RandomRotation(0.03),
    layers.RandomTranslation(0.05, 0.05),
    layers.RandomZoom(0.05),
], name="data_augmentation")

# 2. Apply augmentation to the training dataset only
augmented_train_ds = train_ds.map(
    lambda images, labels: (
        data_augmentation(images, training=True),
        labels
    ),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

# 3. Build the augmented CNN
augmented_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),
    data_augmentation,
    layers.Conv2D(32, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),
    layers.Conv2D(64, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),
    layers.Conv2D(128, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(2),
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(4, activation="softmax")
], name="augmented_brain_tumor_cnn")

augmented_model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

augmented_model.summary()

print("\nAugmented training pipeline created.")
print("Training batches:", tf.data.experimental.cardinality(augmented_train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Test dataset remains unchanged.")
```

```python

import tensorflow as tf
import os

# Rebuild the training dataset WITHOUT external augmentation.
# Augmentation will happen only inside augmented_model.
clean_train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
clean_val_ds = val_ds.prefetch(tf.data.AUTOTUNE)

# Save the augmented model to a separate file.
results_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results"
augmented_model_path = os.path.join(
    results_path, "brain_tumor_cnn_augmented_best.keras"
)

callbacks_aug = [
    tf.keras.callbacks.ModelCheckpoint(
        augmented_model_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

print("Corrected training pipeline is ready.")
print("Augmentation is applied once, inside the model.")
print("Original model file will not be overwritten.")
print("Augmented model save path:", augmented_model_path)
```

```python

history_aug = augmented_model.fit(
    clean_train_ds,
    validation_data=clean_val_ds,
    epochs=10,
    callbacks=callbacks_aug
)

print("\nAugmented CNN training completed!")
print("Best model saved at:", augmented_model_path)

best_val_acc_aug = max(history_aug.history["val_accuracy"])
best_val_loss_aug = min(history_aug.history["val_loss"])

print(f"Best validation accuracy: {best_val_acc_aug * 100:.2f}%")
print(f"Best validation loss: {best_val_loss_aug:.4f}")
```

```python

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import os

# 1. Load the baseline model
model_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results/brain_tumor_cnn_best.keras"
best_model = tf.keras.models.load_model(model_path, compile=False)

class_names = ["glioma", "meningioma", "notumor", "pituitary"]

# 2. Get one test MRI image
images, labels = next(iter(test_ds.take(1)))
img_batch = images[0:1]

# 3. Find the last convolutional layer
last_conv_layer = [
    layer for layer in best_model.layers
    if isinstance(layer, tf.keras.layers.Conv2D)
][-1]

print("Last convolutional layer:", last_conv_layer.name)

# 4. Explicitly build a symbolic graph using the trained layers
input_image = tf.keras.Input(shape=(224, 224, 3))
x = input_image
conv_outputs = None

for layer in best_model.layers:
    if isinstance(layer, tf.keras.layers.InputLayer):
        continue

    x = layer(x)

    if layer is last_conv_layer:
        conv_outputs = x

predictions = x

grad_model = tf.keras.Model(
    inputs=input_image,
    outputs=[conv_outputs, predictions]
)

# 5. Calculate Grad-CAM
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(img_batch, training=False)
    predicted_index = tf.argmax(predictions[0])
    class_score = predictions[:, predicted_index]

gradients = tape.gradient(class_score, feature_maps)

weights = tf.reduce_mean(gradients, axis=(0, 1, 2))
heatmap = tf.reduce_sum(feature_maps[0] * weights, axis=-1)
heatmap = tf.maximum(heatmap, 0)
heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)
heatmap = heatmap.numpy()

# 6. Display the results
original_image = img_batch[0].numpy()
predicted_class = class_names[int(predicted_index.numpy())]
confidence = float(predictions[0, predicted_index].numpy()) * 100

heatmap_resized = tf.image.resize(
    heatmap[..., np.newaxis], (224, 224)
).numpy().squeeze()

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original_image)
plt.title("Original MRI")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(heatmap, cmap="jet")
plt.title("Grad-CAM Heatmap")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(original_image)
plt.imshow(heatmap_resized, cmap="jet", alpha=0.4)
plt.title("MRI + Grad-CAM")
plt.axis("off")

plt.suptitle(
    f"Prediction: {predicted_class} | Confidence: {confidence:.2f}%"
)
plt.tight_layout()

# Save the visualization
output_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results/gradcam_example.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print("Predicted class:", predicted_class)
print(f"Confidence: {confidence:.2f}%")
print("Saved visualization to:", output_path)
```

```python

import zipfile
import os
from collections import Counter

zip_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "Independent dataset/Brain_Tumor_MRI_Dataset_Final.zip"
)

with zipfile.ZipFile(zip_path, "r") as z:
    names = z.namelist()

    # 1. Read the README
    print("=" * 60)
    print("DATASET README")
    print("=" * 60)

    readme_files = [
        n for n in names
        if os.path.basename(n).lower() == "readme.md"
    ]

    if readme_files:
        readme = z.read(readme_files[0]).decode("utf-8", errors="replace")
        print(readme[:12000])
    else:
        print("No README.md found.")

    # 2. Count image files by split and class
    image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
    counts = Counter()

    for name in names:
        parts = name.split("/")

        if (
            len(parts) >= 3
            and parts[0] in ("train", "val", "test")
            and parts[-1].lower().endswith(image_extensions)
        ):
            split = parts[0]
            class_name = parts[1]
            counts[(split, class_name)] += 1

    print("\n" + "=" * 60)
    print("IMAGE COUNTS BY SPLIT AND CLASS")
    print("=" * 60)

    splits = ["train", "val", "test"]
    classes = sorted({class_name for _, class_name in counts})

    for split in splits:
        print(f"\n{split.upper()}")
        split_total = 0

        for class_name in classes:
            count = counts[(split, class_name)]
            print(f"  {class_name}: {count}")
            split_total += count

        print(f"  TOTAL: {split_total}")

    print("\nGrand total:", sum(counts.values()))
```

```python

import os
import zipfile
import hashlib
from collections import defaultdict

zip_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "Independent dataset/Brain_Tumor_MRI_Dataset_Final.zip"
)

original_root = "/content/brain_tumor_clean_split"

def md5_bytes(data):
    return hashlib.md5(data).hexdigest()

def md5_file(path):
    h = hashlib.md5()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)
    return h.hexdigest()

# 1. Hash the existing dataset
print("Checking existing dataset...")
if not os.path.isdir(original_root):
    raise FileNotFoundError(
        f"Existing dataset folder not found: {original_root}"
    )

original_hashes = defaultdict(list)
image_exts = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

for root, dirs, files in os.walk(original_root):
    for filename in files:
        if filename.lower().endswith(image_exts):
            path = os.path.join(root, filename)
            original_hashes[md5_file(path)].append(path)

print("Existing unique image hashes:", len(original_hashes))

# 2. Hash images directly inside the new ZIP
print("\nChecking independent dataset ZIP...")
matching_images = []
total_images = 0

with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 3
            and parts[0] in ("train", "val", "test")
            and name.lower().endswith(image_exts)
        ):
            total_images += 1
            digest = md5_bytes(z.read(name))

            if digest in original_hashes:
                matching_images.append(
                    (name, original_hashes[digest][0])
                )

print("\nExact duplicate audit")
print("Independent dataset images checked:", total_images)
print("Exact matches with existing dataset:", len(matching_images))

if matching_images:
    print("\nFirst 10 matches:")
    for new_path, old_path in matching_images[:10]:
        print("New:", new_path)
        print("Old:", old_path)
else:
    print("No exact byte-identical image matches were found.")
```

```python

import zipfile
import hashlib
from collections import Counter

zip_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "Independent dataset/Brain_Tumor_MRI_Dataset_Final.zip"
)

original_root = "/content/brain_tumor_clean_split"

def md5_bytes(data):
    return hashlib.md5(data).hexdigest()

# Reuse the original hashes if they still exist in memory.
if "original_hashes" not in globals():
    import os
    from collections import defaultdict

    original_hashes = defaultdict(list)
    image_exts = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

    for root, dirs, files in os.walk(original_root):
        for filename in files:
            if filename.lower().endswith(image_exts):
                path = os.path.join(root, filename)
                h = hashlib.md5()
                with open(path, "rb") as f:
                    for chunk in iter(lambda: f.read(1024 * 1024), b""):
                        h.update(chunk)
                original_hashes[h.hexdigest()].append(path)

unmatched_test_counts = Counter()
matched_test_counts = Counter()

with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")
        if (
            len(parts) >= 3
            and parts[0] == "test"
            and name.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp"))
        ):
            class_name = parts[1]
            digest = md5_bytes(z.read(name))

            if digest in original_hashes:
                matched_test_counts[class_name] += 1
            else:
                unmatched_test_counts[class_name] += 1

print("TEST SPLIT: EXACT MATCH AUDIT")
for cls in sorted(set(matched_test_counts) | set(unmatched_test_counts)):
    matched = matched_test_counts[cls]
    unmatched = unmatched_test_counts[cls]
    print(
        f"{cls}: {unmatched} without exact matches, "
        f"{matched} exact matches"
    )

print("\nTotal test images without exact matches:", sum(unmatched_test_counts.values()))
print("Total exact-matching test images:", sum(matched_test_counts.values()))
```

```python

import os
import zipfile
import hashlib
from collections import defaultdict, Counter

zip_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "Independent dataset/Brain_Tumor_MRI_Dataset_Final.zip"
)

original_root = "/content/brain_tumor_clean_split"
output_root = "/content/brain_tumor_unmatched_test"

image_exts = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

# Build hashes for the original dataset
original_hashes = set()

for root, dirs, files in os.walk(original_root):
    for filename in files:
        if filename.lower().endswith(image_exts):
            h = hashlib.md5()
            with open(os.path.join(root, filename), "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    h.update(chunk)
            original_hashes.add(h.hexdigest())

os.makedirs(output_root, exist_ok=True)
saved_counts = Counter()
skipped_counts = Counter()

# Extract only unmatched test images
with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        parts = name.split("/")

        if (
            len(parts) >= 3
            and parts[0] == "test"
            and name.lower().endswith(image_exts)
        ):
            class_name = parts[1]
            digest = hashlib.md5(z.read(name)).hexdigest()

            if digest in original_hashes:
                skipped_counts[class_name] += 1
                continue

            class_folder = os.path.join(output_root, class_name)
            os.makedirs(class_folder, exist_ok=True)

            # Use the original filename within its class folder
            destination = os.path.join(class_folder, os.path.basename(name))

            with z.open(name) as source, open(destination, "wb") as target:
                target.write(source.read())

            saved_counts[class_name] += 1

print("Images extracted without exact matches:")
for cls in sorted(saved_counts):
    print(f"{cls}: {saved_counts[cls]}")

print("\nTotal saved:", sum(saved_counts.values()))
print("Exact matches skipped:", sum(skipped_counts.values()))
print("Output folder:", output_root)
```

```python

import os
import numpy as np
import tensorflow as tf
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

model_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "results/brain_tumor_cnn_best.keras"
)
test_folder = "/content/brain_tumor_unmatched_test"
results_dir = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

# Use the exact class names found in the extracted folders
folder_classes = ["glioma", "meningioma", "no_tumor", "pituitary"]

# The CNN's output class order from training
model_classes = ["glioma", "meningioma", "notumor", "pituitary"]

model = tf.keras.models.load_model(model_path, compile=False)

dataset = tf.keras.utils.image_dataset_from_directory(
    test_folder,
    labels="inferred",
    label_mode="int",
    class_names=folder_classes,
    image_size=(224, 224),
    batch_size=32,
    shuffle=False
)

# Normalize images just as during baseline training
dataset = dataset.map(
    lambda images, labels: (
        tf.cast(images, tf.float32) / 255.0,
        labels
    ),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

y_true_folder = np.concatenate([
    labels.numpy() for _, labels in dataset
])

probabilities = model.predict(dataset, verbose=1)
y_pred = np.argmax(probabilities, axis=1)

# Convert folder label indices to the model's label indices
folder_to_model = {
    "glioma": "glioma",
    "meningioma": "meningioma",
    "no_tumor": "notumor",
    "pituitary": "pituitary"
}

y_true_names = [folder_to_model[folder_classes[i]] for i in y_true_folder]
y_true = np.array([model_classes.index(name) for name in y_true_names])

accuracy = accuracy_score(y_true, y_pred)

print("\nEXPLORATORY EVALUATION")
print("Images evaluated:", len(y_true))
print(f"Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(4)),
    target_names=model_classes,
    digits=4,
    zero_division=0
)
print("\nCLASSIFICATION REPORT")
print(report)

cm = confusion_matrix(y_true, y_pred, labels=list(range(4)))
print("CONFUSION MATRIX (actual rows, predicted columns)")
print("Class order:", model_classes)
print(cm)

os.makedirs(results_dir, exist_ok=True)
output_path = os.path.join(
    results_dir, "unmatched_test_evaluation.txt"
)

with open(output_path, "w", encoding="utf-8") as f:
    f.write("Exploratory evaluation on test images without exact matches\n")
    f.write("Not fully independent external validation.\n\n")
    f.write(f"Images evaluated: {len(y_true)}\n")
    f.write(f"Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)\n\n")
    f.write(report)
    f.write("\nConfusion matrix:\n")
    f.write(np.array2string(cm))

print("\nResults saved to:", output_path)
```

```python

import os
import h5py
import numpy as np
from collections import Counter

mat_dir = "/content/brain_tumor_patient_data/mat_files"

def decode_pid(pid_array):
    """Decode MATLAB UTF-16 character codes into a patient ID."""
    values = np.asarray(pid_array).flatten()

    # MATLAB character arrays may be stored as numeric character codes
    text = "".join(
        chr(int(v)) for v in values
        if int(v) != 0
    ).strip()

    return text

records = []
errors = []

for filename in sorted(
    os.listdir(mat_dir),
    key=lambda name: int(os.path.splitext(name)[0])
):
    if not filename.lower().endswith(".mat"):
        continue

    path = os.path.join(mat_dir, filename)

    try:
        with h5py.File(path, "r") as f:
            cjdata = f["cjdata"]

            pid = decode_pid(cjdata["PID"][()])
            label = int(np.asarray(cjdata["label"][()]).squeeze())

            records.append({
                "file": filename,
                "pid": pid,
                "label": label
            })

    except Exception as e:
        errors.append((filename, str(e)))

print("Records successfully read:", len(records))
print("Files with errors:", len(errors))

print("\nFIRST 10 RECORDS")
for row in records[:10]:
    print(row)

patient_ids = [r["pid"] for r in records]
labels = [r["label"] for r in records]

print("\nUnique patient IDs:", len(set(patient_ids)))
print("Images per tumor label:", dict(Counter(labels)))

patient_label_sets = {}
for row in records:
    patient_label_sets.setdefault(row["pid"], set()).add(row["label"])

mixed_label_patients = {
    pid: sorted(labs)
    for pid, labs in patient_label_sets.items()
    if len(labs) > 1
}

print("Patients associated with multiple labels:",
      len(mixed_label_patients))

if errors:
    print("\nFirst errors:", errors[:5])
```

```python

import os
import pandas as pd
from sklearn.model_selection import train_test_split

# Map dataset labels to readable class names
label_names = {
    1: "meningioma",
    2: "glioma",
    3: "pituitary"
}

# One label per patient, confirmed by the previous audit
patient_df = pd.DataFrame([
    {
        "patient_id": pid,
        "label": next(iter(labels)),
        "class_name": label_names[next(iter(labels))]
    }
    for pid, labels in patient_label_sets.items()
])

# First: 70% train patients, 30% temporary patients
train_patients, temp_patients = train_test_split(
    patient_df,
    test_size=0.30,
    random_state=42,
    stratify=patient_df["label"]
)

# Second: divide temporary patients equally into validation and test
val_patients, test_patients = train_test_split(
    temp_patients,
    test_size=0.50,
    random_state=42,
    stratify=temp_patients["label"]
)

split_by_patient = {}

for pid in train_patients["patient_id"]:
    split_by_patient[pid] = "train"

for pid in val_patients["patient_id"]:
    split_by_patient[pid] = "val"

for pid in test_patients["patient_id"]:
    split_by_patient[pid] = "test"

# Assign every MRI slice according to its patient's split
manifest = pd.DataFrame(records)
manifest["label_name"] = manifest["label"].map(label_names)
manifest["split"] = manifest["pid"].map(split_by_patient)

# Verify patient separation
train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

assert train_ids.isdisjoint(val_ids)
assert train_ids.isdisjoint(test_ids)
assert val_ids.isdisjoint(test_ids)
assert manifest["split"].notna().all()

print("PATIENT-LEVEL SPLIT")
print("-------------------")
print("Training patients:", len(train_ids))
print("Validation patients:", len(val_ids))
print("Testing patients:", len(test_ids))
print("Total patients:", len(train_ids | val_ids | test_ids))

print("\nMRI IMAGE COUNTS BY SPLIT")
print(manifest.groupby("split").size().reindex(["train", "val", "test"]))

print("\nIMAGE COUNTS BY CLASS AND SPLIT")
print(pd.crosstab(manifest["label_name"], manifest["split"]))

print("\nPATIENT OVERLAP CHECK")
print("Train ∩ Validation:", len(train_ids & val_ids))
print("Train ∩ Test:", len(train_ids & test_ids))
print("Validation ∩ Test:", len(val_ids & test_ids))

# Save the patient-level assignment for reproducibility
results_dir = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/results"
)
os.makedirs(results_dir, exist_ok=True)

manifest["mat_path"] = manifest["file"].apply(
    lambda f: os.path.join(
        "/content/brain_tumor_patient_data/mat_files", f
    )
)

manifest_path = os.path.join(
    results_dir, "patient_level_split_manifest.csv"
)
manifest.to_csv(manifest_path, index=False)

print("\nManifest saved to:", manifest_path)
```

```python

import os
import h5py
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

manifest_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "results/patient_level_split_manifest.csv"
)
output_root = "/content/brain_tumor_patient_images"

manifest = pd.read_csv(manifest_path)

class_names = ["glioma", "meningioma", "pituitary"]
splits = ["train", "val", "test"]

# Create the required folder structure
for split in splits:
    for class_name in class_names:
        os.makedirs(
            os.path.join(output_root, split, class_name),
            exist_ok=True
        )

print("Converting MATLAB MRI images...")

for row in tqdm(
    manifest.itertuples(index=False),
    total=len(manifest)
):
    with h5py.File(row.mat_path, "r") as f:
        image = np.asarray(f["cjdata"]["image"][()]).squeeze()

    # Convert MRI intensity values to 8-bit grayscale
    image = image.astype(np.float32)
    min_val = np.min(image)
    max_val = np.max(image)

    if max_val > min_val:
        image = (image - min_val) / (max_val - min_val) * 255.0
    else:
        image = np.zeros_like(image, dtype=np.float32)

    image = np.clip(image, 0, 255).astype(np.uint8)

    output_path = os.path.join(
        output_root,
        row.split,
        row.label_name,
        os.path.splitext(row.file)[0] + ".jpg"
    )

    Image.fromarray(image).save(
        output_path, format="JPEG", quality=95
    )

print("\nConversion complete!")

# Verify actual image counts
for split in splits:
    total = 0
    print(f"\n{split.upper()}")
    for class_name in class_names:
        folder = os.path.join(output_root, split, class_name)
        count = sum(
            1 for name in os.listdir(folder)
            if name.lower().endswith((".jpg", ".jpeg", ".png"))
        )
        total += count
        print(f"  {class_name}: {count}")
    print("  Total:", total)

print("\nPatient-level image folders saved at:")
print(output_root)
```

```python

import os
import tensorflow as tf
import pandas as pd
from tensorflow import keras
from tensorflow.keras import layers

# Paths
data_dir = "/content/brain_tumor_patient_images"
results_dir = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/results"
)
os.makedirs(results_dir, exist_ok=True)

# Reproducibility
SEED = 42
tf.keras.utils.set_random_seed(SEED)

image_size = (224, 224)
batch_size = 32
class_names = ["glioma", "meningioma", "pituitary"]

# Load each split with the same class order
train_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(data_dir, "train"),
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(data_dir, "val"),
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(data_dir, "test"),
    class_names=class_names,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    shuffle=False
)

# Normalize pixel values to [0, 1]
def normalize_images(images, labels):
    images = tf.cast(images, tf.float32) / 255.0
    labels = tf.one_hot(labels, depth=3)
    return images, labels

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.map(
    normalize_images, num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

val_ds = val_ds.map(
    normalize_images, num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

test_ds = test_ds.map(
    normalize_images, num_parallel_calls=AUTOTUNE
).prefetch(AUTOTUNE)

# Build a separate three-class CNN from scratch
model = keras.Sequential([
    layers.Input(shape=(224, 224, 3)),

    layers.Conv2D(32, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(),

    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(3, activation="softmax")
])

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Save the best model based on validation loss
model_path = os.path.join(
    results_dir, "patient_level_cnn_best.keras"
)

callbacks = [
    keras.callbacks.ModelCheckpoint(
        model_path,
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.CSVLogger(
        os.path.join(
            results_dir, "patient_level_training_history.csv"
        )
    )
]

# Train: up to 15 epochs, stopping early if validation stalls
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=callbacks
)

# Load the best validation checkpoint
best_model = keras.models.load_model(
    model_path, compile=False
)

print("\nTraining finished!")
print("Best model saved to:", model_path)
print("Class order:", class_names)
print("Best validation accuracy:",
      f"{max(history.history['val_accuracy']):.4f}")
```

```python

import h5py
import numpy as np
import os

sample_path = os.path.join(mat_dir, "119.mat")

with h5py.File(sample_path, "r") as f:
    mask = np.array(f["cjdata"]["tumorMask"])
    border = np.array(f["cjdata"]["tumorBorder"]).squeeze()

# Get tumour mask pixel coordinates
original_pixels = np.argwhere(mask > 0)
transposed_pixels = np.argwhere(mask.T > 0)

# Calculate centroids in (row, column) order
original_centroid = original_pixels.mean(axis=0)
transposed_centroid = transposed_pixels.mean(axis=0)

# Interpret border points as (x, y)
border_points = border.reshape(-1, 2)
border_centroid = border_points.mean(axis=0)

# Convert mask centroids into (x, y) = (column, row)
original_xy = original_centroid[::-1]
transposed_xy = transposed_centroid[::-1]

original_distance = np.linalg.norm(
    original_xy - border_centroid
)

transposed_distance = np.linalg.norm(
    transposed_xy - border_centroid
)

print("===== NUMERICAL MASK ALIGNMENT =====")
print("Border centroid (x, y):", border_centroid.round(2))

print("\nOriginal mask centroid (x, y):",
      original_xy.round(2))
print("Transposed mask centroid (x, y):",
      transposed_xy.round(2))

print("\nOriginal mask distance from border centroid:",
      round(original_distance, 3))

print("Transposed mask distance from border centroid:",
      round(transposed_distance, 3))
```

```python

import h5py
import numpy as np
import os
import pandas as pd

alignment_results = []

for filename in sorted(required_files):

    path = os.path.join(mat_dir, filename)

    with h5py.File(path, "r") as f:

        mask = np.array(f["cjdata"]["tumorMask"])
        border = np.array(
            f["cjdata"]["tumorBorder"]
        ).squeeze()

    if border.size < 4 or border.size % 2 != 0:
        continue

    border_points = border.reshape(-1, 2)
    border_centroid = border_points.mean(axis=0)

    original_pixels = np.argwhere(mask > 0)
    transposed_pixels = np.argwhere(mask.T > 0)

    if len(original_pixels) == 0:
        continue

    original_xy = original_pixels.mean(axis=0)[::-1]
    transposed_xy = transposed_pixels.mean(axis=0)[::-1]

    original_distance = np.linalg.norm(
        original_xy - border_centroid
    )

    transposed_distance = np.linalg.norm(
        transposed_xy - border_centroid
    )

    alignment_results.append({
        "filename": filename,
        "original_distance": original_distance,
        "transposed_distance": transposed_distance,
        "preferred_orientation": (
            "transposed"
            if transposed_distance < original_distance
            else "original"
        )
    })

alignment_df = pd.DataFrame(alignment_results)

print("===== 91-FILE MASK ORIENTATION AUDIT =====")
print("Files evaluated:", len(alignment_df))

print("\nPreferred orientation counts:")
print(alignment_df["preferred_orientation"].value_counts())

print("\nMean centroid distances:")
print(
    alignment_df[
        ["original_distance", "transposed_distance"]
    ].mean().round(3)
)

print("\nFirst 10 results:")
display(alignment_df.head(10).round(3))
```

```python

from scipy.ndimage import distance_transform_edt
import numpy as np
import pandas as pd
import h5py
import os

border_alignment = []

for filename in sorted(required_files):

    path = os.path.join(mat_dir, filename)

    with h5py.File(path, "r") as f:
        mask = np.array(f["cjdata"]["tumorMask"])
        border = np.array(
            f["cjdata"]["tumorBorder"]
        ).squeeze()

    if border.size < 4 or border.size % 2 != 0:
        continue

    points = border.reshape(-1, 2)

    # Test both orientations.
    # Border coordinates are interpreted as (x, y).
    for orientation, current_mask in [
        ("original", mask),
        ("transposed", mask.T)
    ]:

        # Distance from every pixel to nearest foreground pixel
        distance_map = distance_transform_edt(
            current_mask == 0
        )

        x = np.rint(points[:, 0]).astype(int)
        y = np.rint(points[:, 1]).astype(int)

        # Test MATLAB 1-based and 0-based coordinates
        for offset in [0, 1]:

            rows = y - offset
            cols = x - offset

            valid = (
                (rows >= 0) & (rows < current_mask.shape[0]) &
                (cols >= 0) & (cols < current_mask.shape[1])
            )

            distances = distance_map[rows[valid], cols[valid]]

            border_alignment.append({
                "filename": filename,
                "orientation": orientation,
                "coordinate_offset": offset,
                "mean_border_distance": distances.mean(),
                "median_border_distance": np.median(distances),
                "points_within_3px_pct": (
                    np.mean(distances <= 3) * 100
                )
            })

border_alignment_df = pd.DataFrame(border_alignment)

print("===== BORDER-TO-MASK ALIGNMENT AUDIT =====")

summary = (
    border_alignment_df.groupby(
        ["orientation", "coordinate_offset"]
    )
    .agg(
        Files=("filename", "nunique"),
        Mean_Distance=("mean_border_distance", "mean"),
        Median_Distance=("median_border_distance", "mean"),
        Within_3px_Percentage=("points_within_3px_pct", "mean")
    )
    .reset_index()
)

display(summary.round(3))
```

```python

from PIL import Image
import numpy as np
import h5py
import os

image_output_dir = (
    "/content/brain_tumor_patient_images/val/meningioma"
)

os.makedirs(image_output_dir, exist_ok=True)

restored_images = []
failed_images = []

print("===== RECONSTRUCTING VALIDATION IMAGES =====")

for filename in sorted(required_files):

    mat_path = os.path.join(mat_dir, filename)

    try:
        with h5py.File(mat_path, "r") as f:
            image = np.array(
                f["cjdata"]["image"]
            ).astype(np.float32)

        # Original preprocessing
        image_min = image.min()
        image_max = image.max()

        if image_max > image_min:
            image = (
                (image - image_min) /
                (image_max - image_min) * 255
            )
        else:
            image = np.zeros_like(image)

        image = image.astype(np.uint8)

        # Original RGB conversion and JPEG settings
        pil_image = Image.fromarray(image).convert("RGB")

        output_filename = (
            os.path.splitext(filename)[0] + ".jpg"
        )

        output_path = os.path.join(
            image_output_dir,
            output_filename
        )

        pil_image.save(
            output_path,
            format="JPEG",
            quality=95
        )

        restored_images.append(output_filename)

    except Exception as e:
        failed_images.append((filename, str(e)))

print("\n===== RECONSTRUCTION SUMMARY =====")
print("Required images:", len(required_files))
print("Successfully reconstructed:", len(restored_images))
print("Failed:", len(failed_images))
print("Output directory:", image_output_dir)

if failed_images:
    print("\nFailures:", failed_images[:5])
```

```python

from tensorflow.keras import Model
from tensorflow.keras.layers import Dense
import numpy as np
import tensorflow as tf

print("===== REBUILDING LOGIT MODEL =====")

# Extract features before the final Dense classification layer
feature_model = Model(
    inputs=model.inputs,
    outputs=model.layers[-1].input
)

# Create a linear layer to expose logits
logits_layer = Dense(
    3,
    activation=None,
    name="logits"
)

logits_output = logits_layer(feature_model.output)

logit_model = Model(
    inputs=feature_model.inputs,
    outputs=logits_output
)

# Copy trained classifier weights
logit_model.layers[-1].set_weights(
    model.layers[-1].get_weights()
)

# Verify softmax(logits) against original model probabilities
sample_logits = logit_model(
    X_val_meningioma[:5],
    training=False
).numpy()

sample_probabilities = tf.nn.softmax(
    sample_logits,
    axis=1
).numpy()

original_probabilities = model.predict(
    X_val_meningioma[:5],
    verbose=0
)

max_difference = np.max(
    np.abs(
        sample_probabilities -
        original_probabilities
    )
)

print("Maximum probability difference:", max_difference)

# Build Grad-CAM model
grad_model = Model(
    inputs=logit_model.inputs,
    outputs=[
        logit_model.get_layer("conv2d_11").output,
        logit_model.output
    ]
)

print("\n===== GRAD-CAM MODEL READY =====")
print("Target layer: conv2d_11")
print("Feature maps:", grad_model.output[0].shape)
print("Logits:", grad_model.output[1].shape)
```

```python

print("===== GRADIENT CONNECTION DIAGNOSTIC =====")

sample_input = tf.convert_to_tensor(
    X_val_meningioma[:1],
    dtype=tf.float32
)

with tf.GradientTape() as tape:

    # Call the logit model directly
    feature_maps = logit_model.get_layer(
        "conv2d_11"
    ).output

    # Rebuild a model that returns both outputs
    diagnostic_model = tf.keras.Model(
        inputs=logit_model.inputs,
        outputs=[
            logit_model.get_layer("conv2d_11").output,
            logit_model.output
        ]
    )

    conv_output, logits = diagnostic_model(
        sample_input,
        training=False
    )

    tape.watch(conv_output)

    meningioma_score = logits[:, 1]

gradients = tape.gradient(
    meningioma_score,
    conv_output
)

print("Feature map shape:", conv_output.shape)
print("Logit shape:", logits.shape)
print("Gradient is None:", gradients is None)

if gradients is not None:
    print("Gradient shape:", gradients.shape)
    print(
        "Gradient norm:",
        float(tf.norm(gradients).numpy())
    )
```

```python

print("===== CORRECTED GRADIENT CONNECTION TEST =====")

# Extract convolutional feature maps
feature_extractor = tf.keras.Model(
    inputs=model.inputs,
    outputs=model.get_layer("conv2d_11").output
)

sample_input = tf.convert_to_tensor(
    X_val_meningioma[:1],
    dtype=tf.float32
)

with tf.GradientTape() as tape:

    # Calculate feature maps
    conv_outputs = feature_extractor(
        sample_input,
        training=False
    )

    # Start tracking the intermediate feature maps
    tape.watch(conv_outputs)

    # Explicitly pass through the remaining trained layers
    x = model.get_layer("max_pooling2d_11")(conv_outputs)
    x = model.get_layer("global_average_pooling2d_2")(x)
    x = model.get_layer("dense_6")(x)
    x = model.get_layer("dropout_3")(x, training=False)

    logits = model.get_layer("dense_7")(x)

    meningioma_score = logits[:, 1]

gradients = tape.gradient(
    meningioma_score,
    conv_outputs
)

print("Feature maps:", conv_outputs.shape)
print("Logits:", logits.shape)
print("Gradient is None:", gradients is None)

if gradients is not None:
    print("Gradient shape:", gradients.shape)
    print("Gradient norm:", float(tf.norm(gradients).numpy()))

    # Verify manually reconstructed logits against the logit model
    reference_logits = logit_model(
        sample_input,
        training=False
    )

    print(
        "Maximum logit difference:",
        float(tf.reduce_max(
            tf.abs(logits - reference_logits)
        ).numpy())
    )
```

```python

print("===== LAYER-BY-LAYER OUTPUT AUDIT =====")

sample_input = tf.convert_to_tensor(
    X_val_meningioma[:1],
    dtype=tf.float32
)

# Original CNN intermediate outputs
layer_names = [
    "conv2d_11",
    "max_pooling2d_11",
    "global_average_pooling2d_2",
    "dense_6",
    "dropout_3",
    "dense_7"
]

original_intermediate = tf.keras.Model(
    inputs=model.inputs,
    outputs=[
        model.get_layer(name).output
        for name in layer_names
    ]
)

original_outputs = original_intermediate(
    sample_input,
    training=False
)

# Manual forward pass
manual_outputs = []

x = feature_extractor(
    sample_input,
    training=False
)
manual_outputs.append(x)

x = model.get_layer("max_pooling2d_11")(x)
manual_outputs.append(x)

x = model.get_layer("global_average_pooling2d_2")(x)
manual_outputs.append(x)

x = model.get_layer("dense_6")(x)
manual_outputs.append(x)

x = model.get_layer("dropout_3")(x, training=False)
manual_outputs.append(x)

x = model.get_layer("dense_7")(x)
manual_outputs.append(x)

for name, original, manual in zip(
    layer_names,
    original_outputs,
    manual_outputs
):
    difference = tf.reduce_max(
        tf.abs(original - manual)
    ).numpy()

    print(name, "maximum difference:", float(difference))
```

```python

print("===== FINAL LOGIT CONSISTENCY CHECK =====")

sample_input = tf.convert_to_tensor(
    X_val_meningioma[:1],
    dtype=tf.float32
)

# Original model probabilities
original_probs = model(
    sample_input,
    training=False
).numpy()

# Reference logits
reference_logits = logit_model(
    sample_input,
    training=False
).numpy()

reference_probs = tf.nn.softmax(
    reference_logits,
    axis=1
).numpy()

# Manually reconstructed logits
conv_outputs = feature_extractor(
    sample_input,
    training=False
)

x = model.get_layer("max_pooling2d_11")(conv_outputs)
x = model.get_layer("global_average_pooling2d_2")(x)
x = model.get_layer("dense_6")(x)
x = model.get_layer("dropout_3")(x, training=False)
manual_logits = model.get_layer("dense_7")(x).numpy()

print(
    "Reference vs manual logit difference:",
    np.max(np.abs(reference_logits - manual_logits))
)

print(
    "Original vs reference probability difference:",
    np.max(np.abs(original_probs - reference_probs))
)

print("Original probabilities:", original_probs)
print("Reference probabilities:", reference_probs)
```

```python

print("===== RAW LOGIT AND GRADIENT VERIFICATION =====")

sample_input = tf.convert_to_tensor(
    X_val_meningioma[:1],
    dtype=tf.float32
)

# Get the original feature maps
with tf.GradientTape() as tape:

    conv_outputs = feature_extractor(
        sample_input,
        training=False
    )

    tape.watch(conv_outputs)

    # Reproduce the trained CNN's downstream layers
    x = model.get_layer("max_pooling2d_11")(conv_outputs)
    x = model.get_layer("global_average_pooling2d_2")(x)
    x = model.get_layer("dense_6")(x)
    x = model.get_layer("dropout_3")(x, training=False)

    # Bypass the final Dense softmax activation
    final_layer = model.get_layer("dense_7")
    kernel, bias = final_layer.get_weights()

    manual_logits = tf.linalg.matmul(
        x,
        tf.convert_to_tensor(kernel)
    ) + tf.convert_to_tensor(bias)

    meningioma_score = manual_logits[:, 1]

gradients = tape.gradient(
    meningioma_score,
    conv_outputs
)

# Compare with our independent linear logit model
reference_logits = logit_model(
    sample_input,
    training=False
)

print(
    "Raw logit difference:",
    float(tf.reduce_max(
        tf.abs(manual_logits - reference_logits)
    ).numpy())
)

print("Gradient is None:", gradients is None)

if gradients is not None:
    print("Gradient shape:", gradients.shape)
    print("Gradient norm:", float(tf.norm(gradients).numpy()))

print(
    "Softmax probability difference:",
    float(tf.reduce_max(
        tf.abs(
            tf.nn.softmax(manual_logits, axis=1) -
            model(sample_input, training=False)
        )
    ).numpy())
)
```

```python

import os
import numpy as np
import pandas as pd
import h5py
import tensorflow as tf

print("===== CORRECTED GRAD-CAM: 91 VALIDATION SLICES =====")

corrected_cam_records = []

# Preserve the exact order of reconstructed images
filename_to_index = {
    filename: index
    for index, filename in enumerate(filenames)
}

for i, row in meningioma_val.iterrows():

    filename = str(row["file"])
    jpg_filename = os.path.splitext(filename)[0] + ".jpg"
    image_index = filename_to_index[jpg_filename]

    sample_input = tf.convert_to_tensor(
        X_val_meningioma[image_index:image_index + 1],
        dtype=tf.float32
    )

    # Calculate gradients through the verified forward pass
    with tf.GradientTape(persistent=True) as tape:

        conv_outputs = feature_extractor(
            sample_input,
            training=False
        )

        tape.watch(conv_outputs)

        x = model.get_layer("max_pooling2d_11")(conv_outputs)
        x = model.get_layer("global_average_pooling2d_2")(x)
        x = model.get_layer("dense_6")(x)
        x = model.get_layer("dropout_3")(x, training=False)

        kernel, bias = model.get_layer("dense_7").get_weights()

        logits = tf.linalg.matmul(
            x,
            tf.convert_to_tensor(kernel)
        ) + tf.convert_to_tensor(bias)

    # Load verified transposed tumour mask
    mat_path = os.path.join(mat_dir, filename)

    with h5py.File(mat_path, "r") as f:
        original_mask = np.array(f["cjdata"]["tumorMask"])

    tumour_mask = tf.image.resize(
        original_mask.T.astype(np.float32)[..., None],
        [224, 224],
        method="nearest"
    ).numpy().squeeze() > 0.5

    for class_index, class_name in enumerate(class_names):

        with tf.GradientTape() as class_tape:
            # Recompute the class score while recording its path
            class_tape.watch(conv_outputs)

            x = model.get_layer("max_pooling2d_11")(conv_outputs)
            x = model.get_layer("global_average_pooling2d_2")(x)
            x = model.get_layer("dense_6")(x)
            x = model.get_layer("dropout_3")(x, training=False)

            logits_for_class = (
                tf.linalg.matmul(
                    x,
                    tf.convert_to_tensor(kernel)
                ) + tf.convert_to_tensor(bias)
            )

            score = logits_for_class[:, class_index]

        gradients = class_tape.gradient(
            score,
            conv_outputs
        )

        if gradients is None:
            raise RuntimeError(
                f"Missing gradient for {filename}, {class_name}"
            )

        weights = tf.reduce_mean(
            gradients,
            axis=(1, 2)
        )

        raw_cam = tf.reduce_sum(
            conv_outputs[0] * weights[0][None, None, :],
            axis=-1
        )

        cam = tf.maximum(raw_cam, 0)

        cam = tf.image.resize(
            cam[..., None],
            [224, 224]
        ).numpy().squeeze()

        if cam.max() > 0:
            cam = cam / cam.max()

        inside = cam[tumour_mask]
        outside = cam[~tumour_mask]

        positive_energy = (
            inside.sum() / cam.sum()
            if cam.sum() > 0 else np.nan
        )

        threshold = np.percentile(cam, 90)
        top_pixels = cam >= threshold

        top_overlap = (
            np.logical_and(top_pixels, tumour_mask).sum()
            / max(top_pixels.sum(), 1)
        )

        corrected_cam_records.append({
            "file": filename,
            "pid": row["pid"],
            "class": class_name,
            "true_class": "meningioma",
            "mean_cam_inside": float(inside.mean()),
            "mean_cam_outside": float(outside.mean()),
            "positive_energy_inside": float(positive_energy),
            "top10_tumor_overlap": float(top_overlap),
            "tumor_area_fraction": float(tumour_mask.mean())
        })

    del tape

    if (i + 1) % 10 == 0:
        print(f"Processed {i + 1}/91 slices")

corrected_cam_df = pd.DataFrame(corrected_cam_records)

corrected_path = os.path.join(
    results_dir,
    "meningioma_logit_gradcam_91_slices_transposed_mask.csv"
)

corrected_cam_df.to_csv(corrected_path, index=False)

print("\n===== CORRECTED ANALYSIS COMPLETE =====")
print("Total records:", len(corrected_cam_df))
print("Unique slices:", corrected_cam_df["file"].nunique())
print("Saved:", corrected_path)

print("\n===== CLASS-WISE LOCALISATION SUMMARY =====")

print(
    corrected_cam_df.groupby("class")[
        [
            "mean_cam_inside",
            "mean_cam_outside",
            "positive_energy_inside",
            "top10_tumor_overlap"
        ]
    ].mean().round(5)
)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
import h5py
import os

print("===== RAW GRAD-CAM VISUAL VERIFICATION =====")

kernel, bias = model.get_layer("dense_7").get_weights()

example_pids = ["101082", "104695"]

fig, axes = plt.subplots(2, 3, figsize=(15, 9))

for row_idx, pid in enumerate(example_pids):

    patient_rows = meningioma_val[
        meningioma_val["pid"].astype(str) == pid
    ].sort_values("file")

    if patient_rows.empty:
        raise ValueError(f"No validation images found for PID {pid}")

    selected = patient_rows.iloc[0]
    filename = str(selected["file"])

    jpg_filename = os.path.splitext(filename)[0] + ".jpg"
    image_index = filename_to_index[jpg_filename]

    image_tensor = tf.convert_to_tensor(
        X_val_meningioma[image_index:image_index + 1],
        dtype=tf.float32
    )

    with h5py.File(os.path.join(mat_dir, filename), "r") as f:
        original_mask = np.array(f["cjdata"]["tumorMask"])

    mask = tf.image.resize(
        original_mask.T.astype(np.float32)[..., None],
        [224, 224],
        method="nearest"
    ).numpy().squeeze() > 0.5

    with tf.GradientTape() as tape:

        conv_outputs = feature_extractor(
            image_tensor,
            training=False
        )

        tape.watch(conv_outputs)

        x = model.get_layer("max_pooling2d_11")(conv_outputs)
        x = model.get_layer("global_average_pooling2d_2")(x)
        x = model.get_layer("dense_6")(x)
        x = model.get_layer("dropout_3")(x, training=False)

        logits = (
            tf.linalg.matmul(x, tf.convert_to_tensor(kernel))
            + tf.convert_to_tensor(bias)
        )

        score = logits[:, 1]

    gradients = tape.gradient(score, conv_outputs)
    weights = tf.reduce_mean(gradients, axis=(1, 2))

    raw_cam = tf.reduce_sum(
        conv_outputs[0] * weights[0][None, None, :],
        axis=-1
    )

    raw_cam = tf.image.resize(
        raw_cam[..., None],
        [224, 224]
    ).numpy().squeeze()

    positive_cam = np.maximum(raw_cam, 0)

    if positive_cam.max() > 0:
        positive_cam /= positive_cam.max()

    print(f"\nPID: {pid} | File: {filename}")
    print("Raw CAM minimum:", raw_cam.min())
    print("Raw CAM maximum:", raw_cam.max())
    print("Positive CAM pixel fraction:", np.mean(raw_cam > 0))
    print("Tumour mask pixel fraction:", mask.mean())
    print("Raw CAM mean inside tumour:", raw_cam[mask].mean())

    axes[row_idx, 0].imshow(
        image_tensor.numpy()[0, :, :, 0], cmap="gray"
    )
    axes[row_idx, 0].contour(mask, levels=[0.5], colors="lime")
    axes[row_idx, 0].set_title(f"PID {pid}: MRI + mask")

    axes[row_idx, 1].imshow(raw_cam, cmap="coolwarm")
    axes[row_idx, 1].contour(mask, levels=[0.5], colors="lime")
    axes[row_idx, 1].set_title("Raw signed CAM")

    axes[row_idx, 2].imshow(
        image_tensor.numpy()[0, :, :, 0], cmap="gray"
    )
    axes[row_idx, 2].imshow(
        positive_cam, cmap="jet", alpha=0.45
    )
    axes[row_idx, 2].contour(mask, levels=[0.5], colors="lime")
    axes[row_idx, 2].set_title("Positive Grad-CAM")

    for ax in axes[row_idx]:
        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.spatial.distance import cdist

# Use the already loaded image, mask and border
mask_original = mask.astype(bool)
mask_transposed = mask_original.T

points = border.reshape(-1, 2)

# Interpret border as (row, column)
points_row_col = np.rint(points).astype(int)

# Interpret border as (x, y), converting to (row, column)
points_xy = np.rint(points[:, ::-1]).astype(int)

def border_distance(mask_array, points_rc):
    points_rc = points_rc[
        (points_rc[:, 0] >= 0) &
        (points_rc[:, 0] < mask_array.shape[0]) &
        (points_rc[:, 1] >= 0) &
        (points_rc[:, 1] < mask_array.shape[1])
    ]

    mask_coords = np.argwhere(mask_array)

    distances = cdist(points_rc, mask_coords)
    nearest = distances.min(axis=1)

    return nearest.mean(), np.median(nearest), np.mean(nearest <= 3)

results = {
    "Original mask + row/column border": border_distance(
        mask_original, points_row_col
    ),
    "Original mask + x/y border": border_distance(
        mask_original, points_xy
    ),
    "Transposed mask + row/column border": border_distance(
        mask_transposed, points_row_col
    ),
    "Transposed mask + x/y border": border_distance(
        mask_transposed, points_xy
    )
}

print("===== PID 104695: BORDER COORDINATE AUDIT =====")

for name, (mean_d, median_d, within_3) in results.items():
    print(f"\n{name}")
    print(f"Mean distance: {mean_d:.3f} pixels")
    print(f"Median distance: {median_d:.3f} pixels")
    print(f"Points within 3 pixels: {within_3 * 100:.2f}%")
```

```python
import h5py
import numpy as np
import pandas as pd
from scipy.ndimage import distance_transform_edt

audit_records = []

def evaluate_alignment(mask_array, border_points, swap_xy, offset):
    # Convert border coordinates to row/column order
    points = border_points[:, ::-1] if swap_xy else border_points.copy()

    # Correct for coordinate indexing convention
    points = np.rint(points - offset).astype(int)

    # Distance from every pixel to the nearest tumour-mask pixel
    distance_map = distance_transform_edt(~mask_array.astype(bool))

    valid = (
        (points[:, 0] >= 0) &
        (points[:, 0] < mask_array.shape[0]) &
        (points[:, 1] >= 0) &
        (points[:, 1] < mask_array.shape[1])
    )

    points = points[valid]

    if len(points) == 0:
        return np.nan, np.nan, np.nan

    distances = distance_map[points[:, 0], points[:, 1]]

    return (
        distances.mean(),
        np.median(distances),
        np.mean(distances <= 3) * 100
    )


for _, row in meningioma_val.iterrows():

    with h5py.File(row["mat_path"], "r") as f:
        mask = np.array(f["cjdata"]["tumorMask"]).astype(bool)
        border = np.array(f["cjdata"]["tumorBorder"]).squeeze()

    border_points = border.reshape(-1, 2)

    for orientation, current_mask in [
        ("original", mask),
        ("transposed", mask.T)
    ]:
        for coordinate_order, swap_xy in [
            ("row_column", False),
            ("x_y", True)
        ]:
            for offset in [0, 1]:

                mean_d, median_d, within_3 = evaluate_alignment(
                    current_mask,
                    border_points,
                    swap_xy,
                    offset
                )

                audit_records.append({
                    "pid": str(row["pid"]),
                    "file": row["file"],
                    "orientation": orientation,
                    "coordinate_order": coordinate_order,
                    "offset": offset,
                    "mean_distance": mean_d,
                    "median_distance": median_d,
                    "within_3_percent": within_3
                })


alignment_audit = pd.DataFrame(audit_records)

summary = (
    alignment_audit
    .groupby(["orientation", "coordinate_order", "offset"])
    .agg(
        files=("file", "nunique"),
        mean_distance=("mean_distance", "mean"),
        median_distance=("median_distance", "median"),
        mean_within_3_percent=("within_3_percent", "mean")
    )
    .reset_index()
    .sort_values("mean_distance")
)

print("===== FULL 91-FILE MASK ALIGNMENT AUDIT =====")
display(summary.round(3))

# Save as a separate file; preserve previous audits
output_path = (
    "/content/drive/MyDrive/results/"
    "meningioma_mask_alignment_coordinate_reaudit.csv"
)

alignment_audit.to_csv(output_path, index=False)

print("\nDetailed audit saved to:", output_path)
print("Files evaluated:", alignment_audit["file"].nunique())
```

```python
import os
import h5py
import numpy as np
import pandas as pd
import tensorflow as tf

print("===== ORIGINAL-MASK GRAD-CAM RECOMPUTATION =====")

# Final classifier weights
kernel, bias = model.get_layer("dense_7").get_weights()
kernel = tf.convert_to_tensor(kernel)
bias = tf.convert_to_tensor(bias)

class_names = ["glioma", "meningioma", "pituitary"]

# Map each MAT file to its saved path and patient ID
mat_lookup = meningioma_val.set_index("file")["mat_path"].to_dict()
pid_lookup = meningioma_val.set_index("file")["pid"].to_dict()

corrected_records = []

for image_number, jpg_filename in enumerate(filenames):

    jpg_filename = str(jpg_filename)
    mat_filename = os.path.splitext(jpg_filename)[0] + ".mat"

    if mat_filename not in mat_lookup:
        raise ValueError(f"Missing MAT file: {mat_filename}")

    # Retrieve the ORIGINAL tumour mask (no transpose)
    with h5py.File(mat_lookup[mat_filename], "r") as f:
        original_mask = np.array(
            f["cjdata"]["tumorMask"]
        ).astype(np.float32)

    # Resize mask to match the model input dimensions
    mask_224 = tf.image.resize(
        original_mask[..., None],
        [224, 224],
        method="nearest"
    ).numpy().squeeze() > 0.5

    # Retrieve the exact validation image tensor
    image_tensor = tf.convert_to_tensor(
        X_val_meningioma[image_number:image_number + 1],
        dtype=tf.float32
    )

    for class_index, class_name in enumerate(class_names):

        with tf.GradientTape() as tape:

            conv_outputs = feature_extractor(
                image_tensor, training=False
            )

            tape.watch(conv_outputs)

            x = model.get_layer("max_pooling2d_11")(conv_outputs)
            x = model.get_layer("global_average_pooling2d_2")(x)
            x = model.get_layer("dense_6")(x)
            x = model.get_layer("dropout_3")(x, training=False)

            logits = tf.linalg.matmul(x, kernel) + bias
            score = logits[:, class_index]

        gradients = tape.gradient(score, conv_outputs)

        if gradients is None:
            raise RuntimeError(
                f"Gradient is None for {jpg_filename}, {class_name}"
            )

        # Grad-CAM channel weights
        weights = tf.reduce_mean(
            gradients, axis=(1, 2)
        )

        # Signed raw Grad-CAM
        raw_cam = tf.reduce_sum(
            conv_outputs[0] * weights[0][None, None, :],
            axis=-1
        )

        raw_cam = tf.image.resize(
            raw_cam[None, ..., None],
            [224, 224],
            method="bilinear"
        ).numpy().squeeze()

        # Positive Grad-CAM (ReLU)
        positive_cam = np.maximum(raw_cam, 0)

        tumour_pixels = mask_224
        outside_pixels = ~tumour_pixels

        tumour_area = tumour_pixels.mean()

        cam_inside = (
            raw_cam[tumour_pixels].mean()
            if tumour_pixels.any() else np.nan
        )

        cam_outside = (
            raw_cam[outside_pixels].mean()
            if outside_pixels.any() else np.nan
        )

        # Fraction of total positive CAM energy inside tumour
        total_positive_energy = positive_cam.sum()

        if total_positive_energy > 0:
            positive_energy_inside = (
                positive_cam[tumour_pixels].sum()
                / total_positive_energy
            )

            # Top 10% of positive CAM pixels
            flat_cam = positive_cam.ravel()
            top_n = int(np.ceil(0.10 * flat_cam.size))
            top_indices = np.argsort(flat_cam)[-top_n:]

            top10_overlap = (
                tumour_pixels.ravel()[top_indices].mean()
            )
        else:
            positive_energy_inside = 0.0
            top10_overlap = 0.0

        corrected_records.append({
            "Patient_ID": int(pid_lookup[mat_filename]),
            "Image": jpg_filename,
            "Class": class_name,
            "Logit": float(logits.numpy()[0, class_index]),
            "Tumour_Area": float(tumour_area),
            "CAM_Mean_Inside": float(cam_inside),
            "CAM_Mean_Outside": float(cam_outside),
            "Positive_CAM_Energy_Inside": float(
                positive_energy_inside
            ),
            "Top10_Overlap": float(top10_overlap)
        })

    if (image_number + 1) % 10 == 0:
        print(
            f"Processed {image_number + 1}/"
            f"{len(filenames)} images"
        )

# Create corrected dataframe
corrected_cam_df = pd.DataFrame(corrected_records)

# Save separately; do not overwrite previous results
corrected_path = os.path.join(
    results_dir,
    "meningioma_logit_gradcam_91_slices_original_mask.csv"
)

corrected_cam_df.to_csv(corrected_path, index=False)

print("\n===== RECOMPUTATION COMPLETE =====")
print("Records:", len(corrected_cam_df))
print("Unique images:", corrected_cam_df["Image"].nunique())
print("Unique patients:", corrected_cam_df["Patient_ID"].nunique())
print("Classes:", corrected_cam_df["Class"].unique())
print("Saved:", corrected_path)

print("\n===== CLASS-WISE SUMMARY =====")

display(
    corrected_cam_df.groupby("Class")[
        [
            "CAM_Mean_Inside",
            "CAM_Mean_Outside",
            "Positive_CAM_Energy_Inside",
            "Top10_Overlap"
        ]
    ].mean().round(6)
)
```

```python
print("===== SLICE vs PATIENT PREDICTION AUDIT =====")

class_names = ["glioma", "meningioma", "pituitary"]

prob_cols = [
    "glioma_probability",
    "meningioma_probability",
    "pituitary_probability"
]

# Keep only slices belonging to validation meningioma patients
meningioma_pids = meningioma_errors["pid"].astype(str).tolist()

meningioma_slices = slice_df[
    slice_df["pid"].astype(str).isin(meningioma_pids)
].copy()

# Get each slice's predicted class
meningioma_slices["slice_predicted_class"] = np.array(class_names)[
    np.argmax(meningioma_slices[prob_cols].values, axis=1)
]

# Summarise slice predictions for every patient
slice_summary = (
    meningioma_slices
    .groupby("pid")
    .agg(
        total_slices=("filename", "count"),
        glioma_slices=("slice_predicted_class",
                       lambda x: (x == "glioma").sum()),
        meningioma_slices=("slice_predicted_class",
                           lambda x: (x == "meningioma").sum()),
        pituitary_slices=("slice_predicted_class",
                          lambda x: (x == "pituitary").sum())
    )
    .reset_index()
)

# Add patient-level predictions
comparison = meningioma_errors[
    ["pid", "predicted_class", "correct_prediction", "slice_count"]
].merge(
    slice_summary,
    on="pid",
    how="left"
)

comparison["slice_prediction_agreement"] = (
    comparison[
        ["glioma_slices", "meningioma_slices", "pituitary_slices"]
    ].max(axis=1) / comparison["total_slices"]
)

display(comparison.sort_values("pid").round(3))

print("\nTotal meningioma validation slices:",
      len(meningioma_slices))

print("Patients represented:",
      comparison["pid"].nunique())
```

```python
print("===== PATIENT vs MEAN SLICE PROBABILITY AUDIT =====")

# Average class probabilities across slices for each patient
mean_slice_probs = (
    meningioma_slices
    .groupby("pid")[prob_cols]
    .mean()
    .reset_index()
)

# Compare with saved patient-level probabilities
aggregation_check = meningioma_errors[
    [
        "pid",
        "predicted_class",
        "glioma_probability",
        "meningioma_probability",
        "pituitary_probability"
    ]
].merge(
    mean_slice_probs,
    on="pid",
    suffixes=("_patient", "_slice_mean"),
    validate="one_to_one"
)

for cls in class_names:
    aggregation_check[f"{cls}_difference"] = (
        aggregation_check[f"{cls}_probability_patient"]
        - aggregation_check[f"{cls}_probability_slice_mean"]
    )

display(
    aggregation_check.round(6).sort_values("pid")
)

print("\n===== AGGREGATION DIFFERENCE SUMMARY =====")

difference_cols = [f"{cls}_difference" for cls in class_names]

display(
    aggregation_check[difference_cols]
    .abs()
    .agg(["mean", "max"])
    .round(8)
)
```

```python
print("===== MENINGIOMA SLICE PROBABILITY STABILITY =====")

stability = meningioma_slices.copy()

# Rank each class probability within every slice
stability["meningioma_rank"] = (
    stability[prob_cols]
    .rank(axis=1, ascending=False, method="min")[
        "meningioma_probability"
    ]
)

# Calculate patient-level slice statistics
stability_summary = (
    stability.groupby("pid")
    .agg(
        slice_count=("filename", "count"),

        meningioma_mean=("meningioma_probability", "mean"),
        meningioma_std=("meningioma_probability", "std"),
        meningioma_min=("meningioma_probability", "min"),
        meningioma_max=("meningioma_probability", "max"),

        glioma_mean=("glioma_probability", "mean"),
        glioma_std=("glioma_probability", "std"),

        pituitary_mean=("pituitary_probability", "mean"),
        pituitary_std=("pituitary_probability", "std"),

        meningioma_rank1=("meningioma_rank",
                          lambda x: (x == 1).sum()),
        meningioma_rank2=("meningioma_rank",
                          lambda x: (x == 2).sum()),
        meningioma_rank3=("meningioma_rank",
                          lambda x: (x == 3).sum())
    )
    .reset_index()
)

stability_summary = stability_summary.merge(
    meningioma_errors[
        ["pid", "predicted_class", "correct_prediction"]
    ],
    on="pid",
    validate="one_to_one"
)

display(
    stability_summary.sort_values("pid").round(4)
)

print("\n===== OVERALL SLICE RANK COUNTS =====")

print(
    stability["meningioma_rank"]
    .value_counts()
    .sort_index()
    .rename(index={
        1: "Meningioma ranked 1st",
        2: "Meningioma ranked 2nd",
        3: "Meningioma ranked 3rd"
    })
)
```

```python
print("===== SLICE-LEVEL MENINGIOMA COMPETITION AUDIT =====")

gap_df = meningioma_slices.copy()

other_cols = ["glioma_probability", "pituitary_probability"]

gap_df["strongest_competitor"] = np.where(
    gap_df["glioma_probability"] >= gap_df["pituitary_probability"],
    "glioma",
    "pituitary"
)

gap_df["competitor_probability"] = gap_df[
    other_cols
].max(axis=1)

# Positive = meningioma has higher probability
# Negative = another class has higher probability
gap_df["meningioma_gap"] = (
    gap_df["meningioma_probability"]
    - gap_df["competitor_probability"]
)

gap_df["meningioma_wins"] = gap_df["meningioma_gap"] > 0

print("\n===== PATIENT-LEVEL GAP SUMMARY =====")

gap_summary = (
    gap_df.groupby("pid")
    .agg(
        slice_count=("filename", "count"),
        mean_gap=("meningioma_gap", "mean"),
        min_gap=("meningioma_gap", "min"),
        max_gap=("meningioma_gap", "max"),
        gap_std=("meningioma_gap", "std"),
        slices_meningioma_wins=("meningioma_wins", "sum"),
        glioma_competitor_slices=(
            "strongest_competitor",
            lambda x: (x == "glioma").sum()
        ),
        pituitary_competitor_slices=(
            "strongest_competitor",
            lambda x: (x == "pituitary").sum()
        )
    )
    .reset_index()
    .merge(
        meningioma_errors[
            ["pid", "predicted_class", "correct_prediction"]
        ],
        on="pid",
        validate="one_to_one"
    )
)

display(gap_summary.sort_values("pid").round(4))

print("\n===== OVERALL SLICE GAP =====")
print("Slices where meningioma wins:",
      gap_df["meningioma_wins"].sum())

print("Slices where another class wins:",
      (~gap_df["meningioma_wins"]).sum())

print("\nMean meningioma gap:",
      round(gap_df["meningioma_gap"].mean(), 5))
```

```python
import os
import io
import zipfile
import shutil

archive_path = "/content/drive/MyDrive/Datasets_MP - 1.zip"
mat_dir = "/content/brain_tumor_patient_data/mat_files"

os.makedirs(mat_dir, exist_ok=True)

expected_files = set(manifest["file"].astype(str))
restored = []
verified_existing = []
mismatched_existing = []

print("===== SAFE MAT FILE RESTORATION =====")

with zipfile.ZipFile(archive_path, "r") as outer:

    inner_names = [
        name for name in outer.namelist()
        if os.path.basename(name).startswith("brainTumorDataPublic_")
        and name.endswith(".zip")
    ]

    # Calculate required disk space before extraction
    required_bytes = 0

    for inner_name in inner_names:
        with outer.open(inner_name) as stream:
            with zipfile.ZipFile(stream, "r") as inner:
                for info in inner.infolist():
                    filename = info.filename.replace("\\", "/").split("/")[-1]

                    if filename in expected_files:
                        output_path = os.path.join(mat_dir, filename)

                        if not os.path.exists(output_path):
                            required_bytes += info.file_size

    free_bytes = shutil.disk_usage("/content").free

    print("Space required (GB):", round(required_bytes / 1e9, 2))
    print("Space available (GB):", round(free_bytes / 1e9, 2))

    if required_bytes > free_bytes:
        raise RuntimeError("Insufficient disk space. No files restored.")

    # Restore and verify
    for inner_name in inner_names:
        with outer.open(inner_name) as stream:
            with zipfile.ZipFile(stream, "r") as inner:

                for info in inner.infolist():
                    filename = info.filename.replace("\\", "/").split("/")[-1]

                    if filename not in expected_files:
                        continue

                    data = inner.read(info.filename)
                    output_path = os.path.join(mat_dir, filename)

                    if os.path.exists(output_path):
                        with open(output_path, "rb") as f:
                            existing_data = f.read()

                        if existing_data == data:
                            verified_existing.append(filename)
                        else:
                            mismatched_existing.append(filename)
                    else:
                        with open(output_path, "wb") as f:
                            f.write(data)

                        restored.append(filename)

disk_files = {
    f for f in os.listdir(mat_dir)
    if f.lower().endswith(".mat")
}

print("\n===== RESTORATION SUMMARY =====")
print("Newly restored:", len(restored))
print("Existing files verified:", len(verified_existing))
print("Existing files with content mismatch:", len(mismatched_existing))
print("Total MAT files now present:", len(disk_files))
print("Expected MAT files:", len(expected_files))
print("Missing files:", len(expected_files - disk_files))
print("Unexpected files:", len(disk_files - expected_files))

if mismatched_existing:
    print("Mismatched filenames:", mismatched_existing[:10])
```

```python
import os
import h5py
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

# Source and destination
mat_dir = "/content/brain_tumor_patient_data/mat_files"
output_root = "/content/brain_tumor_patient_images_reconstructed"

class_names = ["glioma", "meningioma", "pituitary"]
splits = ["train", "val", "test"]

# Do not overwrite an earlier reconstruction
if os.path.exists(output_root) and any(os.scandir(output_root)):
    raise FileExistsError(
        f"Destination already contains data: {output_root}"
    )

# Create exact split/class folder structure
for split in splits:
    for class_name in class_names:
        os.makedirs(
            os.path.join(output_root, split, class_name),
            exist_ok=True
        )

print("===== RECONSTRUCTING ORIGINAL IMAGE DATASET =====")
print("Source MAT files:", len(manifest))
print("Destination:", output_root)

successful = 0
failed = []

for row in tqdm(manifest.itertuples(index=False), total=len(manifest)):

    mat_path = os.path.join(mat_dir, str(row.file))

    try:
        with h5py.File(mat_path, "r") as f:
            image = np.asarray(
                f["cjdata"]["image"][()]
            ).squeeze()

        # Original Cell 61 intensity preprocessing
        image = image.astype(np.float32)
        min_val = np.min(image)
        max_val = np.max(image)

        if max_val > min_val:
            image = (
                (image - min_val) /
                (max_val - min_val) * 255.0
            )
        else:
            image = np.zeros_like(image, dtype=np.float32)

        image = np.clip(image, 0, 255).astype(np.uint8)

        output_path = os.path.join(
            output_root,
            str(row.split),
            str(row.label_name),
            os.path.splitext(str(row.file))[0] + ".jpg"
        )

        # Match the original Cell 61 save operation
        Image.fromarray(image).save(
            output_path,
            format="JPEG",
            quality=95
        )

        successful += 1

    except Exception as e:
        failed.append((str(row.file), str(e)))

print("\n===== RECONSTRUCTION SUMMARY =====")
print("Successfully converted:", successful)
print("Failed:", len(failed))

if failed:
    print("First failures:", failed[:10])

print("\n===== IMAGE COUNTS =====")

for split in splits:
    total = 0

    print(f"\n{split.upper()}")

    for class_name in class_names:
        folder = os.path.join(output_root, split, class_name)

        count = len([
            f for f in os.listdir(folder)
            if f.lower().endswith(".jpg")
        ])

        total += count
        print(f"{class_name}: {count}")

    print("Total:", total)

print("\nReconstructed dataset:", output_root)
```

```python
import os
from PIL import Image
from collections import Counter

root = "/content/brain_tumor_patient_images_reconstructed"

expected = {}

for row in manifest.itertuples(index=False):
    filename = os.path.splitext(str(row.file))[0] + ".jpg"
    key = (str(row.split), str(row.label_name), filename)
    expected[key] = str(row.pid)

actual = set()
corrupted = []
dimension_counts = Counter()
mode_counts = Counter()
label_mismatches = []

for split in ["train", "val", "test"]:
    for class_name in class_names:
        folder = os.path.join(root, split, class_name)

        for filename in os.listdir(folder):
            if not filename.lower().endswith(".jpg"):
                continue

            key = (split, class_name, filename)
            actual.add(key)

            path = os.path.join(folder, filename)

            try:
                with Image.open(path) as img:
                    img.verify()

                with Image.open(path) as img:
                    dimension_counts[img.size] += 1
                    mode_counts[img.mode] += 1

            except Exception as e:
                corrupted.append((key, str(e)))

            if key not in expected:
                label_mismatches.append(key)

expected_keys = set(expected.keys())

missing = expected_keys - actual
unexpected = actual - expected_keys

# Patient leakage audit
patient_splits = manifest.groupby("pid")["split"].nunique()
leaked_patients = patient_splits[patient_splits > 1]

print("===== DATASET INTEGRITY AUDIT =====")
print("Expected images:", len(expected_keys))
print("Actual images:", len(actual))
print("Missing images:", len(missing))
print("Unexpected images:", len(unexpected))
print("Corrupted images:", len(corrupted))
print("Label/path mismatches:", len(label_mismatches))

print("\n===== IMAGE FORMAT AUDIT =====")
print("Image dimensions:", dict(dimension_counts))
print("Image modes:", dict(mode_counts))

print("\n===== PATIENT SPLIT LEAKAGE AUDIT =====")
print("Unique patients:", manifest["pid"].nunique())
print("Patients appearing in multiple splits:", len(leaked_patients))

print("\n===== FINAL STATUS =====")

passed = (
    len(expected_keys) == 3064
    and len(actual) == 3064
    and len(missing) == 0
    and len(unexpected) == 0
    and len(corrupted) == 0
    and len(label_mismatches) == 0
    and len(leaked_patients) == 0
)

print("ALL CHECKS PASSED:", passed)

if missing:
    print("Missing examples:", list(missing)[:5])

if unexpected:
    print("Unexpected examples:", list(unexpected)[:5])

if corrupted:
    print("Corrupted examples:", corrupted[:5])

if len(leaked_patients):
    print("Leaked patient IDs:", leaked_patients.index.tolist())
```

```python
def evaluate_split(split_name):

    dataset = tf.keras.utils.image_dataset_from_directory(
        os.path.join(root, split_name),
        image_size=(224, 224),
        batch_size=32,
        class_names=class_names,
        shuffle=False
    )

    normalized = dataset.map(
        lambda images, labels: (
            tf.cast(images, tf.float32) / 255.0,
            labels
        )
    )

    probabilities = model.predict(normalized, verbose=0)

    filenames = [
        os.path.splitext(os.path.basename(path))[0]
        for path in dataset.file_paths
    ]

    df = pd.DataFrame(
        probabilities,
        columns=probability_columns
    )

    df["file_key"] = filenames
    df["pid"] = df["file_key"].map(
        lambda key: metadata[key]["pid"]
    )
    df["true_class"] = df["file_key"].map(
        lambda key: metadata[key]["label_name"]
    )

    df["predicted_class"] = np.array(class_names)[
        np.argmax(probabilities, axis=1)
    ]

    return df


# Inference only
train_predictions = evaluate_split("train")
val_predictions_new = evaluate_split("val")


def print_metrics(df, level_name):

    print(f"\n===== {level_name} =====")
    print("Samples:", len(df))
    print(
        "Accuracy:",
        round(
            accuracy_score(
                df["true_class"],
                df["predicted_class"]
            ),
            4
        )
    )

    print("\nClassification report:")
    print(
        classification_report(
            df["true_class"],
            df["predicted_class"],
            labels=class_names,
            target_names=class_names,
            zero_division=0,
            digits=4
        )
    )

    print("\nConfusion matrix:")
    print(
        pd.DataFrame(
            confusion_matrix(
                df["true_class"],
                df["predicted_class"],
                labels=class_names
            ),
            index=class_names,
            columns=class_names
        )
    )


def aggregate_by_patient(df):

    patient_df = (
        df.groupby("pid")
        .agg(
            true_class=("true_class", "first"),
            **{
                col: (col, "mean")
                for col in probability_columns
            }
        )
        .reset_index()
    )

    patient_df["predicted_class"] = np.array(class_names)[
        np.argmax(
            patient_df[probability_columns].to_numpy(),
            axis=1
        )
    ]

    return patient_df


# Slice-level comparison
print_metrics(train_predictions, "TRAIN - SLICE LEVEL")
print_metrics(val_predictions_new, "VALIDATION - SLICE LEVEL")

# Patient-level comparison
train_patient_predictions = aggregate_by_patient(
    train_predictions
)

val_patient_predictions_new = aggregate_by_patient(
    val_predictions_new
)

print_metrics(
    train_patient_predictions,
    "TRAIN - PATIENT LEVEL"
)

print_metrics(
    val_patient_predictions_new,
    "VALIDATION - PATIENT LEVEL"
)
```

```python
import pandas as pd
import numpy as np

# Patient-level probability distributions
def probability_profile(df, split_name):

    print(f"\n===== {split_name}: PATIENT-LEVEL PROBABILITY PROFILE =====")

    for true_class in class_names:

        subset = df[df["true_class"] == true_class]

        print(f"\nTrue class: {true_class}")
        print("Patients:", len(subset))

        for probability_col in probability_columns:

            values = subset[probability_col]

            print(
                probability_col,
                "| Mean:", round(values.mean(), 4),
                "| Median:", round(values.median(), 4),
                "| Std:", round(values.std(), 4)
            )


probability_profile(
    train_patient_predictions,
    "TRAIN"
)

probability_profile(
    val_patient_predictions_new,
    "VALIDATION"
)


# Focus specifically on meningioma patients
print("\n===== MENINGIOMA PATIENT CONFIDENCE ANALYSIS =====")

for split_name, df in [
    ("TRAIN", train_patient_predictions),
    ("VALIDATION", val_patient_predictions_new)
]:

    men = df[df["true_class"] == "meningioma"].copy()

    men["strongest_competitor"] = men[
        ["glioma_probability", "pituitary_probability"]
    ].max(axis=1)

    men["meningioma_margin"] = (
        men["meningioma_probability"] -
        men["strongest_competitor"]
    )

    print(f"\n{split_name}")
    print("Patient count:", len(men))
    print(
        "Mean meningioma probability:",
        round(men["meningioma_probability"].mean(), 4)
    )
    print(
        "Mean strongest competitor probability:",
        round(men["strongest_competitor"].mean(), 4)
    )
    print(
        "Mean meningioma margin:",
        round(men["meningioma_margin"].mean(), 4)
    )
    print(
        "Patients with positive meningioma margin:",
        int((men["meningioma_margin"] > 0).sum()),
        "/",
        len(men)
    )
```

```python
import numpy as np
import pandas as pd

# Count training slices contributed by each patient
patient_slice_counts = (
    manifest[manifest["split"] == "train"]
    .groupby(["pid", "label_name"])
    .size()
    .reset_index(name="slice_count")
)

# Add patient-level prediction results
patient_analysis = patient_slice_counts.merge(
    train_patient_predictions,
    on="pid",
    how="inner"
)

patient_analysis["strongest_competitor"] = (
    patient_analysis[
        ["glioma_probability", "pituitary_probability"]
    ].max(axis=1)
)

patient_analysis["true_class_margin"] = (
    patient_analysis.apply(
        lambda row: (
            row[f"{row['true_class']}_probability"]
            - max(
                row[
                    f"{c}_probability"
                ]
                for c in class_names
                if c != row["true_class"]
            )
        ),
        axis=1
    )
)

print("===== PATIENT CONTRIBUTION BY CLASS =====")

summary = (
    patient_analysis.groupby("label_name")
    .agg(
        patients=("pid", "nunique"),
        total_slices=("slice_count", "sum"),
        mean_slices=("slice_count", "mean"),
        median_slices=("slice_count", "median"),
        min_slices=("slice_count", "min"),
        max_slices=("slice_count", "max")
    )
)

summary["slice_share_percent"] = (
    summary["total_slices"] /
    summary["total_slices"].sum() * 100
)

display(summary.round(2))

print("\n===== MENINGIOMA PATIENT ANALYSIS =====")

men = patient_analysis[
    patient_analysis["label_name"] == "meningioma"
].copy()

print("Meningioma training patients:", len(men))

print(
    "Correlation between slice count and true-class margin:",
    round(
        men["slice_count"].corr(
            men["true_class_margin"]
        ),
        4
    )
)

print("\nMeningioma patient-level details:")

display(
    men[
        [
            "pid",
            "slice_count",
            "glioma_probability",
            "meningioma_probability",
            "pituitary_probability",
            "true_class_margin",
            "predicted_class"
        ]
    ].sort_values("slice_count", ascending=False).round(4)
)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

class_order = ["glioma", "meningioma", "pituitary"]

prob_cols = [
    "glioma_probability",
    "meningioma_probability",
    "pituitary_probability"
]

# Verify each patient has only one true class
patient_class_check = slice_df.groupby("pid")["true_class"].nunique()

assert patient_class_check.max() == 1, \
    "At least one patient has inconsistent true-class labels."

# Aggregate slice probabilities into one prediction per patient
patient_df = (
    slice_df.groupby("pid")
    .agg(
        true_class=("true_class", "first"),
        glioma_probability=("glioma_probability", "mean"),
        meningioma_probability=("meningioma_probability", "mean"),
        pituitary_probability=("pituitary_probability", "mean"),
        number_of_slices=("file", "count")
    )
    .reset_index()
)

# Assign predicted class using the highest mean probability
patient_df["predicted_class"] = (
    patient_df[prob_cols]
    .to_numpy()
    .argmax(axis=1)
)

patient_df["predicted_class"] = [
    class_order[i] for i in patient_df["predicted_class"]
]

y_true = patient_df["true_class"]
y_pred = patient_df["predicted_class"]

print("===== PATIENT-LEVEL VALIDATION RESULTS =====")
print("Total patients:", len(patient_df))
print("Patient-level accuracy:", round(accuracy_score(y_true, y_pred) * 100, 2), "%")

print("\n===== CLASSIFICATION REPORT =====")
print(classification_report(
    y_true,
    y_pred,
    labels=class_order,
    zero_division=0,
    digits=4
))

print("\n===== CONFUSION MATRIX =====")
print("Rows = True class | Columns = Predicted class")
display(pd.DataFrame(
    confusion_matrix(y_true, y_pred, labels=class_order),
    index=class_order,
    columns=class_order
))

print("\n===== MENINGIOMA PATIENT-LEVEL RECALL =====")
meningioma_patients = patient_df[
    patient_df["true_class"] == "meningioma"
]
meningioma_correct = (
    meningioma_patients["predicted_class"] == "meningioma"
).sum()

print(
    f"{meningioma_correct}/{len(meningioma_patients)} = "
    f"{meningioma_correct / len(meningioma_patients) * 100:.2f}%"
)

print("\n===== PATIENT SLICE COUNTS =====")
display(
    patient_df.groupby("true_class")["number_of_slices"]
    .describe()
    .round(2)
)

print("\n===== PATIENT PREDICTIONS =====")
display(patient_df.head(15))
```

```python
import os
import numpy as np
import pandas as pd

DATA_ROOT = "/content/brain_tumor_patient_images_reconstructed"

# Get the original training rows from the patient-level manifest
train_weight_df = manifest[
    manifest["split"] == "train"
].copy()

# Convert MAT filenames to reconstructed JPG filenames
train_weight_df["image_file"] = train_weight_df["file"].apply(
    lambda x: os.path.splitext(os.path.basename(str(x)))[0] + ".jpg"
)

# Build the expected image paths
train_weight_df["image_path"] = train_weight_df.apply(
    lambda row: os.path.join(
        DATA_ROOT, "train", row["label_name"], row["image_file"]
    ),
    axis=1
)

# Verify every image exists
train_weight_df["exists"] = train_weight_df["image_path"].apply(
    os.path.isfile
)

# Count slices belonging to each patient
train_weight_df["patient_slice_count"] = (
    train_weight_df.groupby("pid")["pid"].transform("size")
)

# Each patient contributes a total raw weight of 1
train_weight_df["raw_weight"] = (
    1.0 / train_weight_df["patient_slice_count"]
)

# Normalize weights so their overall mean is 1
train_weight_df["sample_weight"] = (
    train_weight_df["raw_weight"] /
    train_weight_df["raw_weight"].mean()
)

print("===== DATA INTEGRITY =====")
print("Training manifest rows:", len(train_weight_df))
print("Existing image files:", train_weight_df["exists"].sum())
print("Missing image files:", (~train_weight_df["exists"]).sum())
print("Unique patients:", train_weight_df["pid"].nunique())
print("Duplicate image filenames:", train_weight_df["image_file"].duplicated().sum())

print("\n===== PATIENT-BALANCED WEIGHT SUMMARY =====")
display(
    train_weight_df.groupby("label_name").agg(
        patients=("pid", "nunique"),
        slices=("image_file", "count"),
        mean_slices_per_patient=("patient_slice_count", "mean"),
        minimum_weight=("sample_weight", "min"),
        maximum_weight=("sample_weight", "max"),
        total_weight=("sample_weight", "sum")
    ).round(3)
)

print("\nOverall mean sample weight:",
      round(train_weight_df["sample_weight"].mean(), 4))

print("\n===== WEIGHT EXAMPLES =====")
display(
    train_weight_df[
        ["image_file", "pid", "label_name",
         "patient_slice_count", "sample_weight"]
    ].head(15)
)

assert train_weight_df["exists"].all(), "Some training images are missing."
assert train_weight_df["image_file"].is_unique, "Duplicate image filenames detected."
assert train_weight_df["pid"].notna().all(), "Some patient IDs are missing."
```

```python
# One row per patient, avoiding slice-weighted averages
patient_counts = (
    train_weight_df.groupby(["pid", "label_name"])
    .size()
    .reset_index(name="number_of_slices")
)

print("===== TRUE PATIENT-LEVEL SLICE DISTRIBUTION =====")

display(
    patient_counts.groupby("label_name")["number_of_slices"]
    .agg(
        patients="count",
        mean="mean",
        median="median",
        minimum="min",
        maximum="max",
        std="std"
    )
    .round(2)
)

print("\n===== PATIENT WEIGHT VALIDATION =====")

patient_weight_totals = (
    train_weight_df.groupby("pid")["raw_weight"].sum()
)

print("Minimum total raw weight per patient:",
      round(patient_weight_totals.min(), 6))
print("Maximum total raw weight per patient:",
      round(patient_weight_totals.max(), 6))
print("All patients equally weighted:",
      np.allclose(patient_weight_totals.values, 1.0))
```

```python
import tensorflow as tf
import numpy as np
import os

# Fixed experiment configuration
CLASS_ORDER = ["glioma", "meningioma", "pituitary"]
CLASS_TO_INDEX = {
    name: index for index, name in enumerate(CLASS_ORDER)
}

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

# Prepare training arrays
train_paths = train_weight_df["image_path"].to_numpy()

train_labels = np.array([
    CLASS_TO_INDEX[name]
    for name in train_weight_df["label_name"]
], dtype=np.int32)

train_weights = train_weight_df["sample_weight"].to_numpy(
    dtype=np.float32
)

# Prepare validation arrays directly from the manifest
val_manifest = manifest[
    manifest["split"] == "val"
].copy()

val_manifest["image_file"] = val_manifest["file"].apply(
    lambda x: os.path.splitext(os.path.basename(str(x)))[0] + ".jpg"
)

val_manifest["image_path"] = val_manifest.apply(
    lambda row: os.path.join(
        DATA_ROOT, "val", row["label_name"], row["image_file"]
    ),
    axis=1
)

val_paths = val_manifest["image_path"].to_numpy()

val_labels = np.array([
    CLASS_TO_INDEX[name]
    for name in val_manifest["label_name"]
], dtype=np.int32)

# Image loading function: RGB, resize, normalize
def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, IMAGE_SIZE)
    image = tf.cast(image, tf.float32) / 255.0

    label = tf.one_hot(label, depth=len(CLASS_ORDER))
    return image, label

# Weighted training loader
def load_weighted_image(path, label, weight):
    image, label = load_image(path, label)
    return image, label, weight

train_ds_weighted = tf.data.Dataset.from_tensor_slices(
    (train_paths, train_labels, train_weights)
)

train_ds_weighted = (
    train_ds_weighted
    .shuffle(
        buffer_size=len(train_paths),
        seed=SEED,
        reshuffle_each_iteration=True
    )
    .map(load_weighted_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# Unweighted validation loader
val_ds_weighted_experiment = tf.data.Dataset.from_tensor_slices(
    (val_paths, val_labels)
)

val_ds_weighted_experiment = (
    val_ds_weighted_experiment
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

# Verify dataset sizes and batch structure
print("===== DATASET VERIFICATION =====")
print("Training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Training batches:", tf.data.experimental.cardinality(train_ds_weighted).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds_weighted_experiment).numpy())

for images, labels, weights in train_ds_weighted.take(1):
    print("\nTraining batch image shape:", images.shape)
    print("Training batch label shape:", labels.shape)
    print("Training batch weight shape:", weights.shape)
    print("Sample weights:", weights[:5].numpy())

for images, labels in val_ds_weighted_experiment.take(1):
    print("\nValidation batch image shape:", images.shape)
    print("Validation batch label shape:", labels.shape)

print("\nClass order:", CLASS_ORDER)
print("Training weight mean:", round(float(train_weights.mean()), 4))
print("All validation image paths exist:", all(os.path.isfile(p) for p in val_paths))
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

WEIGHTED_CKPT = "/content/drive/MyDrive/results/patient_balanced_cnn_best.keras"

assert os.path.exists(WEIGHTED_CKPT), "Saved weighted checkpoint not found."

# Load best saved checkpoint, not the interrupted in-memory model
weighted_best = tf.keras.models.load_model(
    WEIGHTED_CKPT,
    compile=False
)

CLASS_ORDER = ["glioma", "meningioma", "pituitary"]
PROB_COLS = [
    "glioma_probability",
    "meningioma_probability",
    "pituitary_probability"
]

# Predict validation images (validation dataset is not shuffled)
val_probabilities = weighted_best.predict(
    val_ds_weighted_experiment,
    verbose=1
)

assert len(val_probabilities) == len(val_manifest)

# Construct slice-level results using the matching manifest order
slice_df_weighted = val_manifest[
    ["file", "pid", "label_name"]
].copy().reset_index(drop=True)

slice_df_weighted["true_class"] = slice_df_weighted["label_name"]

for i, col in enumerate(PROB_COLS):
    slice_df_weighted[col] = val_probabilities[:, i]

slice_df_weighted["predicted_class"] = [
    CLASS_ORDER[i] for i in val_probabilities.argmax(axis=1)
]

# Slice-level evaluation
print("===== SLICE-LEVEL VALIDATION =====")
print("Images:", len(slice_df_weighted))
print("Accuracy:",
      round(accuracy_score(
          slice_df_weighted["true_class"],
          slice_df_weighted["predicted_class"]
      ) * 100, 2), "%")

print(classification_report(
    slice_df_weighted["true_class"],
    slice_df_weighted["predicted_class"],
    labels=CLASS_ORDER,
    zero_division=0,
    digits=4
))

print("Slice-level confusion matrix:")
display(pd.DataFrame(
    confusion_matrix(
        slice_df_weighted["true_class"],
        slice_df_weighted["predicted_class"],
        labels=CLASS_ORDER
    ),
    index=CLASS_ORDER,
    columns=CLASS_ORDER
))

# Verify patient-level labels are consistent
assert slice_df_weighted.groupby("pid")["true_class"].nunique().max() == 1

# Aggregate probabilities by patient
patient_df_weighted = (
    slice_df_weighted.groupby("pid")
    .agg(
        true_class=("true_class", "first"),
        glioma_probability=("glioma_probability", "mean"),
        meningioma_probability=("meningioma_probability", "mean"),
        pituitary_probability=("pituitary_probability", "mean"),
        number_of_slices=("file", "count")
    )
    .reset_index()
)

patient_df_weighted["predicted_class"] = [
    CLASS_ORDER[i]
    for i in patient_df_weighted[PROB_COLS].to_numpy().argmax(axis=1)
]

# Patient-level evaluation
y_true = patient_df_weighted["true_class"]
y_pred = patient_df_weighted["predicted_class"]

print("\n===== PATIENT-LEVEL VALIDATION =====")
print("Patients:", len(patient_df_weighted))
print("Accuracy:",
      round(accuracy_score(y_true, y_pred) * 100, 2), "%")

print(classification_report(
    y_true,
    y_pred,
    labels=CLASS_ORDER,
    zero_division=0,
    digits=4
))

print("Patient-level confusion matrix:")
display(pd.DataFrame(
    confusion_matrix(y_true, y_pred, labels=CLASS_ORDER),
    index=CLASS_ORDER,
    columns=CLASS_ORDER
))

meningioma_patients = patient_df_weighted[
    patient_df_weighted["true_class"] == "meningioma"
]

correct_meningioma = (
    meningioma_patients["predicted_class"] == "meningioma"
).sum()

print("\nMeningioma patient recall:",
      f"{correct_meningioma}/{len(meningioma_patients)} = "
      f"{correct_meningioma / len(meningioma_patients) * 100:.2f}%")

print("\n===== MENINGIOMA PATIENT PROBABILITIES =====")
display(
    meningioma_patients[
        ["pid", "meningioma_probability",
         "glioma_probability", "pituitary_probability",
         "predicted_class", "number_of_slices"]
    ].round(4)
)
```

```python
import numpy as np
import pandas as pd
import os

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

CLASS_ORDER = ["glioma", "meningioma", "pituitary"]
PROB_COLS = [
    "glioma_probability",
    "meningioma_probability",
    "pituitary_probability"
]

# Generate test predictions
test_probabilities = final_candidate.predict(
    test_ds,
    verbose=1
)

assert test_probabilities.shape == (462, 3)

# Build slice-level prediction dataframe
test_results = test_manifest_ordered.copy().reset_index(drop=True)

test_results["true_label"] = (
    test_results["label_name"].astype(str)
)

for i, col in enumerate(PROB_COLS):
    test_results[col] = test_probabilities[:, i]

test_results["predicted_label"] = [
    CLASS_ORDER[i]
    for i in np.argmax(test_probabilities, axis=1)
]

# Slice-level evaluation
y_slice_true = test_results["true_label"]
y_slice_pred = test_results["predicted_label"]

print("===== SLICE-LEVEL TEST RESULTS =====")
print("Images:", len(test_results))
print("Accuracy:", round(accuracy_score(y_slice_true, y_slice_pred), 4))
print(classification_report(
    y_slice_true,
    y_slice_pred,
    labels=CLASS_ORDER,
    zero_division=0
))

print("Slice-level confusion matrix:")
print(pd.DataFrame(
    confusion_matrix(
        y_slice_true,
        y_slice_pred,
        labels=CLASS_ORDER
    ),
    index=CLASS_ORDER,
    columns=CLASS_ORDER
))

# Patient-level aggregation: mean slice probabilities
patient_results = (
    test_results
    .groupby("pid", as_index=False)
    .agg(
        true_label=("true_label", "first"),
        glioma_probability=("glioma_probability", "mean"),
        meningioma_probability=("meningioma_probability", "mean"),
        pituitary_probability=("pituitary_probability", "mean"),
        slice_count=("file", "count")
    )
)

# Confirm each patient has only one true class
assert (
    test_results.groupby("pid")["true_label"].nunique().max() == 1
)

patient_results["predicted_label"] = [
    CLASS_ORDER[i]
    for i in np.argmax(
        patient_results[PROB_COLS].to_numpy(),
        axis=1
    )
]

y_patient_true = patient_results["true_label"]
y_patient_pred = patient_results["predicted_label"]

print("\n===== PATIENT-LEVEL TEST RESULTS =====")
print("Patients:", len(patient_results))
print("Accuracy:", round(
    accuracy_score(y_patient_true, y_patient_pred), 4
))

print(classification_report(
    y_patient_true,
    y_patient_pred,
    labels=CLASS_ORDER,
    zero_division=0
))

print("Patient-level confusion matrix:")
print(pd.DataFrame(
    confusion_matrix(
        y_patient_true,
        y_patient_pred,
        labels=CLASS_ORDER
    ),
    index=CLASS_ORDER,
    columns=CLASS_ORDER
))

# Save the untouched test predictions
RESULTS_DIR = "/content/drive/MyDrive/results"

slice_output = os.path.join(
    RESULTS_DIR,
    "patient_balanced_test_slice_predictions.csv"
)

patient_output = os.path.join(
    RESULTS_DIR,
    "patient_balanced_test_patient_predictions.csv"
)

assert not os.path.exists(slice_output)
assert not os.path.exists(patient_output)

test_results.to_csv(slice_output, index=False)
patient_results.to_csv(patient_output, index=False)

print("\n===== TEST ARTIFACTS SAVED =====")
print(slice_output, os.path.exists(slice_output))
print(patient_output, os.path.exists(patient_output))

print("\n===== MENINGIOMA PATIENT RECALL =====")
meningioma_mask = y_patient_true == "meningioma"
correct_meningioma = (
    (y_patient_pred == "meningioma") & meningioma_mask
).sum()

total_meningioma = meningioma_mask.sum()

print(
    f"{correct_meningioma}/{total_meningioma} = "
    f"{correct_meningioma / total_meningioma:.2%}"
)
```

```python
import os
import zipfile

ZIP_PATH = "/content/drive/MyDrive/external_BDNeuro_MRI/Brain_Tumor_MRI_Dataset_Final.zip"

EXTRACT_DIR = "/content/external_BDNeuro_MRI_extracted"

os.makedirs(EXTRACT_DIR, exist_ok=True)

print("===== SAFE EXTRACTION =====")

with zipfile.ZipFile(ZIP_PATH, "r") as archive:

    members = archive.namelist()
    destination = os.path.realpath(EXTRACT_DIR)

    # Check every archive path before extracting
    unsafe_paths = []

    for member in members:
        target = os.path.realpath(
            os.path.join(destination, member)
        )

        if os.path.commonpath([destination, target]) != destination:
            unsafe_paths.append(member)

    assert len(unsafe_paths) == 0, (
        f"Unsafe archive paths detected: {unsafe_paths[:10]}"
    )

    print("Path safety check: PASSED")

    archive.extractall(EXTRACT_DIR)

print("Extraction completed.")

print("\n===== EXTRACTED DIRECTORY =====")

for root, dirs, files in os.walk(EXTRACT_DIR):
    level = root.replace(EXTRACT_DIR, "").count(os.sep)

    if level > 2:
        dirs[:] = []
        continue

    print(
        "  " * level,
        os.path.basename(root) or "external dataset",
        f"| {len(files)} files"
    )

print("\nExtraction directory:", EXTRACT_DIR)
```

```python
import os
from collections import Counter, defaultdict
from PIL import Image

EXTRACT_DIR = "/content/external_BDNeuro_MRI_extracted"

SPLITS = ["train", "val", "test"]
CLASSES = ["glioma", "meningioma", "pituitary", "no_tumor"]

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")

dimension_counts = Counter()
mode_counts = Counter()
format_counts = Counter()

split_counts = defaultdict(Counter)

corrupted_images = []
image_records = []

print("===== EXTERNAL IMAGE QUALITY AUDIT =====")

for split in SPLITS:
    for class_name in CLASSES:

        folder = os.path.join(EXTRACT_DIR, split, class_name)

        assert os.path.isdir(folder), f"Missing folder: {folder}"

        for filename in os.listdir(folder):

            if not filename.lower().endswith(image_extensions):
                continue

            path = os.path.join(folder, filename)

            try:
                with Image.open(path) as img:
                    img.verify()

                with Image.open(path) as img:
                    width, height = img.size
                    mode = img.mode
                    fmt = img.format

                dimension_counts[(width, height)] += 1
                mode_counts[mode] += 1
                format_counts[fmt] += 1

                split_counts[split][class_name] += 1

                image_records.append({
                    "split": split,
                    "class": class_name,
                    "filename": filename,
                    "path": path,
                    "width": width,
                    "height": height,
                    "mode": mode,
                    "format": fmt
                })

            except Exception as e:
                corrupted_images.append((path, str(e)))

print("\n===== IMAGE COUNTS =====")

print("Readable images:", len(image_records))
print("Corrupted images:", len(corrupted_images))

for split in SPLITS:
    print(f"\n{split.upper()}:")
    for class_name in CLASSES:
        print(
            f"{class_name}: "
            f"{split_counts[split][class_name]}"
        )

print("\n===== IMAGE DIMENSIONS =====")
for item, count in dimension_counts.most_common(15):
    print(f"{item}: {count}")

print("\n===== IMAGE COLOUR MODES =====")
print(dict(mode_counts))

print("\n===== IMAGE FORMATS =====")
print(dict(format_counts))

print("\n===== CORRUPTED IMAGE EXAMPLES =====")
for item in corrupted_images[:10]:
    print(item)

print("\n===== SAMPLE IMAGE RECORDS =====")
for record in image_records[:5]:
    print(record)
```

```python
import os
import sys
import subprocess
import hashlib
import numpy as np
import pandas as pd
from PIL import Image

try:
    import imagehash
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "ImageHash"
    ])
    import imagehash

EXTERNAL_DIR = "/content/external_BDNeuro_MRI_extracted"
ORIGINAL_DIR = "/content/brain_tumor_patient_images_reconstructed"

SPLITS = ["train", "val", "test"]
CLASSES = ["glioma", "meningioma", "pituitary", "no_tumor"]

EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")

def collect_valid_images(root, external=False):
    records = []

    for split in SPLITS:
        for class_name in CLASSES:

            folder = os.path.join(root, split, class_name)

            if not os.path.isdir(folder):
                continue

            for filename in os.listdir(folder):
                if filename.lower().endswith(EXTENSIONS):
                    path = os.path.join(folder, filename)

                    records.append({
                        "path": path,
                        "split": split,
                        "class": class_name,
                        "filename": filename
                    })

    return records

external_records = collect_valid_images(EXTERNAL_DIR)
original_records = collect_valid_images(ORIGINAL_DIR)

print("External MRI images:", len(external_records))
print("Original MRI images:", len(original_records))

assert len(external_records) == 5941
assert len(original_records) == 3064

def calculate_hashes(records, label):

    sha_hashes = []
    phash_values = []

    for i, record in enumerate(records):

        path = record["path"]

        sha = hashlib.sha256()

        with open(path, "rb") as f:
            for block in iter(lambda: f.read(1024 * 1024), b""):
                sha.update(block)

        sha_hashes.append(sha.hexdigest())

        with Image.open(path) as img:
            phash = imagehash.phash(
                img.convert("L"), hash_size=8
            )

        phash_values.append(int(str(phash), 16))

        if (i + 1) % 1000 == 0:
            print(f"{label}: {i + 1}/{len(records)}")

    return sha_hashes, np.array(phash_values, dtype=np.uint64)

print("\nCalculating external hashes...")
external_sha, external_phash = calculate_hashes(
    external_records, "External"
)

print("\nCalculating original hashes...")
original_sha, original_phash = calculate_hashes(
    original_records, "Original"
)

# Exact matching
original_sha_set = set(original_sha)

exact_matches = [
    i for i, value in enumerate(external_sha)
    if value in original_sha_set
]

print("\n===== EXACT MATCH RESULTS =====")
print("Exact duplicate matches:", len(exact_matches))

# Perceptual matching
lookup = np.array(
    [bin(i).count("1") for i in range(256)],
    dtype=np.uint8
)

near_matches = []
nearest_distances = []

print("\n===== PERCEPTUAL HASH SCREENING =====")

CHUNK_SIZE = 128

for start in range(0, len(external_phash), CHUNK_SIZE):

    end = min(start + CHUNK_SIZE, len(external_phash))

    xor = np.bitwise_xor(
        external_phash[start:end, None],
        original_phash[None, :]
    )

    byte_view = xor.view(np.uint8).reshape(
        end - start, len(original_phash), 8
    )

    distances = lookup[byte_view].sum(axis=2)

    nearest_indices = distances.argmin(axis=1)
    minimum_distances = distances[
        np.arange(end - start), nearest_indices
    ]

    for local_i, (nearest_j, distance) in enumerate(
        zip(nearest_indices, minimum_distances)
    ):

        external_i = start + local_i

        nearest_distances.append(int(distance))

        if distance <= 5:
            near_matches.append({
                "external_index": external_i,
                "original_index": int(nearest_j),
                "distance": int(distance)
            })

    if end % 1000 == 0 or end == len(external_phash):
        print(f"Compared {end}/{len(external_phash)}")

print("\n===== NEAR MATCH SUMMARY =====")
print("Matches with pHash distance <= 5:", len(near_matches))
print("Minimum nearest distance:", min(nearest_distances))
print("Median nearest distance:", np.median(nearest_distances))

candidate_rows = []

for match in near_matches[:100]:

    ext = external_records[match["external_index"]]
    orig = original_records[match["original_index"]]

    candidate_rows.append({
        "distance": match["distance"],
        "external_class": ext["class"],
        "original_class": orig["class"],
        "external_path": ext["path"],
        "original_path": orig["path"]
    })

candidate_df = pd.DataFrame(candidate_rows)

print("\n===== FIRST 30 CANDIDATE MATCHES =====")

if len(candidate_df):
    print(candidate_df.head(30).to_string(index=False))
else:
    print("No candidate matches.")

print("\nAudit completed.")
```

```python
import os
import hashlib
import numpy as np
import pandas as pd
from PIL import Image

import imagehash

EXTERNAL_DIR = "/content/external_BDNeuro_MRI_extracted"

SPLITS = ["train", "val", "test"]
CLASSES = ["glioma", "meningioma", "pituitary", "no_tumor"]

EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff")

print("===== CROSS-SPLIT DUPLICATE AUDIT =====")

# Collect only images in valid split/class folders
records = []

for split in SPLITS:
    for class_name in CLASSES:

        folder = os.path.join(EXTERNAL_DIR, split, class_name)

        for filename in os.listdir(folder):
            if filename.lower().endswith(EXTENSIONS):
                records.append({
                    "path": os.path.join(folder, filename),
                    "split": split,
                    "class": class_name,
                    "filename": filename
                })

print("Total images:", len(records))

assert len(records) == 5941

# Calculate hashes
sha_values = []
phash_values = []

for i, record in enumerate(records):

    sha = hashlib.sha256()

    with open(record["path"], "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            sha.update(block)

    sha_values.append(sha.hexdigest())

    with Image.open(record["path"]) as img:
        phash = imagehash.phash(
            img.convert("L"), hash_size=8
        )

    phash_values.append(int(str(phash), 16))

    if (i + 1) % 1000 == 0:
        print(f"Hashed {i + 1}/{len(records)}")

phash_values = np.array(phash_values, dtype=np.uint64)

# Exact duplicates across different splits
sha_to_indices = {}

for i, sha in enumerate(sha_values):
    sha_to_indices.setdefault(sha, []).append(i)

exact_cross_split = []

for indices in sha_to_indices.values():
    for a in range(len(indices)):
        for b in range(a + 1, len(indices)):
            i, j = indices[a], indices[b]

            if records[i]["split"] != records[j]["split"]:
                exact_cross_split.append((i, j))

print("\n===== EXACT CROSS-SPLIT MATCHES =====")
print("Exact matching pairs:", len(exact_cross_split))

# Perceptual hash comparison across split combinations
lookup = np.array(
    [bin(i).count("1") for i in range(256)],
    dtype=np.uint8
)

split_pairs = [
    ("train", "val"),
    ("train", "test"),
    ("val", "test")
]

near_matches = []

for split_a, split_b in split_pairs:

    indices_a = [
        i for i, r in enumerate(records)
        if r["split"] == split_a
    ]

    indices_b = [
        i for i, r in enumerate(records)
        if r["split"] == split_b
    ]

    hashes_a = phash_values[indices_a]
    hashes_b = phash_values[indices_b]

    print(f"\nComparing {split_a} vs {split_b}")

    for start in range(0, len(hashes_a), 128):

        end = min(start + 128, len(hashes_a))

        xor = np.bitwise_xor(
            hashes_a[start:end, None],
            hashes_b[None, :]
        )

        byte_view = xor.view(np.uint8).reshape(
            end - start, len(hashes_b), 8
        )

        distances = lookup[byte_view].sum(axis=2)

        candidate_positions = np.argwhere(distances <= 5)

        for local_i, j in candidate_positions:

            i = indices_a[start + local_i]
            k = indices_b[j]

            near_matches.append({
                "split_a": split_a,
                "split_b": split_b,
                "distance": int(distances[local_i, j]),
                "class_a": records[i]["class"],
                "class_b": records[k]["class"],
                "path_a": records[i]["path"],
                "path_b": records[k]["path"]
            })

    print("Completed comparison.")

# Summary
print("\n===== NEAR-DUPLICATE SUMMARY =====")
print("Total cross-split candidates (distance <= 5):", len(near_matches))

print(
    "Very close candidates (distance <= 3):",
    sum(m["distance"] <= 3 for m in near_matches)
)

if near_matches:

    result_df = pd.DataFrame(near_matches)

    print("\n===== CANDIDATES BY SPLIT PAIR =====")
    print(
        result_df.groupby(
            ["split_a", "split_b"]
        ).size().to_string()
    )

    print("\n===== CANDIDATES BY CLASS PAIR =====")
    print(
        result_df.groupby(
            ["class_a", "class_b"]
        ).size().to_string()
    )

    print("\n===== FIRST 30 CANDIDATES =====")
    print(result_df.head(30).to_string(index=False))

else:
    print("No cross-split candidates found.")

print("\n===== AUDIT COMPLETE =====")
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image
from skimage.metrics import structural_similarity as ssim

ROOT = "/content/external_BDNeuro_MRI_extracted"

pairs = [
    ("train/glioma/glioma_train_00732.jpg", "val/meningioma/meningioma_val_00138.jpg", 4),
    ("train/glioma/glioma_train_01515.jpg", "val/meningioma/meningioma_val_00076.jpg", 4),
    ("train/glioma/glioma_train_00013.jpg", "val/meningioma/meningioma_val_00134.jpg", 4),
    ("train/meningioma/meningioma_train_00004.jpg", "val/glioma/glioma_val_00145.jpg", 4),
    ("train/meningioma/meningioma_train_00561.jpg", "val/glioma/glioma_val_00303.jpg", 2),
    ("train/meningioma/meningioma_train_00122.jpg", "val/glioma/glioma_val_00136.jpg", 4),
    ("train/meningioma/meningioma_train_00332.jpg", "val/glioma/glioma_val_00018.jpg", 4),
    ("train/pituitary/pituitary_train_00617.jpg", "val/meningioma/meningioma_val_00098.jpg", 0),
    ("train/no_tumor/no_tumor_train_00189.jpg", "val/glioma/glioma_val_00068.jpg", 4),
    ("train/glioma/glioma_train_01271.jpg", "test/no_tumor/no_tumor_test_00004.jpg", 4),
    ("train/meningioma/meningioma_train_00636.jpg", "test/glioma/glioma_test_00033.jpg", 4)
]

results = []

for path_a, path_b, phash_distance in pairs:

    with Image.open(os.path.join(ROOT, path_a)) as img:
        a = np.array(
            img.convert("L").resize((224, 224))
        )

    with Image.open(os.path.join(ROOT, path_b)) as img:
        b = np.array(
            img.convert("L").resize((224, 224))
        )

    similarity = ssim(a, b, data_range=255)

    normalized_mae = np.mean(
        np.abs(a.astype(float) - b.astype(float))
    ) / 255

    results.append({
        "image_a": path_a.split("/")[-1],
        "image_b": path_b.split("/")[-1],
        "pHash_distance": phash_distance,
        "SSIM": round(float(similarity), 4),
        "normalized_MAE": round(float(normalized_mae), 4)
    })

result_df = pd.DataFrame(results)

print("===== SECONDARY SIMILARITY AUDIT =====")
print(result_df.to_string(index=False))

print("\nMean SSIM:", round(result_df["SSIM"].mean(), 4))
print("Maximum SSIM:", round(result_df["SSIM"].max(), 4))
```

```python
import os
import hashlib
import numpy as np
import pandas as pd
import tensorflow as tf

from PIL import Image
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

# --------------------------------------------------
# 1. Paths and frozen evaluation configuration
# --------------------------------------------------

ROOT = "/content/external_BDNeuro_MRI_extracted"
RESULTS_DIR = "/content/drive/MyDrive/results"

MODEL_PATH = os.path.join(
    RESULTS_DIR, "patient_balanced_cnn_best.keras"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR, "BDNeuro_external_test_predictions.csv"
)

CLASS_NAMES = ["glioma", "meningioma", "pituitary"]
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 32

print("===== EXTERNAL TEST EVALUATION =====")

# --------------------------------------------------
# 2. Load frozen checkpoint
# --------------------------------------------------

assert os.path.exists(MODEL_PATH), "Model checkpoint missing."

model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

assert model.input_shape == (None, 224, 224, 3)
assert model.output_shape == (None, 3)
assert model.count_params() == 110147

# Record checkpoint hash
sha = hashlib.sha256()

with open(MODEL_PATH, "rb") as f:
    for block in iter(lambda: f.read(1024 * 1024), b""):
        sha.update(block)

model_hash = sha.hexdigest()

print("Checkpoint verified.")
print("SHA-256:", model_hash)

# --------------------------------------------------
# 3. Build external TEST manifest
#    Only the three tumour classes are included.
# --------------------------------------------------

records = []

for class_name in CLASS_NAMES:

    folder = os.path.join(
        ROOT, "test", class_name
    )

    assert os.path.isdir(folder), (
        f"Missing class directory: {folder}"
    )

    for filename in sorted(os.listdir(folder)):

        if filename.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp")
        ):

            records.append({
                "path": os.path.join(folder, filename),
                "filename": filename,
                "true_label": class_name
            })

manifest = pd.DataFrame(records)

expected_counts = {
    "glioma": 328,
    "meningioma": 221,
    "pituitary": 222
}

actual_counts = manifest["true_label"].value_counts().to_dict()

print("\nExternal test image counts:")
print(actual_counts)

assert actual_counts == expected_counts
assert len(manifest) == 771

print("Total tumour-positive test images:", len(manifest))
print("No-tumour images excluded: 118")

# --------------------------------------------------
# 4. Predict once using original preprocessing
#    RGB, 224x224, pixel values divided by 255.
# --------------------------------------------------

all_probabilities = []

for start in range(0, len(manifest), BATCH_SIZE):

    batch_records = manifest.iloc[
        start:start + BATCH_SIZE
    ]

    batch_images = []

    for path in batch_records["path"]:

        with Image.open(path) as img:

            img = img.convert("RGB")
            img = img.resize(
                IMAGE_SIZE,
                Image.Resampling.BILINEAR
            )

            array = np.asarray(
                img, dtype=np.float32
            ) / 255.0

            batch_images.append(array)

    batch_array = np.stack(batch_images)

    probabilities = model(
        batch_array, training=False
    ).numpy()

    all_probabilities.append(probabilities)

    print(
        f"Predicted {min(start + BATCH_SIZE, len(manifest))}"
        f"/{len(manifest)}"
    )

probabilities = np.concatenate(
    all_probabilities, axis=0
)

assert probabilities.shape == (771, 3)
assert np.isfinite(probabilities).all()

# --------------------------------------------------
# 5. Save raw predictions and probabilities
# --------------------------------------------------

true_labels = manifest["true_label"].to_numpy()

true_indices = np.array([
    CLASS_NAMES.index(label)
    for label in true_labels
])

predicted_indices = np.argmax(
    probabilities, axis=1
)

predicted_labels = [
    CLASS_NAMES[i] for i in predicted_indices
]

prediction_df = manifest[
    ["filename", "true_label"]
].copy()

prediction_df["predicted_label"] = predicted_labels

for i, class_name in enumerate(CLASS_NAMES):
    prediction_df[f"prob_{class_name}"] = probabilities[:, i]

prediction_df["model_sha256"] = model_hash

prediction_df.to_csv(
    OUTPUT_PATH, index=False
)

print("\nPredictions saved:", OUTPUT_PATH)

# --------------------------------------------------
# 6. Calculate external image-level metrics
# --------------------------------------------------

print("\n===== EXTERNAL TEST RESULTS =====")

accuracy = accuracy_score(
    true_indices, predicted_indices
)

balanced_accuracy = balanced_accuracy_score(
    true_indices, predicted_indices
)

macro_f1 = f1_score(
    true_indices,
    predicted_indices,
    average="macro"
)

print("Image-level accuracy:", round(accuracy, 4))
print("Balanced accuracy:", round(balanced_accuracy, 4))
print("Macro F1:", round(macro_f1, 4))

print("\nClassification report:")

print(
    classification_report(
        true_indices,
        predicted_indices,
        labels=[0, 1, 2],
        target_names=CLASS_NAMES,
        digits=4,
        zero_division=0
    )
)

print("Confusion matrix:")
print(
    confusion_matrix(
        true_indices,
        predicted_indices,
        labels=[0, 1, 2]
    )
)

print("\n===== EVALUATION COMPLETE =====")
print("Checkpoint unchanged.")
print("No training or fine-tuning performed.")
print("Patient-level metrics: NOT AVAILABLE.")
```

```python
import os
import hashlib
import pandas as pd
from PIL import Image
import imagehash

ROOT = "/content/external_BDNeuro_MRI_extracted"
CLASSES = ["glioma", "meningioma", "pituitary"]

records = []

for class_name in CLASSES:
    folder = os.path.join(ROOT, "test", class_name)

    for filename in sorted(os.listdir(folder)):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):
            path = os.path.join(folder, filename)

            with open(path, "rb") as f:
                sha = hashlib.sha256(f.read()).hexdigest()

            with Image.open(path) as img:
                phash = imagehash.phash(
                    img.convert("L"), hash_size=8
                )

            records.append({
                "filename": filename,
                "class": class_name,
                "path": path,
                "sha256": sha,
                "phash": phash
            })

print("===== TEST-SPLIT INTERNAL DUPLICATE AUDIT =====")
print("Images inspected:", len(records))

# Exact duplicate groups
sha_groups = {}

for r in records:
    sha_groups.setdefault(r["sha256"], []).append(r)

exact_groups = [
    group for group in sha_groups.values()
    if len(group) > 1
]

print("\nExact duplicate groups:", len(exact_groups))

# Perceptual near-duplicate pairs
near_matches = []

for i in range(len(records)):
    for j in range(i + 1, len(records)):

        distance = records[i]["phash"] - records[j]["phash"]

        if distance <= 5:
            near_matches.append({
                "image_a": records[i]["filename"],
                "class_a": records[i]["class"],
                "image_b": records[j]["filename"],
                "class_b": records[j]["class"],
                "phash_distance": distance
            })

near_df = pd.DataFrame(near_matches)

print("\nNear-duplicate pairs (pHash distance <= 5):",
      len(near_df))

if len(near_df):
    print("\nCandidates by class combination:")
    print(
        near_df.groupby(
            ["class_a", "class_b"]
        ).size().to_string()
    )

    print("\nFirst 30 candidates:")
    print(near_df.head(30).to_string(index=False))

print("\n===== AUDIT COMPLETE =====")
```

```python
import os
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    recall_score
)

RESULTS_DIR = "/content/drive/MyDrive/results"

PREDICTION_PATH = os.path.join(
    RESULTS_DIR,
    "BDNeuro_external_test_predictions.csv"
)

OUTPUT_PATH = os.path.join(
    RESULTS_DIR,
    "BDNeuro_external_bootstrap_CI.csv"
)

CLASS_NAMES = ["glioma", "meningioma", "pituitary"]

N_BOOTSTRAPS = 5000
RANDOM_SEED = 42

print("===== EXTERNAL TEST STATISTICAL ANALYSIS =====")

# Load the original saved predictions
df = pd.read_csv(PREDICTION_PATH)

assert len(df) == 771
assert set(df["true_label"].unique()) == set(CLASS_NAMES)

y_true = df["true_label"].to_numpy()
y_pred = df["predicted_label"].to_numpy()

# Original observed metrics
observed = {
    "Accuracy": accuracy_score(y_true, y_pred),
    "Balanced accuracy": balanced_accuracy_score(y_true, y_pred),
    "Macro F1": f1_score(
        y_true, y_pred,
        labels=CLASS_NAMES,
        average="macro",
        zero_division=0
    )
}

for class_name in CLASS_NAMES:
    observed[f"{class_name} recall"] = recall_score(
        y_true,
        y_pred,
        labels=[class_name],
        average=None,
        zero_division=0
    )[0]

# Stratified bootstrap:
# Resample images within each actual class,
# preserving the original class support.
rng = np.random.default_rng(RANDOM_SEED)

class_indices = {
    class_name: np.where(y_true == class_name)[0]
    for class_name in CLASS_NAMES
}

bootstrap_results = {
    metric: []
    for metric in observed
}

for iteration in range(N_BOOTSTRAPS):

    sampled_indices = np.concatenate([
        rng.choice(
            class_indices[class_name],
            size=len(class_indices[class_name]),
            replace=True
        )
        for class_name in CLASS_NAMES
    ])

    boot_true = y_true[sampled_indices]
    boot_pred = y_pred[sampled_indices]

    bootstrap_results["Accuracy"].append(
        accuracy_score(boot_true, boot_pred)
    )

    bootstrap_results["Balanced accuracy"].append(
        balanced_accuracy_score(boot_true, boot_pred)
    )

    bootstrap_results["Macro F1"].append(
        f1_score(
            boot_true,
            boot_pred,
            labels=CLASS_NAMES,
            average="macro",
            zero_division=0
        )
    )

    for class_name in CLASS_NAMES:
        bootstrap_results[
            f"{class_name} recall"
        ].append(
            recall_score(
                boot_true,
                boot_pred,
                labels=[class_name],
                average=None,
                zero_division=0
            )[0]
        )

# Calculate percentile confidence intervals
rows = []

for metric, values in bootstrap_results.items():

    lower, upper = np.percentile(
        values, [2.5, 97.5]
    )

    rows.append({
        "Metric": metric,
        "Observed": observed[metric],
        "95% CI lower": lower,
        "95% CI upper": upper
    })

ci_df = pd.DataFrame(rows)

print("\n===== BOOTSTRAP RESULTS =====")

display(
    ci_df.style.format({
        "Observed": "{:.4f}",
        "95% CI lower": "{:.4f}",
        "95% CI upper": "{:.4f}"
    })
)

ci_df.to_csv(OUTPUT_PATH, index=False)

print("\nBootstrap resamples:", N_BOOTSTRAPS)
print("Random seed:", RANDOM_SEED)
print("Saved:", OUTPUT_PATH)

print("\nIMPORTANT:")
print(
    "These are image-level bootstrap confidence intervals. "
    "Unknown patient clustering may make them too narrow."
)
print("No model training or inference performed.")
```

```python
import os
import numpy as np
import pandas as pd

from scipy.stats import binomtest
from sklearn.metrics import accuracy_score, f1_score

RESULTS_DIR = "/content/drive/MyDrive/results"

PATH = os.path.join(
    RESULTS_DIR,
    "baseline_vs_patient_balanced_validation_comparison.csv"
)

df = pd.read_csv(PATH)

classes = ["glioma", "meningioma", "pituitary"]

y = df["true_label"].values
base = df["baseline_prediction"].values
bal = df["patient_balanced_prediction"].values

base_correct = (base == y)
bal_correct = (bal == y)

# --------------------------------------------------
# 1. Paired correctness comparison
# --------------------------------------------------

baseline_wrong_balanced_correct = int(
    np.sum((~base_correct) & bal_correct)
)

baseline_correct_balanced_wrong = int(
    np.sum(base_correct & (~bal_correct))
)

both_correct = int(np.sum(base_correct & bal_correct))
both_wrong = int(np.sum((~base_correct) & (~bal_correct)))

print("===== PAIRED PATIENT OUTCOMES =====")
print("Both models correct:", both_correct)
print("Both models incorrect:", both_wrong)
print("Baseline wrong, balanced correct:",
      baseline_wrong_balanced_correct)
print("Baseline correct, balanced wrong:",
      baseline_correct_balanced_wrong)

# Exact two-sided McNemar test
discordant = (
    baseline_wrong_balanced_correct
    + baseline_correct_balanced_wrong
)

if discordant > 0:
    mcnemar_result = binomtest(
        baseline_wrong_balanced_correct,
        n=discordant,
        p=0.5,
        alternative="two-sided"
    )

    print("\n===== EXACT MCNEMAR TEST =====")
    print("Discordant patient pairs:", discordant)
    print("Exact two-sided p-value:",
          round(mcnemar_result.pvalue, 6))
else:
    print("\nNo discordant patient pairs; McNemar test not applicable.")

# --------------------------------------------------
# 2. Observed metric differences
# --------------------------------------------------

base_acc = accuracy_score(y, base)
bal_acc = accuracy_score(y, bal)

base_f1 = f1_score(
    y, base, labels=classes,
    average="macro", zero_division=0
)

bal_f1 = f1_score(
    y, bal, labels=classes,
    average="macro", zero_division=0
)

print("\n===== OBSERVED PAIRED DIFFERENCES =====")
print("Accuracy difference (balanced - baseline):",
      round(bal_acc - base_acc, 4))

print("Macro F1 difference (balanced - baseline):",
      round(bal_f1 - base_f1, 4))

# --------------------------------------------------
# 3. Class-stratified paired bootstrap
# --------------------------------------------------

rng = np.random.default_rng(42)
N_BOOTSTRAP = 10000

class_indices = {
    cls: np.where(y == cls)[0]
    for cls in classes
}

accuracy_differences = []
macro_f1_differences = []

for _ in range(N_BOOTSTRAP):

    sampled_indices = np.concatenate([
        rng.choice(
            class_indices[cls],
            size=len(class_indices[cls]),
            replace=True
        )
        for cls in classes
    ])

    y_b = y[sampled_indices]
    base_b = base[sampled_indices]
    bal_b = bal[sampled_indices]

    accuracy_differences.append(
        accuracy_score(y_b, bal_b)
        - accuracy_score(y_b, base_b)
    )

    macro_f1_differences.append(
        f1_score(
            y_b, bal_b,
            labels=classes,
            average="macro",
            zero_division=0
        )
        -
        f1_score(
            y_b, base_b,
            labels=classes,
            average="macro",
            zero_division=0
        )
    )

def percentile_ci(values):
    return np.percentile(values, [2.5, 97.5])

acc_ci = percentile_ci(accuracy_differences)
f1_ci = percentile_ci(macro_f1_differences)

print("\n===== STRATIFIED PAIRED BOOTSTRAP =====")
print("Bootstrap repetitions:", N_BOOTSTRAP)

print(
    "Accuracy difference 95% CI:",
    np.round(acc_ci, 4)
)

print(
    "Macro F1 difference 95% CI:",
    np.round(f1_ci, 4)
)

# --------------------------------------------------
# 4. Save statistical results
# --------------------------------------------------

stats = pd.DataFrame([
    {
        "metric": "accuracy_difference",
        "observed_difference": bal_acc - base_acc,
        "ci_lower": acc_ci[0],
        "ci_upper": acc_ci[1]
    },
    {
        "metric": "macro_f1_difference",
        "observed_difference": bal_f1 - base_f1,
        "ci_lower": f1_ci[0],
        "ci_upper": f1_ci[1]
    }
])

stats_path = os.path.join(
    RESULTS_DIR,
    "patient_balanced_paired_bootstrap_validation.csv"
)

stats.to_csv(stats_path, index=False)

print("\nSaved:", stats_path)
print("\nStatistical comparison completed.")
```

```python
import os
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.image import ImageDataGenerator

SEED = 2026
BATCH_SIZE = 32
IMAGE_SIZE = (224, 224)
LEARNING_RATE = 0.001

tf.keras.utils.set_random_seed(SEED)

# --------------------------------------------------
# 1. Prepare dataframes
# --------------------------------------------------

train_unweighted_df = train_df[
    ["image_path", "label_name"]
].copy()

train_weighted_df = train_df[
    ["image_path", "label_name", "patient_weight"]
].copy()

val_generator_df = val_df[
    ["image_path", "label_name"]
].copy()

# --------------------------------------------------
# 2. Create identical preprocessing pipelines
# --------------------------------------------------

train_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

val_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0
)

train_unweighted = train_datagen.flow_from_dataframe(
    dataframe=train_unweighted_df,
    x_col="image_path",
    y_col="label_name",
    target_size=IMAGE_SIZE,
    color_mode="rgb",
    classes=CLASS_NAMES,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED,
    interpolation="nearest"
)

train_weighted = train_datagen.flow_from_dataframe(
    dataframe=train_weighted_df,
    x_col="image_path",
    y_col="label_name",
    weight_col="patient_weight",
    target_size=IMAGE_SIZE,
    color_mode="rgb",
    classes=CLASS_NAMES,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=True,
    seed=SEED,
    interpolation="nearest"
)

val_generator = val_datagen.flow_from_dataframe(
    dataframe=val_generator_df,
    x_col="image_path",
    y_col="label_name",
    target_size=IMAGE_SIZE,
    color_mode="rgb",
    classes=CLASS_NAMES,
    class_mode="categorical",
    batch_size=BATCH_SIZE,
    shuffle=False,
    interpolation="nearest"
)

# --------------------------------------------------
# 3. Define the original CNN architecture
# --------------------------------------------------

def build_original_cnn():

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(224, 224, 3)),

        tf.keras.layers.Conv2D(
            32, (3, 3),
            activation="relu",
            padding="same"
        ),
        tf.keras.layers.MaxPooling2D((2, 2)),

        tf.keras.layers.Conv2D(
            64, (3, 3),
            activation="relu",
            padding="same"
        ),
        tf.keras.layers.MaxPooling2D((2, 2)),

        tf.keras.layers.Conv2D(
            128, (3, 3),
            activation="relu",
            padding="same"
        ),
        tf.keras.layers.MaxPooling2D((2, 2)),

        tf.keras.layers.GlobalAveragePooling2D(),

        tf.keras.layers.Dense(
            128,
            activation="relu"
        ),

        tf.keras.layers.Dropout(0.5),

        tf.keras.layers.Dense(
            3,
            activation="softmax"
        )
    ])

    return model


# --------------------------------------------------
# 4. Create identical initial model weights
# --------------------------------------------------

tf.keras.utils.set_random_seed(SEED)

unweighted_model = build_original_cnn()

initial_weights = [
    weight.copy()
    for weight in unweighted_model.get_weights()
]

weighted_model = build_original_cnn()

weighted_model.set_weights(initial_weights)

# Compile both with fresh, identical optimizers
unweighted_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

weighted_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# --------------------------------------------------
# 5. Verify initial weights and optimizer states
# --------------------------------------------------

weights_identical = all(
    np.array_equal(a, b)
    for a, b in zip(
        unweighted_model.get_weights(),
        weighted_model.get_weights()
    )
)

print("===== MODEL VERIFICATION =====")

print("Unweighted parameters:", unweighted_model.count_params())
print("Weighted parameters:", weighted_model.count_params())

print("Initial weights identical:", weights_identical)

print(
    "Unweighted optimizer iterations:",
    int(unweighted_model.optimizer.iterations.numpy())
)

print(
    "Weighted optimizer iterations:",
    int(weighted_model.optimizer.iterations.numpy())
)

assert weights_identical
assert unweighted_model.count_params() == 110147
assert weighted_model.count_params() == 110147

# --------------------------------------------------
# 6. Verify matching batch order and sample weights
# --------------------------------------------------

unweighted_batch = next(train_unweighted)
weighted_batch = next(train_weighted)

x_unweighted, y_unweighted = unweighted_batch
x_weighted, y_weighted, sample_weights = weighted_batch

print("\n===== TRAINING BATCH VERIFICATION =====")

print("Image batch shape:", x_unweighted.shape)
print("Labels batch shape:", y_unweighted.shape)
print("Sample weights shape:", sample_weights.shape)

print(
    "Same images in first batch:",
    np.allclose(x_unweighted, x_weighted)
)

print(
    "Same labels in first batch:",
    np.array_equal(y_unweighted, y_weighted)
)

print(
    "Weighted batch mean weight:",
    round(float(np.mean(sample_weights)), 6)
)

print(
    "Weighted batch minimum:",
    round(float(np.min(sample_weights)), 6)
)

print(
    "Weighted batch maximum:",
    round(float(np.max(sample_weights)), 6)
)

assert np.allclose(x_unweighted, x_weighted)
assert np.array_equal(y_unweighted, y_weighted)

# Reset iterators so training begins at the first batch
train_unweighted.reset()
train_weighted.reset()
val_generator.reset()

print("\n===== VALIDATION GENERATOR =====")
print("Validation images:", val_generator.samples)
print("Class order:", val_generator.class_indices)

print("\nBoth models are prepared.")
print("No training performed.")
print("No existing checkpoint modified.")
print("Test split not accessed.")
```

```python

import os
import h5py
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

mat_dir = "/content/brain_tumor_patient_data/mat_files"
image_dir = "/content/brain_tumor_patient_images_reconstructed"

os.makedirs(image_dir, exist_ok=True)

# Cheng dataset label mapping
label_mapping = {
    1: "meningioma",
    2: "glioma",
    3: "pituitary"
}

reconstructed = []
errors = []
metadata_mismatches = []

print("===== HDF5 IMAGE RECONSTRUCTION =====")

for _, row in tqdm(manifest.iterrows(), total=len(manifest)):
    mat_path = os.path.join(mat_dir, str(row["file"]))
    split = str(row["split"])
    label_name = str(row["label_name"])

    output_dir = os.path.join(image_dir, split, label_name)
    os.makedirs(output_dir, exist_ok=True)

    output_path = os.path.join(
        output_dir,
        os.path.splitext(str(row["file"]))[0] + ".jpg"
    )

    try:
        with h5py.File(mat_path, "r") as f:
            cjdata = f["cjdata"]

            image_array = np.array(cjdata["image"])
            stored_pid = "".join(
                chr(int(x)) for x in np.array(cjdata["PID"]).flatten()
            )
            stored_label = int(np.array(cjdata["label"]).squeeze())

        # Verify original metadata against the saved patient manifest
        if stored_pid != str(row["pid"]):
            metadata_mismatches.append(
                (row["file"], "PID", stored_pid, str(row["pid"]))
            )
            continue

        expected_label = int(row["label"])
        if stored_label != expected_label:
            metadata_mismatches.append(
                (row["file"], "label", stored_label, expected_label)
            )
            continue

        if image_array.shape != (512, 512):
            raise ValueError(f"Unexpected image shape: {image_array.shape}")

        # Preserve original intensity values and convert to 8-bit for JPEG.
        image_array = np.clip(image_array, 0, 255).astype(np.uint8)
        Image.fromarray(image_array).save(
            output_path, format="JPEG", quality=95
        )

        reconstructed.append({
            "file": row["file"],
            "pid": row["pid"],
            "label_name": label_name,
            "split": split,
            "image_path": output_path
        })

    except Exception as exc:
        errors.append((row["file"], str(exc)))

reconstructed_df = pd.DataFrame(reconstructed)

print("\n===== RECONSTRUCTION SUMMARY =====")
print("Successfully reconstructed:", len(reconstructed_df))
print("Errors:", len(errors))
print("Metadata mismatches:", len(metadata_mismatches))

print("\nSplit and class counts:")
if not reconstructed_df.empty:
    display(
        reconstructed_df.groupby(["split", "label_name"])
        .size().unstack(fill_value=0)
    )

print("\nFirst errors:", errors[:5])
print("First metadata mismatches:", metadata_mismatches[:5])
```

```python

import os
import h5py
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

MANIFEST_PATH = "/content/gdrive/MyDrive/results/patient_level_split_manifest.csv"
MAT_DIR = "/content/brain_tumor_patient_data/mat_files"
IMAGE_DIR = "/content/brain_tumor_patient_images_reconstructed"

df = pd.read_csv(MANIFEST_PATH)

errors = []
metadata_mismatches = []
reconstructed = []

for _, row in tqdm(df.iterrows(), total=len(df)):
    filename = row["file"]
    mat_path = os.path.join(MAT_DIR, filename)

    try:
        with h5py.File(mat_path, "r") as f:
            cjdata = f["cjdata"]
            image_array = np.array(cjdata["image"])
            pid_array = np.array(cjdata["PID"]).flatten()
            label_value = int(np.array(cjdata["label"]).squeeze())

        pid = "".join(chr(int(x)) for x in pid_array)

        expected_pid = str(row["pid"])
        expected_label = int(row["label"])

        if pid != expected_pid or label_value != expected_label:
            metadata_mismatches.append(
                (filename, pid, expected_pid,
                 label_value, expected_label)
            )
            continue

        # Original per-image min-max normalization
        image_array = image_array.astype(np.float32)
        img_min = image_array.min()
        img_max = image_array.max()

        if img_max > img_min:
            normalized = (
                (image_array - img_min) /
                (img_max - img_min) * 255.0
            )
        else:
            normalized = np.zeros_like(image_array)

        image_uint8 = normalized.astype(np.uint8)

        # Preserve native resolution; convert grayscale to RGB
        image = Image.fromarray(image_uint8).convert("RGB")

        output_dir = os.path.join(
            IMAGE_DIR,
            str(row["split"]),
            str(row["label_name"])
        )
        os.makedirs(output_dir, exist_ok=True)

        output_path = os.path.join(
            output_dir,
            os.path.splitext(filename)[0] + ".jpg"
        )

        image.save(output_path, format="JPEG", quality=95)

        reconstructed.append({
            "file": filename,
            "split": row["split"],
            "label_name": row["label_name"],
            "shape": image_array.shape
        })

    except Exception as e:
        errors.append((filename, str(e)))

print("\n===== CORRECTED RECONSTRUCTION SUMMARY =====")
print("Expected images:", len(df))
print("Successfully reconstructed:", len(reconstructed))
print("Errors:", len(errors))
print("Metadata mismatches:", len(metadata_mismatches))

result_df = pd.DataFrame(reconstructed)

print("\nSplit and class counts:")
print(pd.crosstab(result_df["split"], result_df["label_name"]))

print("\nOriginal image dimensions:")
print(result_df["shape"].value_counts())

print("\nFirst errors:", errors[:5])
print("First metadata mismatches:", metadata_mismatches[:5])
```

```python
import os
import json
import zipfile
import hashlib
from pathlib import Path

RESULTS_DIR = Path("/content/gdrive/MyDrive/results")
ABLATION_DIR = RESULTS_DIR / "controlled_ablation"

targets = [
    ABLATION_DIR / "unweighted_best.keras",
    ABLATION_DIR / "patient_balanced_best.keras"
]

print("========== CHECKPOINT FILE DETAILS ==========")

for path in targets:
    print(f"\nFILE: {path.name}")

    if not path.exists():
        print("NOT FOUND")
        continue

    stat = path.stat()
    print("Size:", stat.st_size, "bytes")
    print("Modified:", __import__("datetime").datetime.fromtimestamp(
        stat.st_mtime
    ).isoformat())

    sha = hashlib.sha256(path.read_bytes()).hexdigest()
    print("SHA256:", sha)

    if zipfile.is_zipfile(path):
        with zipfile.ZipFile(path) as z:
            print("Archive contents:", z.namelist())

            for name in ["metadata.json", "config.json"]:
                if name in z.namelist():
                    print(f"\n--- {name} ---")
                    content = z.read(name).decode("utf-8")
                    print(content[:3000])
    else:
        print("Not recognised as a Keras ZIP archive.")

print("\n========== RELATED EXPERIMENT RECORDS ==========")

for folder in [RESULTS_DIR, ABLATION_DIR]:
    print(f"\nFolder: {folder}")

    if not folder.exists():
        print("Folder not found.")
        continue

    for path in sorted(folder.iterdir()):
        if path.is_file() and (
            path.suffix.lower() in [".csv", ".json", ".txt", ".log"]
            or "ablation" in path.name.lower()
            or "balanced" in path.name.lower()
            or "unweighted" in path.name.lower()
        ):
            print(path.name, "|", path.stat().st_size, "bytes")
```

```python
import tensorflow as tf
import numpy as np
from pathlib import Path

R = Path("/content/gdrive/MyDrive/results")
A = R / "controlled_ablation"

models = {
    "baseline": R / "patient_level_cnn_best.keras",
    "earlier_balanced": R / "patient_balanced_cnn_best.keras",
    "ablation_unweighted": A / "unweighted_best.keras",
    "ablation_balanced": A / "patient_balanced_best.keras",
}

loaded = {}

for name, path in models.items():
    print(f"\nLoading: {name}")
    model = tf.keras.models.load_model(path, compile=False)
    loaded[name] = model
    print("Parameters:", model.count_params())
    print("Input:", model.input_shape)
    print("Output:", model.output_shape)
    print("Weight arrays:", len(model.get_weights()))

print("\n========== WEIGHT COMPARISONS ==========")

pairs = [
    ("baseline", "earlier_balanced"),
    ("ablation_unweighted", "ablation_balanced"),
    ("baseline", "ablation_unweighted"),
    ("earlier_balanced", "ablation_balanced"),
]

for a, b in pairs:
    wa = loaded[a].get_weights()
    wb = loaded[b].get_weights()

    same_shapes = (
        len(wa) == len(wb)
        and all(x.shape == y.shape for x, y in zip(wa, wb))
    )

    print(f"\n{a} vs {b}")
    print("Same weight shapes:", same_shapes)

    if same_shapes:
        diffs = [
            np.max(np.abs(x.astype(np.float64) - y.astype(np.float64)))
            for x, y in zip(wa, wb)
        ]
        print("Exactly identical:", all(d == 0 for d in diffs))
        print("Maximum absolute difference:", max(diffs))
```

```python
import h5py
from pathlib import Path
from collections import Counter

mat_dir = Path("/content/brain_tumor_patient_data/mat_files")

shape_counts = Counter()
dtype_counts = Counter()
errors = []

for path in sorted(mat_dir.glob("*.mat")):
    try:
        with h5py.File(path, "r") as f:
            image = f["cjdata"]["image"]
            shape_counts[tuple(image.shape)] += 1
            dtype_counts[str(image.dtype)] += 1
    except Exception as e:
        errors.append((path.name, str(e)))

print("Total MAT files:", len(list(mat_dir.glob("*.mat"))))
print("\nImage dimensions:")
for shape, count in shape_counts.items():
    print(shape, ":", count)

print("\nData types:")
for dtype, count in dtype_counts.items():
    print(dtype, ":", count)

print("\nUnreadable files:", len(errors))
if errors:
    print(errors[:10])
```

```python
import h5py
import numpy as np
from pathlib import Path
from PIL import Image
import pandas as pd

MAT_DIR = Path("/content/brain_tumor_patient_data/mat_files")
IMAGE_DIR = Path("/content/brain_tumor_patient_data/reconstructed_images")
IMAGE_DIR.mkdir(parents=True, exist_ok=True)

manifest = pd.read_csv(
    "/content/gdrive/MyDrive/results/patient_level_split_manifest.csv"
)

success = 0
errors = []

for _, row in manifest.iterrows():
    filename = Path(row["file"]).stem
    mat_path = MAT_DIR / f"{filename}.mat"
    output_path = IMAGE_DIR / f"{filename}.jpg"

    try:
        with h5py.File(mat_path, "r") as f:
            image = np.array(f["cjdata"]["image"], dtype=np.float32)

        low, high = image.min(), image.max()

        if high > low:
            image = (image - low) / (high - low) * 255.0
        else:
            image = np.zeros_like(image)

        image = np.clip(image, 0, 255).astype(np.uint8)

        Image.fromarray(image).convert("RGB").save(
            output_path, format="JPEG", quality=95
        )

        success += 1

    except Exception as e:
        errors.append((row["file"], str(e)))

print("Successfully reconstructed:", success)
print("Errors:", len(errors))

if errors:
    print("First errors:", errors[:10])

print("Output directory:", IMAGE_DIR)
print("Saved JPEG count:", len(list(IMAGE_DIR.glob("*.jpg"))))
```

```python
import os
import random
import hashlib
import numpy as np
import pandas as pd
import tensorflow as tf
from pathlib import Path

SEED = 2026
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

tf.keras.utils.set_random_seed(SEED)
try:
    tf.config.experimental.enable_op_determinism()
except Exception:
    pass

EXP_DIR = Path(
    "/content/gdrive/MyDrive/results/controlled_ablation_verified_v2"
)
IMG_DIR = Path("/content/brain_tumor_patient_data/reconstructed_images")

df = pd.read_csv(
    "/content/gdrive/MyDrive/results/patient_level_split_manifest.csv"
)

train_df = df[df["split"] == "train"].copy().reset_index(drop=True)
val_df = df[df["split"] == "val"].copy().reset_index(drop=True)

# Class order: meningioma, glioma, pituitary
class_to_index = {1: 0, 2: 1, 3: 2}

def prepare_arrays(frame):
    paths = np.array([
        str(IMG_DIR / (Path(f).stem + ".jpg"))
        for f in frame["file"]
    ])
    labels = np.array([
        class_to_index[int(label)] for label in frame["label"]
    ], dtype=np.int32)
    return paths, labels

train_paths, train_labels = prepare_arrays(train_df)
val_paths, val_labels = prepare_arrays(val_df)

# Patient-balanced sample weights, training only
patient_counts = train_df.groupby("pid")["file"].transform("count")
train_weights = (1.0 / patient_counts).to_numpy(dtype=np.float32)
train_weights /= train_weights.mean()

def load_image(path, label):
    image_bytes = tf.io.read_file(path)
    image = tf.io.decode_jpeg(image_bytes, channels=3)
    image = tf.image.resize(image, IMG_SIZE)
    image = tf.cast(image, tf.float32) / 255.0
    label = tf.one_hot(label, depth=3)
    return image, label

def make_train_dataset(weighted):
    ds = tf.data.Dataset.from_tensor_slices(
        (train_paths, train_labels, train_weights)
    )
    ds = ds.shuffle(
        len(train_paths), seed=SEED, reshuffle_each_iteration=True
    )

    if weighted:
        ds = ds.map(
            lambda p, y, w: (
                *load_image(p, y), w
            ),
            num_parallel_calls=tf.data.AUTOTUNE,
            deterministic=True
        )
    else:
        ds = ds.map(
            lambda p, y, w: load_image(p, y),
            num_parallel_calls=tf.data.AUTOTUNE,
            deterministic=True
        )

    return ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

def make_validation_dataset():
    ds = tf.data.Dataset.from_tensor_slices(
        (val_paths, val_labels)
    )
    ds = ds.map(
        load_image,
        num_parallel_calls=tf.data.AUTOTUNE,
        deterministic=True
    )
    return ds.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

train_unweighted = make_train_dataset(weighted=False)
train_balanced = make_train_dataset(weighted=True)
validation_ds = make_validation_dataset()

def build_model():
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(224, 224, 3)),
        tf.keras.layers.Conv2D(32, 3, activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(64, 3, activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.Conv2D(128, 3, activation="relu"),
        tf.keras.layers.MaxPooling2D(),
        tf.keras.layers.GlobalAveragePooling2D(),
        tf.keras.layers.Dense(128, activation="relu"),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(3, activation="softmax")
    ])
    return model

# Build one initial model and clone its exact weights
tf.keras.utils.set_random_seed(SEED)
initial_model = build_model()
initial_model(np.zeros((1, 224, 224, 3), dtype=np.float32))

unweighted_model = tf.keras.models.clone_model(initial_model)
balanced_model = tf.keras.models.clone_model(initial_model)

unweighted_model.set_weights(initial_model.get_weights())
balanced_model.set_weights(initial_model.get_weights())

# Independent fresh optimizers
unweighted_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

balanced_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# Save shared initial weights
initial_path = EXP_DIR / "shared_initial.weights.h5"
initial_model.save_weights(initial_path)

# Verify exact equality
w1 = unweighted_model.get_weights()
w2 = balanced_model.get_weights()

identical = (
    len(w1) == len(w2)
    and all(np.array_equal(a, b) for a, b in zip(w1, w2))
)

digest = hashlib.sha256(initial_path.read_bytes()).hexdigest()

print("Unweighted parameters:", unweighted_model.count_params())
print("Balanced parameters:", balanced_model.count_params())
print("Initial weights exactly identical:", identical)
print("Initial weights SHA256:", digest)
print("Initial weights saved:", initial_path.exists())
print("Training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Unweighted optimizer iteration:", int(unweighted_model.optimizer.iterations))
print("Balanced optimizer iteration:", int(balanced_model.optimizer.iterations))
print("Training datasets prepared:", True)
print("Training started: NO")
```

```python
import tensorflow as tf
import numpy as np
from pathlib import Path

def load_unweighted(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, (224, 224))
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

def load_weighted(path, label, weight):
    image = tf.io.read_file(path)
    image = tf.io.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, (224, 224))
    image = tf.cast(image, tf.float32) / 255.0
    return image, label, weight

def prepare_arrays(df):
    paths = np.asarray([
        str(IMAGE_DIR / f"{Path(str(name)).stem}.jpg")
        for name in df["file"]
    ], dtype=str)

    labels = np.asarray([
        class_to_index[int(label)]
        for label in df["label"]
    ])

    one_hot = tf.keras.utils.to_categorical(
        labels, num_classes=3
    ).astype("float32")

    return paths, one_hot

train_paths, train_labels = prepare_arrays(train_df)
val_paths, val_labels = prepare_arrays(val_df)

# Verify weight alignment before constructing the weighted dataset
assert len(train_weights) == len(train_paths), (
    "Sample-weight count does not match training rows."
)

# Check that all image paths exist
assert all(Path(p).is_file() for p in train_paths), (
    "A training image is missing."
)
assert all(Path(p).is_file() for p in val_paths), (
    "A validation image is missing."
)

train_weight_array = np.asarray(
    train_weights, dtype="float32"
)

unweighted_data = tf.data.Dataset.from_tensor_slices(
    (train_paths, train_labels)
)

weighted_data = tf.data.Dataset.from_tensor_slices(
    (train_paths, train_labels, train_weight_array)
)

validation_data = tf.data.Dataset.from_tensor_slices(
    (val_paths, val_labels)
)

train_ds = (
    unweighted_data
    .shuffle(len(train_paths), seed=SEED,
             reshuffle_each_iteration=True)
    .map(load_unweighted, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(32)
    .prefetch(tf.data.AUTOTUNE)
)

balanced_train_ds = (
    weighted_data
    .shuffle(len(train_paths), seed=SEED,
             reshuffle_each_iteration=True)
    .map(load_weighted, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(32)
    .prefetch(tf.data.AUTOTUNE)
)

val_ds = (
    validation_data
    .map(load_unweighted, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(32)
    .prefetch(tf.data.AUTOTUNE)
)

print("Unweighted training images:", len(train_paths))
print("Balanced training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Unweighted batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Balanced batches:", tf.data.experimental.cardinality(balanced_train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
print("Sample weight mean:", float(np.mean(train_weight_array)))
print("Dataset reconstruction successful.")
```

```python
import json
import pandas as pd
import tensorflow as tf
from pathlib import Path

balanced_checkpoint = EXP_DIR / "patient_balanced_best.keras"
balanced_history_path = EXP_DIR / "patient_balanced_history.csv"
balanced_status_path = EXP_DIR / "patient_balanced_run_status.json"

class RecoveryCallback(tf.keras.callbacks.Callback):
    def __init__(self):
        super().__init__()
        self.rows = []
        self.best_loss = float("inf")
        self.best_epoch = None

    def on_epoch_end(self, epoch, logs=None):
        logs = logs or {}

        row = {"epoch": epoch + 1}
        row.update({
            key: float(value)
            for key, value in logs.items()
        })
        self.rows.append(row)

        if logs.get("val_loss", float("inf")) < self.best_loss:
            self.best_loss = float(logs["val_loss"])
            self.best_epoch = epoch + 1

        # Save completed epoch history immediately
        pd.DataFrame(self.rows).to_csv(
            balanced_history_path, index=False
        )

        status = {
            "experiment": "Paired patient-balanced ablation",
            "arm": "patient_balanced",
            "completed_epochs": len(self.rows),
            "best_epoch": self.best_epoch,
            "best_validation_loss": self.best_loss,
            "training_interrupted": False,
            "status": "running",
            "test_set_used_for_training_or_selection": False
        }

        with open(balanced_status_path, "w") as f:
            json.dump(status, f, indent=2)

    def on_train_end(self, logs=None):
        status = {
            "experiment": "Paired patient-balanced ablation",
            "arm": "patient_balanced",
            "completed_epochs": len(self.rows),
            "best_epoch": self.best_epoch,
            "best_validation_loss": self.best_loss,
            "training_interrupted": False,
            "status": "completed",
            "test_set_used_for_training_or_selection": False
        }

        with open(balanced_status_path, "w") as f:
            json.dump(status, f, indent=2)


recovery_callback = RecoveryCallback()

checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath=str(balanced_checkpoint),
    monitor="val_loss",
    mode="min",
    save_best_only=True,
    verbose=1
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    mode="min",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

print("Starting patient-balanced training...")
print("Training images:", len(train_paths))
print("Validation images:", len(val_paths))
print("Maximum epochs: 20")
print("Early stopping patience: 5")

balanced_history = balanced_model.fit(
    balanced_train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=[
        checkpoint_callback,
        early_stopping,
        recovery_callback
    ],
    verbose=1
)

print("\nTraining finished.")
print("Best checkpoint:", balanced_checkpoint)
print("History saved:", balanced_history_path)
print("Run status saved:", balanced_status_path)
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    recall_score,
    confusion_matrix,
    classification_report
)

class_names = ["meningioma", "glioma", "pituitary"]

# Ground truth in the exact validation dataset order
validation_truth = pd.DataFrame({
    "file": val_df["file"].astype(str).values,
    "pid": val_df["pid"].values,
    "true_class": [
        class_to_index[int(x)] for x in val_df["label"]
    ]
})

# Confirm each patient has only one ground-truth class
assert validation_truth.groupby("pid")["true_class"].nunique().max() == 1

all_patient_results = {}

for arm in ["unweighted", "patient_balanced"]:

    print(f"\n{'='*55}")
    print("Evaluating:", arm)

    model = tf.keras.models.load_model(
        EXP_DIR / f"{arm}_best.keras",
        compile=False
    )

    # val_ds is unshuffled and follows val_df order
    probabilities = model.predict(val_ds, verbose=1)

    assert len(probabilities) == len(validation_truth)

    slice_results = validation_truth.copy()

    for i, class_name in enumerate(class_names):
        slice_results[f"{class_name}_probability"] = probabilities[:, i]

    slice_results["predicted_class"] = np.argmax(
        probabilities, axis=1
    )

    # Save slice-level predictions
    slice_results.to_csv(
        EXP_DIR / f"{arm}_validation_slice_predictions.csv",
        index=False
    )

    # Aggregate probabilities per patient
    probability_columns = [
        f"{name}_probability" for name in class_names
    ]

    patient_results = (
        slice_results
        .groupby("pid", as_index=False)
        .agg(
            true_class=("true_class", "first"),
            **{
                col: (col, "mean")
                for col in probability_columns
            }
        )
    )

    patient_probabilities = patient_results[
        probability_columns
    ].to_numpy()

    patient_results["predicted_class"] = np.argmax(
        patient_probabilities, axis=1
    )

    # Save patient-level predictions
    patient_results.to_csv(
        EXP_DIR / f"{arm}_validation_patient_predictions.csv",
        index=False
    )

    y_true = patient_results["true_class"].to_numpy()
    y_pred = patient_results["predicted_class"].to_numpy()

    metrics = {
        "patients": len(patient_results),
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "meningioma_recall": recall_score(
            y_true, y_pred, labels=[0], average=None,
            zero_division=0
        )[0]
    }

    all_patient_results[arm] = patient_results

    print("\nPatient-level metrics:")
    for metric, value in metrics.items():
        print(f"{metric}: {value:.4f}" if isinstance(value, float)
              else f"{metric}: {value}")

    print("\nConfusion matrix (true rows, predicted columns):")
    print(confusion_matrix(y_true, y_pred, labels=[0, 1, 2]))

    print("\nClassification report:")
    print(classification_report(
        y_true,
        y_pred,
        labels=[0, 1, 2],
        target_names=class_names,
        zero_division=0
    ))

print("\nBoth validation prediction files saved.")
print("No test-set images were evaluated.")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import binomtest
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    recall_score
)

# Match both arms by patient ID
u = all_patient_results["unweighted"].copy()
b = all_patient_results["patient_balanced"].copy()

paired = u.merge(
    b,
    on="pid",
    suffixes=("_unweighted", "_balanced"),
    validate="one_to_one"
)

assert len(paired) == 35
assert (
    paired["true_class_unweighted"]
    == paired["true_class_balanced"]
).all()

y = paired["true_class_unweighted"].to_numpy()
pred_u = paired["predicted_class_unweighted"].to_numpy()
pred_b = paired["predicted_class_balanced"].to_numpy()

correct_u = pred_u == y
correct_b = pred_b == y

# Paired discordance table
u_only_correct = int(np.sum(correct_u & ~correct_b))
b_only_correct = int(np.sum(~correct_u & correct_b))
both_correct = int(np.sum(correct_u & correct_b))
both_wrong = int(np.sum(~correct_u & ~correct_b))

print("PAIRED ACCURACY COMPARISON")
print("Both correct:", both_correct)
print("Both wrong:", both_wrong)
print("Only unweighted correct:", u_only_correct)
print("Only balanced correct:", b_only_correct)

# Exact two-sided McNemar test
discordant = u_only_correct + b_only_correct

if discordant:
    mcnemar = binomtest(
        b_only_correct,
        n=discordant,
        p=0.5,
        alternative="two-sided"
    )
    print("\nExact McNemar p-value:", mcnemar.pvalue)
else:
    print("\nNo discordant predictions; McNemar test not informative.")

# Metrics
def calculate_metrics(indices):
    yt = y[indices]
    pu = pred_u[indices]
    pb = pred_b[indices]

    def metric_set(pred):
        return np.array([
            accuracy_score(yt, pred),
            balanced_accuracy_score(yt, pred),
            f1_score(yt, pred, average="macro", zero_division=0),
            recall_score(
                yt, pred, labels=[0], average=None,
                zero_division=0
            )[0]
        ])

    return metric_set(pu), metric_set(pb)

metric_names = [
    "Accuracy",
    "Balanced accuracy",
    "Macro F1",
    "Meningioma recall"
]

observed_u, observed_b = calculate_metrics(
    np.arange(len(y))
)
observed_diff = observed_b - observed_u

# Stratified paired bootstrap: resample patients within each true class
rng = np.random.default_rng(2026)
class_indices = [
    np.where(y == c)[0] for c in range(3)
]

bootstrap_differences = []

for _ in range(10000):
    sampled = np.concatenate([
        rng.choice(idx, size=len(idx), replace=True)
        for idx in class_indices
    ])

    bu, bb = calculate_metrics(sampled)
    bootstrap_differences.append(bb - bu)

bootstrap_differences = np.asarray(bootstrap_differences)

print("\nPAIRED METRIC DIFFERENCES")
print("Difference = patient-balanced minus unweighted")
print("95% percentile bootstrap confidence intervals\n")

for i, name in enumerate(metric_names):
    low, high = np.percentile(
        bootstrap_differences[:, i], [2.5, 97.5]
    )

    print(
        f"{name}: {observed_diff[i]:+.4f} "
        f"(95% CI {low:+.4f} to {high:+.4f})"
    )
```

```python
from pathlib import Path

EXP_DIR = Path(
    "/content/gdrive/MyDrive/results/controlled_ablation_verified_v2"
)

assert EXP_DIR.exists(), "Experiment directory not found."

markdown = """
# Controlled Patient-Balanced Training Ablation

## 1. Objective
Compare ordinary categorical crossentropy with patient-balanced
categorical crossentropy for three-class brain tumour MRI classification.

## 2. Experimental design
- Dataset: Figshare brain tumour MRI
- Classes: Meningioma, Glioma, Pituitary
- Validation cohort: 35 patients
- Architecture: Custom CNN
- Parameters: 110,147
- Input: 224 × 224 × 3
- Optimizer: Adam
- Learning rate: 0.001
- Batch size: 32
- Random seed: 2026
- Shared initial weights: Yes
- Test set used for training or model selection: No

## 3. Training completion

| Arm | Completed epochs | Best epoch | Best validation loss |
|---|---:|---:|---:|
| Unweighted | 19 | 18 | 0.48117 |
| Patient-balanced | 20 | 19 | 0.55401 |

The unweighted arm was interrupted during epoch 20.
The patient-balanced arm completed 20 epochs.

## 4. Patient-level validation results

| Metric | Unweighted | Patient-balanced |
|---|---:|---:|
| Accuracy | 74.29% | 82.86% |
| Balanced accuracy | 75.00% | 83.73% |
| Macro F1 | 71.89% | 83.75% |
| Meningioma recall | 25.00% | 58.33% |

### Confusion matrices

Class order: Meningioma, Glioma, Pituitary.
Rows represent true classes; columns represent predicted classes.

Unweighted:

    [[3, 9, 0],
     [0, 14, 0],
     [0, 0, 9]]

Patient-balanced:

    [[7, 5, 0],
     [1, 13, 0],
     [0, 0, 9]]

## 5. Paired statistical analysis

- Exact two-sided McNemar p-value: 0.453125
- Both models correct: 24 patients
- Both models wrong: 4 patients
- Only unweighted correct: 2 patients
- Only patient-balanced correct: 5 patients
- Bootstrap replicates: 10,000
- Method: Stratified paired patient-level bootstrap
- Confidence level: 95%

| Metric difference | Balanced minus unweighted | 95% CI |
|---|---:|---:|
| Accuracy | +0.0857 | [-0.0571, 0.2000] |
| Balanced accuracy | +0.0873 | [-0.0476, 0.1984] |
| Macro F1 | +0.1186 | [-0.0402, 0.2733] |
| Meningioma recall | +0.3333 | [0.0000, 0.6667] |

## 6. Interpretation

Patient-balanced training showed higher observed patient-level validation
metrics, particularly for meningioma recall. However, the paired McNemar
test did not establish a statistically significant overall accuracy
difference (p = 0.453125).

These findings are promising but exploratory, given the small validation
cohort and wide confidence intervals.

## 7. Limitations

1. Internal validation included only 35 patients.
2. The unweighted arm completed 19 epochs; the balanced arm completed 20.
3. Only one training seed was evaluated.
4. Checkpoint selection used slice-level validation loss, while reported
   comparison metrics are patient-level.
5. The reconstructed image preprocessing pipeline is documented, but
   equivalence to the original historical training pipeline is unverified.
6. The results do not establish clinical effectiveness.

## 8. Conclusion

Patient-balanced training improved the observed validation metrics,
including meningioma recall. However, statistical evidence was
inconclusive. Larger independent patient cohorts and multiple training
seeds are needed for further evaluation.
"""

md_path = EXP_DIR / "consolidated_ablation_report.md"

md_path.write_text(markdown, encoding="utf-8")

print("SUCCESS: Markdown report created.")
print("Location:", md_path)
print("File size:", md_path.stat().st_size, "bytes")
```

```python
import os
import hashlib
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

RESULTS = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

# --------------------------------------------------
# A. RECOMPUTE BDNEURO METRICS
# --------------------------------------------------

print("=" * 70)
print("BDNEURO - INDEPENDENT METRIC VERIFICATION")
print("=" * 70)

df = pd.read_csv(
    os.path.join(RESULTS, "BDNeuro_external_test_predictions.csv")
)

classes = ["glioma", "meningioma", "pituitary"]

y_true = df["true_label"]
y_pred = df["predicted_label"]

print("\nClass counts:")
print(y_true.value_counts().reindex(classes))

print("\nClassification report:")
print(classification_report(
    y_true, y_pred, labels=classes, zero_division=0
))

print("Confusion matrix:")
print(confusion_matrix(y_true, y_pred, labels=classes))

print("Accuracy:", accuracy_score(y_true, y_pred))
print("Balanced accuracy:", balanced_accuracy_score(y_true, y_pred))
print("Macro F1:", f1_score(
    y_true, y_pred, labels=classes,
    average="macro", zero_division=0
))

# --------------------------------------------------
# B. CHECK PROBABILITY CONSISTENCY
# --------------------------------------------------

prob_cols = [
    "prob_glioma",
    "prob_meningioma",
    "prob_pituitary"
]

probabilities = df[prob_cols].to_numpy()

print("\nProbability checks:")
print("All finite:", np.isfinite(probabilities).all())
print("Minimum probability:", probabilities.min())
print("Maximum probability:", probabilities.max())
print(
    "Maximum probability-sum deviation:",
    np.max(np.abs(probabilities.sum(axis=1) - 1))
)

predicted_from_probabilities = np.array(classes)[
    np.argmax(probabilities, axis=1)
]

print(
    "Predicted labels match probability argmax:",
    np.array_equal(
        predicted_from_probabilities,
        y_pred.to_numpy()
    )
)

# --------------------------------------------------
# C. MATCH MODEL SHA-256 TO SAVED CHECKPOINTS
# --------------------------------------------------

print("\n" + "=" * 70)
print("MODEL CHECKPOINT HASH VERIFICATION")
print("=" * 70)

expected_hashes = df["model_sha256"].dropna().unique()

print("Hash recorded in prediction file:", expected_hashes)

for root, dirs, files in os.walk(RESULTS):
    for filename in files:
        if filename.endswith(".keras"):
            path = os.path.join(root, filename)
            sha = hashlib.sha256()

            with open(path, "rb") as f:
                for chunk in iter(lambda: f.read(1024 * 1024), b""):
                    sha.update(chunk)

            digest = sha.hexdigest()

            if digest in expected_hashes:
                print("\nEXACT MODEL MATCH:")
                print(path)
                print(digest)

# --------------------------------------------------
# D. VERIFY PATIENT SPLIT MANIFEST
# --------------------------------------------------

print("\n" + "=" * 70)
print("PATIENT SPLIT MANIFEST")
print("=" * 70)

manifest_path = os.path.join(
    RESULTS, "patient_level_split_manifest.csv"
)

if os.path.exists(manifest_path):
    manifest = pd.read_csv(manifest_path)

    print("Rows:", len(manifest))
    print("Columns:", manifest.columns.tolist())
    print(manifest.head(5).to_string(index=False))
else:
    print("Manifest not found.")
```

```python
import pandas as pd
import os

manifest_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "results/patient_level_split_manifest.csv"
)

df = pd.read_csv(manifest_path)

print("=" * 70)
print("PATIENT-LEVEL SPLIT INTEGRITY AUDIT")
print("=" * 70)

# Image and patient counts
print("\nIMAGE COUNTS BY SPLIT:")
print(df.groupby("split").size())

print("\nUNIQUE PATIENTS BY SPLIT:")
print(df.groupby("split")["pid"].nunique())

# Patient overlap
patient_sets = {
    split: set(group["pid"].unique())
    for split, group in df.groupby("split")
}

splits = list(patient_sets.keys())

print("\nPATIENT OVERLAP CHECK:")
for i in range(len(splits)):
    for j in range(i + 1, len(splits)):
        a, b = splits[i], splits[j]
        overlap = patient_sets[a] & patient_sets[b]

        print(f"{a} vs {b}: {len(overlap)} overlapping patients")

# Label consistency per patient
labels_per_patient = df.groupby("pid")["label"].nunique()

print("\nPATIENT LABEL CONSISTENCY:")
print("Patients with multiple labels:", (labels_per_patient > 1).sum())

# Class counts per split
print("\nCLASS COUNTS BY SPLIT:")
print(pd.crosstab(df["split"], df["label_name"]))

# Overall summary
print("\nOVERALL:")
print("Total images:", len(df))
print("Unique patients:", df["pid"].nunique())
print("Missing patient IDs:", df["pid"].isna().sum())
print("Missing split labels:", df["split"].isna().sum())
```

```python
import zipfile, os, io, collections, time
import imagehash
from PIL import Image

ORIGINAL = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"
EXTERNAL = "/content/drive/MyDrive/Brain_Tumor_CNN/Independent dataset/Brain_Tumor_MRI_Dataset_Final.zip"

EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
TEST_NAMES = {"test", "testing", "test_set", "testset"}

def is_image(p):
    return p.lower().endswith(EXTS)

def parts(p):
    return [x.strip().lower() for x in p.replace("\\", "/").split("/") if x]

def label_of(p):
    for x in parts(p):
        if x in {"glioma", "meningioma", "pituitary", "no_tumor", "notumor"}:
            return "no_tumor" if x == "notumor" else x
    return "unknown"

def phash_from_zip(z, path):
    try:
        with Image.open(io.BytesIO(z.read(path))) as im:
            return int(str(imagehash.phash(im.convert("RGB"))), 16)
    except Exception:
        return None

print("=" * 65)
print("CORRECTED PHASE 2 CROSS-DATASET NEAR-DUPLICATE AUDIT")
print("=" * 65)

with zipfile.ZipFile(ORIGINAL) as oz, zipfile.ZipFile(EXTERNAL) as ez:

    original_files = [p for p in oz.namelist() if is_image(p)]
    external_files = [p for p in ez.namelist() if is_image(p)]

    print("\nOriginal images:", len(original_files))
    print("External archive images:", len(external_files))

    # Identify test images using actual path components
    external_test = [
        p for p in external_files
        if any(x in TEST_NAMES for x in parts(p))
    ]

    print("External test images detected:", len(external_test))

    if not external_test:
        print("\nTEST FOLDER NOT DETECTED.")
        print("External image paths (first 30):")
        for p in external_files[:30]:
            print(p)

        print("\nTop-level external folders:")
        top = collections.Counter(
            parts(p)[0] if parts(p) else "(root)"
            for p in external_files
        )
        for k, v in top.most_common(30):
            print(v, k)

    else:
        print("\nExternal test class counts:")
        for k, v in collections.Counter(label_of(p) for p in external_test).items():
            print(k, v)

        print("\nHashing original images...")
        original_hashes = []
        for i, p in enumerate(original_files, 1):
            h = phash_from_zip(oz, p)
            if h is not None:
                original_hashes.append((p, label_of(p), h))

        print("Successfully hashed original images:", len(original_hashes))

        print("\nComparing external test images...")
        matches = []
        unmatched = []
        failed = 0

        for i, p in enumerate(external_test, 1):
            h = phash_from_zip(ez, p)
            if h is None:
                failed += 1
                continue

            best_distance = 65
            best_match = None

            for op, ol, oh in original_hashes:
                d = (h ^ oh).bit_count()
                if d < best_distance:
                    best_distance = d
                    best_match = (op, ol)

            record = {
                "external_file": p,
                "external_label": label_of(p),
                "closest_original_file": best_match[0],
                "original_label": best_match[1],
                "hamming_distance": best_distance
            }

            if best_distance <= 5:
                matches.append(record)
            else:
                unmatched.append(record)

            if i % 100 == 0:
                print(f"Compared {i}/{len(external_test)}")

        print("\n" + "=" * 65)
        print("FINAL RESULTS (pHash Hamming distance <= 5)")
        print("=" * 65)
        print("External test images:", len(external_test))
        print("Matched candidates:", len(matches))
        print("Unmatched candidates:", len(unmatched))
        print("Images that could not be hashed:", failed)
        print("Label disagreements among matches:",
              sum(m["external_label"] != m["original_label"] for m in matches))

        print("\nMatches by external class:")
        for k, v in collections.Counter(m["external_label"] for m in matches).items():
            print(k, v)

        print("\nFirst 25 candidate matches:")
        for m in matches[:25]:
            print(m)

        print("\nFirst 15 unmatched examples:")
        for m in unmatched[:15]:
            print(m)

print("\nRead-only audit complete. No datasets or checkpoints modified.")
```

```python
import os, zipfile, hashlib

BASE = "/content/drive/MyDrive/Brain_Tumor_CNN"

paths = [
    os.path.join(BASE, "Independent dataset",
                 "Brain_Tumor_MRI_Dataset_Final.zip"),
    os.path.join(BASE, "external_BDNeuro_MRI",
                 "Brain_Tumor_MRI_Dataset_Final.zip")
]

for path in paths:
    print("\nARCHIVE:", path)
    if not os.path.exists(path):
        print("NOT FOUND")
        continue

    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            h.update(chunk)

    print("Size:", os.path.getsize(path))
    print("SHA256:", h.hexdigest())

    with zipfile.ZipFile(path) as z:
        tests = [
            n for n in z.namelist()
            if n.lower().startswith("test/")
            and n.lower().endswith((".jpg", ".jpeg", ".png"))
        ]
        print("Test images:", len(tests))

print("\nBDNEURO RESULT FILES:")
result_dir = os.path.join(BASE, "results")

for filename in [
    "BDNeuro_external_test_predictions.csv",
    "BDNeuro_external_bootstrap_CI.csv"
]:
    p = os.path.join(result_dir, filename)
    print(filename, "| Exists:", os.path.exists(p),
          "| Size:", os.path.getsize(p) if os.path.exists(p) else "-")
```

```python
import os, zipfile, io, hashlib
import pandas as pd
import imagehash
from PIL import Image
from collections import Counter

BASE = "/content/drive/MyDrive/Brain_Tumor_CNN"
ORIGINAL = os.path.join(BASE, "Datasets_MP - 1.zip")
EXTERNAL = os.path.join(
    BASE, "external_BDNeuro_MRI",
    "Brain_Tumor_MRI_Dataset_Final.zip"
)
PREDICTIONS = os.path.join(
    BASE, "results", "BDNeuro_external_test_predictions.csv"
)

EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def phash_from_zip(z, name):
    try:
        img = Image.open(io.BytesIO(z.read(name))).convert("RGB")
        return int(str(imagehash.phash(img)), 16)
    except Exception:
        return None

def norm(s):
    return str(s).replace("\\", "/").strip("/").lower()

def distance(a, b):
    return (a ^ b).bit_count()

print("=" * 65)
print("BDNEURO – EXACT EVALUATED-IMAGE OVERLAP AUDIT")
print("=" * 65)

pred = pd.read_csv(PREDICTIONS)
print("\nPrediction CSV columns:", pred.columns.tolist())
print("Prediction rows:", len(pred))

# Match prediction records to actual archive paths
with zipfile.ZipFile(EXTERNAL) as ez, zipfile.ZipFile(ORIGINAL) as oz:

    external_test = [
        n for n in ez.namelist()
        if norm(n).startswith("test/")
        and n.lower().endswith(EXTS)
    ]

    # Match by full normalized path, then basename if necessary
    path_lookup = {norm(n): n for n in external_test}
    basename_lookup = {}
    for n in external_test:
        basename_lookup.setdefault(os.path.basename(norm(n)), []).append(n)

    matched_records = []
    missing_records = []

    for _, row in pred.iterrows():
        filename = norm(row["filename"])

        archive_path = path_lookup.get(filename)

        if archive_path is None:
            candidates = basename_lookup.get(
                os.path.basename(filename), []
            )
            if len(candidates) == 1:
                archive_path = candidates[0]

        if archive_path is None:
            missing_records.append(row["filename"])
        else:
            matched_records.append((row, archive_path))

    print("\nArchive test images:", len(external_test))
    print("Prediction rows matched to archive:", len(matched_records))
    print("Prediction rows not matched:", len(missing_records))

    if missing_records:
        print("First missing filenames:", missing_records[:10])

    if len(matched_records) != len(pred):
        print("\nSTOP: Not all prediction records were mapped.")
    else:
        print("\nHashing original dataset images...")
        original_hashes = []

        original_files = [
            n for n in oz.namelist()
            if n.lower().endswith(EXTS)
            and (
                "training/" in norm(n)
                or "testing/" in norm(n)
            )
        ]

        for n in original_files:
            h = phash_from_zip(oz, n)
            if h is not None:
                original_hashes.append(
                    (n, norm(n), h)
                )

        print("Original images hashed:", len(original_hashes))

        results = []

        print("\nComparing the exact 771 evaluated images...")

        for i, (row, archive_path) in enumerate(matched_records, 1):
            ext_hash = phash_from_zip(ez, archive_path)

            if ext_hash is None:
                continue

            best_dist = 65
            best_file = None

            for original_path, _, original_hash in original_hashes:
                d = distance(ext_hash, original_hash)
                if d < best_dist:
                    best_dist = d
                    best_file = original_path

            results.append({
                "external_file": archive_path,
                "true_label": str(row["true_label"]).lower(),
                "original_match": best_file,
                "original_label": (
                    norm(best_file).split("/")[-2]
                    if best_file else "unknown"
                ),
                "hamming_distance": best_dist,
                "matched_le_5": best_dist <= 5
            })

            if i % 100 == 0:
                print("Compared:", i)

        df = pd.DataFrame(results)
        matches = df[df["matched_le_5"]]
        disagreements = matches[
            matches["true_label"] != matches["original_label"]
        ]

        print("\n" + "=" * 65)
        print("FINAL BDNEURO EVALUATION OVERLAP RESULTS")
        print("=" * 65)
        print("Successfully compared:", len(df))
        print("pHash matches <= 5:", len(matches))
        print("Unmatched:", len(df) - len(matches))
        print("Label disagreements among matches:", len(disagreements))

        print("\nMatches by true class:")
        print(matches["true_label"].value_counts().to_string())

        print("\nDisagreements:")
        if len(disagreements):
            print(disagreements.to_string(index=False))
        else:
            print("None")

        print("\nDistance distribution:")
        print(df["hamming_distance"].value_counts().sort_index().to_string())

        # Keep detailed results in notebook memory only
        bdneuro_overlap_audit = df.copy()

print("\nRead-only audit complete. No files or checkpoints modified.")
```

```python

# ============================================================
# PHASE 1 – ORIGINAL SPLIT & PREPROCESSING RECOVERY AUDIT
# READ-ONLY: NO TRAINING, NO FILE MODIFICATION
# ============================================================

import os
import glob
import re

ROOT = "/content/drive/MyDrive/Brain_Tumor_CNN"
RESULTS = os.path.join(ROOT, "results")

print("=" * 75)
print("1. SEARCHING PROJECT FILES FOR SPLIT / MANIFEST / PREPROCESSING EVIDENCE")
print("=" * 75)

keywords = [
    "manifest", "split", "train_test", "preprocess",
    "rescale", "normalization", "duplicate", "clean",
    "dataset", "mobilenet", "efficientnet"
]

found_files = []

for current_root, dirs, files in os.walk(ROOT):
    dirs[:] = [d for d in dirs if d not in [
        ".git", "__pycache__", ".ipynb_checkpoints"
    ]]

    for filename in files:
        lower = filename.lower()

        if any(k in lower for k in keywords):
            path = os.path.join(current_root, filename)

            try:
                size = os.path.getsize(path) / (1024 * 1024)
                found_files.append((path, size))
            except OSError:
                pass

for path, size in sorted(found_files):
    print(f"{size:8.2f} MB | {path}")

print("\nTotal matching files:", len(found_files))

print("\n" + "=" * 75)
print("2. SEARCHING EXISTING COLAB EXECUTION HISTORY")
print("=" * 75)

history = globals().get("In", [])

patterns = [
    r"train_test_split",
    r"random_state",
    r"stratify",
    r"ImageDataGenerator",
    r"image_dataset_from_directory",
    r"flow_from_directory",
    r"preprocess_input",
    r"Rescaling",
    r"rescale\s*=",
    r"load_img",
    r"img_to_array",
    r"duplicate",
    r"phash",
    r"224\s*,\s*224",
    r"validation_split"
]

matches = []

for cell_number, cell in enumerate(history):
    if not isinstance(cell, str):
        continue

    matched = [
        p for p in patterns
        if re.search(p, cell, flags=re.IGNORECASE)
    ]

    if matched:
        matches.append((cell_number, matched, cell))

print("Matching history cells:", len(matches))

for cell_number, matched, cell in matches[-35:]:
    print("\n" + "-" * 65)
    print("HISTORY CELL:", cell_number)
    print("MATCHED TOPICS:", ", ".join(matched))

    lines = cell.strip().splitlines()

    # Show relevant lines with nearby context
    selected = set()

    for i, line in enumerate(lines):
        if any(re.search(p, line, re.IGNORECASE) for p in patterns):
            for j in range(max(0, i - 1), min(len(lines), i + 2)):
                selected.add(j)

    for i in sorted(selected):
        print(f"{i+1:04d}: {lines[i][:250]}")

print("\n" + "=" * 75)
print("3. CHECKING FOR POSSIBLE SPLIT MANIFESTS")
print("=" * 75)

manifest_candidates = [
    (path, size) for path, size in found_files
    if any(k in os.path.basename(path).lower()
           for k in ["manifest", "split", "partition"])
]

if manifest_candidates:
    for path, size in manifest_candidates:
        print("CANDIDATE:", path)
else:
    print("No obvious split-manifest filename found.")

print("\n" + "=" * 75)
print("AUDIT COMPLETE – NO FILES WERE MODIFIED")
print("=" * 75)
```

```python

import os
import pandas as pd

ROOT = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

files = [
    "mobilenetv2_training_history.csv",
    "mobilenetv2_finetuned_history.csv",
    "efficientnetb0_training_history.csv"
]

for filename in files:
    path = os.path.join(ROOT, filename)

    print("\n" + "=" * 75)
    print(filename)
    print("=" * 75)

    if not os.path.exists(path):
        print("File not found")
        continue

    df = pd.read_csv(path)

    print("Epochs recorded:", len(df))
    print("Columns:", df.columns.tolist())

    print(df.to_string(index=False))

    for metric in ["val_accuracy", "val_loss"]:
        if metric in df.columns:
            idx = (
                df[metric].idxmax()
                if metric == "val_accuracy"
                else df[metric].idxmin()
            )

            print(
                f"Best {metric}:",
                df.loc[idx, metric],
                "| Epoch:",
                df.loc[idx, "epoch"]
                if "epoch" in df.columns
                else idx + 1
            )

print("\nHISTORY AUDIT COMPLETE – NO MODIFICATIONS")
```

```python

# ============================================================
# CHECKPOINT SELECTION AND SAVING LOGIC AUDIT
# READ-ONLY
# ============================================================

import re
import zipfile
import os
import json

ROOT = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

print("=" * 70)
print("1. SEARCHING NOTEBOOK HISTORY FOR CHECKPOINT SELECTION")
print("=" * 70)

patterns = [
    r"ModelCheckpoint",
    r"save_best_only",
    r"monitor\s*=",
    r"mobilenetv2_finetuned_best",
    r"EarlyStopping",
    r"best_val",
    r"val_loss",
    r"val_accuracy"
]

history = globals().get("In", [])

for i, cell in enumerate(history):
    if not isinstance(cell, str):
        continue

    if any(re.search(p, cell, re.IGNORECASE) for p in patterns):
        print("\n--- HISTORY CELL:", i, "---")
        lines = cell.splitlines()

        for j, line in enumerate(lines):
            if any(re.search(p, line, re.IGNORECASE)
                   for p in patterns):
                start = max(0, j-2)
                end = min(len(lines), j+3)

                for k in range(start, end):
                    print(f"{k+1:04d}: {lines[k][:250]}")
                print()

print("\n" + "=" * 70)
print("2. CHECKING SAVED CHECKPOINT METADATA")
print("=" * 70)

filename = "mobilenetv2_finetuned_best.keras"
path = os.path.join(ROOT, filename)

with zipfile.ZipFile(path, "r") as z:
    print("Checkpoint archive files:")
    for name in z.namelist():
        print(name)

    if "metadata.json" in z.namelist():
        metadata = json.loads(z.read("metadata.json"))
        print("\nKeras metadata:")
        print(json.dumps(metadata, indent=2))

print("\n" + "=" * 70)
print("AUDIT COMPLETE – NO CHECKPOINT MODIFIED")
print("=" * 70)
```

```python

# ============================================================
# PHASE 2 – VISUAL REVIEW OF THE BDNEURO LABEL DISAGREEMENT
# READ-ONLY: OPENS IMAGES DIRECTLY FROM ZIP FILES
# ============================================================

import os
import zipfile
from PIL import Image
import matplotlib.pyplot as plt

PROJECT = "/content/drive/MyDrive/Brain_Tumor_CNN"

original_zip = os.path.join(PROJECT, "Datasets_MP - 1.zip")
external_zip = os.path.join(
    PROJECT,
    "external_BDNeuro_MRI",
    "Brain_Tumor_MRI_Dataset_Final.zip"
)

original_name = "Tr-me_0701.jpg"
external_name = "glioma_test_00050.jpg"

def find_image_in_zip(zip_path, basename):
    with zipfile.ZipFile(zip_path, "r") as z:
        matches = [
            name for name in z.namelist()
            if name.lower().endswith("/" + basename.lower())
            or name.lower() == basename.lower()
        ]

        if not matches:
            raise FileNotFoundError(
                f"{basename} not found in {zip_path}"
            )

        print(f"\nZIP: {os.path.basename(zip_path)}")
        print("Matches:", matches)

        # Use the first exact-basename match.
        with z.open(matches[0]) as f:
            image = Image.open(f).convert("RGB").copy()

        return image, matches[0]

original_image, original_path = find_image_in_zip(
    original_zip, original_name
)

external_image, external_path = find_image_in_zip(
    external_zip, external_name
)

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(original_image)
axes[0].set_title(
    "Original dataset\nLabel: MENINGIOMA\n" + original_path,
    fontsize=9
)
axes[0].axis("off")

axes[1].imshow(external_image)
axes[1].set_title(
    "BDNeuro external test\nLabel: GLIOMA\n" + external_path,
    fontsize=9
)
axes[1].axis("off")

plt.tight_layout()
plt.show()

print("\nIMAGE DETAILS")
print("Original image size:", original_image.size)
print("External image size:", external_image.size)
print("pHash distance previously recorded: 4")
print("\nVisual review only. No files or datasets modified.")
```

```python

# ============================================================
# PHASE 2 – NUMERICAL REVIEW OF LABEL DISAGREEMENT
# READ-ONLY – NO DATASET OR MODEL MODIFICATIONS
# ============================================================

import zipfile
import io
import numpy as np
from PIL import Image
from skimage.metrics import structural_similarity
import imagehash

PROJECT = "/content/drive/MyDrive/Brain_Tumor_CNN"

original_zip = PROJECT + "/Datasets_MP - 1.zip"
external_zip = (
    PROJECT +
    "/external_BDNeuro_MRI/Brain_Tumor_MRI_Dataset_Final.zip"
)

original_name = "Tr-me_0701.jpg"
external_name = "glioma_test_00050.jpg"

def load_image(zip_path, basename):
    with zipfile.ZipFile(zip_path, "r") as z:
        matches = [
            n for n in z.namelist()
            if n.lower().endswith("/" + basename.lower())
        ]
        if not matches:
            raise FileNotFoundError(basename)

        data = z.read(matches[0])
        return Image.open(io.BytesIO(data)).convert("RGB")

img1 = load_image(original_zip, original_name)
img2 = load_image(external_zip, external_name)

# Standardize dimensions for numerical comparison
img1 = img1.resize((512, 512))
img2 = img2.resize((512, 512))

a = np.asarray(img1).astype(np.float32)
b = np.asarray(img2).astype(np.float32)

gray1 = np.asarray(img1.convert("L"))
gray2 = np.asarray(img2.convert("L"))

mae = np.mean(np.abs(a - b))
mse = np.mean((a - b) ** 2)
ssim = structural_similarity(gray1, gray2, data_range=255)

phash1 = imagehash.phash(img1)
phash2 = imagehash.phash(img2)

print("=" * 65)
print("NUMERICAL IMAGE COMPARISON")
print("=" * 65)

print("Original label: MENINGIOMA")
print("External label: GLIOMA")
print("Dimensions:", img1.size, img2.size)
print("pHash distance:", phash1 - phash2)
print("Mean Absolute Error:", round(float(mae), 4))
print("Mean Squared Error:", round(float(mse), 4))
print("SSIM:", round(float(ssim), 6))

print("\nINTERPRETATION")
print("SSIM measures structural similarity, not tumour diagnosis.")
print("Pixel differences can arise from resizing, compression,")
print("contrast changes, cropping, or image transformations.")
print("A similarity score cannot establish the correct tumour label.")

print("\nAUDIT COMPLETE – NO FILES MODIFIED")
```

```python
import os
import hashlib
import random
import shutil
from collections import defaultdict

random.seed(42)

source_path = "/content/brain_tumor_recovery/Datasets_MP - Copy"
clean_path = "/content/brain_tumor_clean_recovered"

extensions = (".jpg", ".jpeg", ".png", ".bmp")
class_images = defaultdict(dict)

# Collect unique images by class using MD5 hashes.
for split in ("Training", "Testing"):
    split_path = os.path.join(source_path, split)

    for label in sorted(os.listdir(split_path)):
        folder = os.path.join(split_path, label)

        if not os.path.isdir(folder):
            continue

        for filename in sorted(os.listdir(folder)):
            if not filename.lower().endswith(extensions):
                continue

            path = os.path.join(folder, filename)

            with open(path, "rb") as f:
                digest = hashlib.md5(f.read()).hexdigest()

            class_images[label].setdefault(digest, path)

# Refuse to overwrite an existing folder.
if os.path.exists(clean_path):
    raise FileExistsError("Recovery folder already exists. Stopping safely.")

train_count = 0
test_count = 0

for label in sorted(class_images):
    unique_paths = list(class_images[label].values())
    random.shuffle(unique_paths)

    split_index = int(len(unique_paths) * 0.8)

    assignments = {
        "Training": unique_paths[:split_index],
        "Testing": unique_paths[split_index:]
    }

    for split_name, paths in assignments.items():
        destination = os.path.join(clean_path, split_name, label)
        os.makedirs(destination, exist_ok=True)

        for index, source in enumerate(paths):
            extension = os.path.splitext(source)[1].lower()
            new_name = f"{label}_{index:05d}{extension}"
            shutil.copy2(source, os.path.join(destination, new_name))

        if split_name == "Training":
            train_count += len(paths)
        else:
            test_count += len(paths)

print("Recovery completed!")
print("Training images:", train_count)
print("Testing images:", test_count)
print("Total unique images:", train_count + test_count)
print("Saved at:", clean_path)
```

```python
import tensorflow as tf
import os

model_path = "/content/drive/MyDrive/Brain_Tumor_CNN/results/brain_tumor_cnn_best.keras"

baseline_model = tf.keras.models.load_model(
    model_path,
    compile=False
)

print("Baseline CNN loaded successfully!")
print("Input shape:", baseline_model.input_shape)
print("Output shape:", baseline_model.output_shape)
```

```python
import tensorflow as tf

clean_path = "/content/drive/MyDrive/Brain_Tumor_CNN/brain_tumor_clean_recovered"

class_names = ["glioma", "meningioma", "notumor", "pituitary"]

test_dataset = tf.keras.utils.image_dataset_from_directory(
    clean_path + "/Testing",
    image_size=(224, 224),
    batch_size=32,
    shuffle=False,
    class_names=class_names
)

print("\nClass order:", test_dataset.class_names)
print("Total test images:", len(test_dataset.file_paths))
```

```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix

y_true = []
for images, labels in test_dataset:
    y_true.extend(labels.numpy())

y_true = np.array(y_true)

# Predict using the same 0–1 normalization used during training.
predictions = baseline_model.predict(
    test_dataset.map(lambda x, y: (x / 255.0, y)),
    verbose=1
)

y_pred = np.argmax(predictions, axis=1)

correct = np.sum(y_true == y_pred)
incorrect = np.sum(y_true != y_pred)

print("\nCorrect predictions:", correct)
print("Incorrect predictions:", incorrect)
print("Accuracy:", round(correct / len(y_true) * 100, 2), "%")

print("\nClassification Report:\n")
print(classification_report(
    y_true,
    y_pred,
    target_names=class_names
))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_true, y_pred))
```

```python
import os
import shutil
import pandas as pd

source_path = os.path.join(clean_path, "Testing")

destination_path = (
    "/content/drive/MyDrive/Brain_Tumor_CNN/"
    "gradcam_recovered_test_images"
)

# Safety check: never overwrite an existing folder
if os.path.exists(destination_path):
    raise FileExistsError(
        "Destination already exists. Nothing was overwritten."
    )

os.makedirs(destination_path)

manifest = []

for class_name in sorted(os.listdir(source_path)):

    class_source = os.path.join(source_path, class_name)

    if not os.path.isdir(class_source):
        continue

    class_destination = os.path.join(
        destination_path, class_name
    )

    os.makedirs(class_destination)

    for filename in sorted(os.listdir(class_source)):

        if not filename.lower().endswith(
            (".jpg", ".jpeg", ".png", ".bmp")
        ):
            continue

        source_file = os.path.join(class_source, filename)
        destination_file = os.path.join(
            class_destination, filename
        )

        shutil.copy2(source_file, destination_file)

        manifest.append({
            "filename": filename,
            "true_class": class_name,
            "image_path": destination_file
        })

manifest_df = pd.DataFrame(manifest)

manifest_path = os.path.join(
    destination_path,
    "recovered_test_image_manifest.csv"
)

manifest_df.to_csv(manifest_path, index=False)

print("Image backup completed!")
print("Images copied:", len(manifest_df))
print("Backup folder:", destination_path)
print("Manifest saved:", manifest_path)
print("\nImages per class:")
print(manifest_df["true_class"].value_counts())
```

```python
import numpy as np

# Prepare the MRI image for MobileNetV2
img_array = np.array(sample_image).astype("float32")
img_array = np.expand_dims(img_array, axis=0)

# Get model prediction
prediction = model_ft.predict(img_array, verbose=0)

predicted_index = np.argmax(prediction[0])
predicted_class = class_names[predicted_index]
confidence = prediction[0][predicted_index] * 100

print("Actual class: glioma")
print("Predicted class:", predicted_class)
print("Confidence:", round(float(confidence), 2), "%")
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# Get MobileNetV2 feature extractor
base_model = model_ft.get_layer("mobilenetv2_1.00_224")

# Create a model that returns feature maps and final predictions
grad_model = tf.keras.models.Model(
    inputs=model_ft.inputs,
    outputs=[
        base_model.get_layer("out_relu").output,
        model_ft.output
    ]
)

# Calculate gradients
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(img_array)
    class_index = tf.argmax(predictions[0])
    class_score = predictions[:, class_index]

gradients = tape.gradient(class_score, feature_maps)

# Calculate Grad-CAM weights
pooled_gradients = tf.reduce_mean(gradients, axis=(0, 1, 2))

# Create heatmap
feature_maps = feature_maps[0]
heatmap = tf.reduce_sum(
    feature_maps * pooled_gradients,
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)
heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

heatmap = heatmap.numpy()

print("Grad-CAM generated successfully!")
print("Heatmap shape:", heatmap.shape)
print("Explained prediction:", predicted_class)
```

```python
# Build a Grad-CAM model using the existing trained weights

base_model = model_ft.get_layer("mobilenetv2_1.00_224")

# Access the final convolutional feature maps
feature_extractor = tf.keras.Model(
    inputs=base_model.input,
    outputs=base_model.get_layer("out_relu").output
)

# Reconnect the original classification layers
inputs = model_ft.input
x = model_ft.get_layer("rescaling")(inputs)

feature_maps = feature_extractor(x)

x = model_ft.get_layer("global_average_pooling2d")(feature_maps)
x = model_ft.get_layer("dense")(x)
x = model_ft.get_layer("dropout")(x, training=False)
predictions = model_ft.get_layer("dense_1")(x)

grad_model = tf.keras.Model(
    inputs=inputs,
    outputs=[feature_maps, predictions]
)

print("Grad-CAM model created successfully!")
```

```python
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(img_array, training=False)

    # Explain the class predicted by the model
    class_index = tf.argmax(predictions[0])
    class_score = predictions[:, class_index]

# Calculate gradients
gradients = tape.gradient(class_score, feature_maps)

# Calculate importance weights
weights = tf.reduce_mean(gradients, axis=(1, 2))

# Generate heatmap
heatmap = tf.reduce_sum(
    feature_maps[0] * weights[0][tf.newaxis, tf.newaxis, :],
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)
heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

heatmap = heatmap.numpy()

print("Grad-CAM heatmap generated!")
print("Heatmap dimensions:", heatmap.shape)
print("Explained class:", predicted_class)
```

```python
from collections import Counter

prediction_counts = Counter()
correct_counts = Counter()

for class_name in class_names:

    folder = os.path.join(test_path, class_name)

    files = sorted([
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    for filename in files:

        path = os.path.join(folder, filename)

        image = load_img(path, target_size=(224, 224))
        image_array = img_to_array(image)
        image_batch = np.expand_dims(image_array, axis=0)

        pred = model_ft.predict(image_batch, verbose=0)[0]
        predicted_class = class_names[np.argmax(pred)]

        prediction_counts[predicted_class] += 1

        if predicted_class == class_name:
            correct_counts[class_name] += 1

print("PREDICTIONS BY MODEL")
print("--------------------")
for name in class_names:
    print(name, ":", prediction_counts[name])

print("\nCORRECT PREDICTIONS BY ACTUAL CLASS")
print("----------------------------------")
for name in class_names:
    print(name, ":", correct_counts[name])
```

```python
import os
import numpy as np
from tensorflow.keras.utils import load_img, img_to_array
from collections import Counter

prediction_counts = Counter()
correct_counts = Counter()

for class_name in class_names:

    folder = os.path.join(test_path, class_name)

    files = sorted([
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    images = []

    for filename in files:
        path = os.path.join(folder, filename)

        image = load_img(path, target_size=(224, 224))
        images.append(img_to_array(image))

    images = np.array(images, dtype="float32")

    # Predict the entire class folder in batches
    predictions = model_ft.predict(images, batch_size=32, verbose=0)

    predicted_indices = np.argmax(predictions, axis=1)

    for index in predicted_indices:
        predicted_class = class_names[index]
        prediction_counts[predicted_class] += 1

    correct_counts[class_name] = np.sum(
        predicted_indices == class_names.index(class_name)
    )

print("PREDICTIONS BY MODEL")
for name in class_names:
    print(name, ":", prediction_counts[name])

print("\nCORRECT PREDICTIONS BY ACTUAL CLASS")
for name in class_names:
    print(name, ":", correct_counts[name])
```

```python
class_to_explain = "meningioma"

folder = os.path.join(
    "/content/drive/MyDrive/Brain_Tumor_CNN/gradcam_recovered_test_images",
    class_to_explain
)

for filename in sorted(os.listdir(folder)):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        path = os.path.join(folder, filename)

        image = tf.keras.utils.load_img(path, target_size=(224, 224))
        image_array = tf.keras.utils.img_to_array(image).astype("float32") / 255.0
        image_array = np.expand_dims(image_array, axis=0)

        scores = model_ft.predict(image_array, verbose=0)[0]
        predicted_class = class_names[np.argmax(scores)]

        if predicted_class == class_to_explain:
            sample_image_path = path
            sample_image = image
            img_array = image_array
            print("Selected image:", filename)
            print("Actual class:", class_to_explain)
            print("Predicted class:", predicted_class)
            print("Confidence:", round(float(np.max(scores)) * 100, 2), "%")
            break
```

```python
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(img_array, training=False)
    class_index = tf.argmax(predictions[0])
    class_score = predictions[:, class_index]

gradients = tape.gradient(class_score, feature_maps)

weights = tf.reduce_mean(gradients, axis=(1, 2))

heatmap = tf.reduce_sum(
    feature_maps[0] * weights[0][tf.newaxis, tf.newaxis, :],
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)
heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

print("Predicted class:", class_names[int(class_index)])
print("Meningioma Grad-CAM heatmap generated!")
print("Heatmap shape:", heatmap.shape)
```

```python
recovered_path = "/content/drive/MyDrive/Brain_Tumor_CNN/gradcam_recovered_test_images"

test_raw = tf.keras.utils.image_dataset_from_directory(
    recovered_path,
    image_size=(224, 224),
    batch_size=32,
    class_names=class_names,
    label_mode="int",
    shuffle=False
)

test_ds = test_raw.map(
    lambda images, labels: (
        tf.cast(images, tf.float32) / 255.0,
        labels
    )
)

all_predictions = []
all_labels = []

for images, labels in test_ds:
    scores = model_ft.predict(images, verbose=0)

    all_predictions.extend(np.argmax(scores, axis=1))
    all_labels.extend(labels.numpy())

all_predictions = np.array(all_predictions)
all_labels = np.array(all_labels)

print("Individual prediction records created!")
print("Total actual labels:", len(all_labels))
print("Total predicted labels:", len(all_predictions))
print("Correct predictions:", np.sum(all_labels == all_predictions))
print("Incorrect predictions:", np.sum(all_labels != all_predictions))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

cm = confusion_matrix(
    all_labels,
    all_predictions
)

print("CONFUSION MATRIX")
print(cm)

print("\nCLASSIFICATION REPORT")
print(classification_report(
    all_labels,
    all_predictions,
    target_names=class_names
))
```

```python
misclassified_indices = np.where(
    all_labels != all_predictions
)[0]

print("Total misclassified images:", len(misclassified_indices))
print("\nMISCLASSIFIED IMAGE DETAILS\n")

for idx in misclassified_indices:
    actual = class_names[all_labels[idx]]
    predicted = class_names[all_predictions[idx]]

    print(
        "Image index:", idx,
        "| Actual:", actual,
        "| Predicted:", predicted
    )
```

```python
# Load the misclassified MRI
sample_image_path = misclassified_path

sample_image = tf.keras.utils.load_img(
    sample_image_path,
    target_size=(224, 224)
)

img_array = tf.keras.utils.img_to_array(
    sample_image
).astype("float32") / 255.0

img_array = np.expand_dims(img_array, axis=0)

# Generate Grad-CAM for the model's predicted class
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(
        img_array, training=False
    )

    class_index = tf.argmax(predictions[0])
    class_score = predictions[:, class_index]

gradients = tape.gradient(class_score, feature_maps)

weights = tf.reduce_mean(gradients, axis=(1, 2))

heatmap = tf.reduce_sum(
    feature_maps[0] * weights[0][tf.newaxis, tf.newaxis, :],
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)
heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

print("Actual class: glioma")
print("Model predicted:", class_names[int(class_index)])
print("Incorrect-prediction Grad-CAM generated!")
```

```python
# Resize the heatmap to MRI dimensions
heatmap_resized = cv2.resize(
    heatmap.numpy(),
    (224, 224)
)

# Apply colour map
heatmap_uint8 = np.uint8(255 * heatmap_resized)

heatmap_colored = cv2.applyColorMap(
    heatmap_uint8,
    cv2.COLORMAP_JET
)

# Original MRI
original_image = np.array(sample_image).astype("uint8")

# Create overlay
overlay = cv2.addWeighted(
    original_image, 0.6,
    heatmap_colored, 0.4,
    0
)

# Display results
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original_image)
plt.title("Actual: Glioma")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(heatmap_resized, cmap="jet")
plt.title("Grad-CAM: Predicted Meningioma")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.title("Incorrect Prediction Overlay")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
import os

results_dir = "/content/drive/MyDrive/Brain_Tumor_CNN/results"

model_paths = {
    "Custom CNN": os.path.join(
        results_dir, "brain_tumor_cnn_best.keras"
    ),
    "MobileNetV2": os.path.join(
        results_dir, "mobilenetv2_brain_tumor_best.keras"
    ),
    "EfficientNetB0": os.path.join(
        results_dir, "efficientnetb0_brain_tumor_best.keras"
    ),
    "Fine-tuned MobileNetV2": os.path.join(
        results_dir, "mobilenetv2_finetuned_best.keras"
    )
}

models = {}

for name, path in model_paths.items():
    print("Loading:", name)
    models[name] = tf.keras.models.load_model(
        path, compile=False
    )
    print("Input:", models[name].input_shape)
    print("Output:", models[name].output_shape)
    print("Loaded successfully!\n")

print("All four models loaded.")
```

```python
for model_name, model in models.items():

    print("\n" + "=" * 60)
    print("MODEL:", model_name)
    print("=" * 60)

    for layer in model.layers:
        print(
            layer.name,
            "|",
            layer.__class__.__name__,
            "|",
            getattr(layer, "output_shape", "Shape not directly available")
        )
```

```python
for model_name in ["MobileNetV2", "EfficientNetB0",
                   "Fine-tuned MobileNetV2"]:

    print("\n" + "=" * 60)
    print("MODEL:", model_name)
    print("=" * 60)

    model = models[model_name]
    backbone = model.layers[2]

    for layer in backbone.layers:
        if isinstance(layer, tf.keras.layers.Conv2D):
            print(layer.name, layer.output.shape)

    print("\nFinal backbone layer:", backbone.layers[-1].name)
    print("Backbone output shape:", backbone.output.shape)
```

```python
model = models["EfficientNetB0"]

print("OUTER MODEL LAYERS:")

for i, layer in enumerate(model.layers):
    print(i, layer.name, layer.__class__.__name__)

print("\nFINDING EFFICIENTNET BACKBONE:")

backbone = next(
    layer for layer in model.layers
    if "efficientnet" in layer.name.lower()
)

print("Backbone found:", backbone.name)
print("Backbone output:", backbone.output.shape)

print("\nLAST CONVOLUTIONAL LAYERS:")

for layer in backbone.layers:
    if isinstance(layer, tf.keras.layers.Conv2D):
        print(layer.name, layer.output.shape)
```

```python
import cv2
import numpy as np

image_path = "/content/drive/MyDrive/Brain_Tumor_CNN/gradcam_recovered_test_images/glioma/glioma_00000.jpg"

# Load and resize the same MRI image
original_image = cv2.imread(image_path)
original_image = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
original_image = cv2.resize(original_image, (224, 224))

# Prepare both input formats
image_0_1 = original_image.astype("float32") / 255.0
image_0_255 = original_image.astype("float32")

# Add batch dimension
input_0_1 = np.expand_dims(image_0_1, axis=0)
input_0_255 = np.expand_dims(image_0_255, axis=0)

print("Image shape:", original_image.shape)
print("0–1 input shape:", input_0_1.shape)
print("0–1 value range:", input_0_1.min(), "to", input_0_1.max())
print("0–255 input shape:", input_0_255.shape)
print("0–255 value range:", input_0_255.min(), "to", input_0_255.max())
```

```python
class_names = ["glioma", "meningioma", "notumor", "pituitary"]

model_inputs = {
    "Custom CNN": input_0_1,
    "MobileNetV2": input_0_1,
    "EfficientNetB0": input_0_255,
    "Fine-tuned MobileNetV2": input_0_1
}

for name, model in models.items():
    prediction = model.predict(model_inputs[name], verbose=0)
    predicted_index = np.argmax(prediction[0])

    print("\nModel:", name)
    print("Predicted class:", class_names[predicted_index])
    print("Confidence:", round(float(prediction[0][predicted_index]) * 100, 2), "%")
```

```python
import tensorflow as tf

custom_model = models["Custom CNN"]
custom_target_layer = custom_model.get_layer("conv2d_5")

grad_model = tf.keras.Model(
    inputs=custom_model.inputs,
    outputs=[custom_target_layer.output, custom_model.output]
)

print("Grad-CAM model created successfully!")
print("Feature map:", grad_model.output[0].shape)
print("Prediction output:", grad_model.output[1].shape)
```

```python
custom_model = models["Custom CNN"]

print("Model type:", type(custom_model).__name__)
print("Model input:", custom_model.input_shape)
print("\nModel summary:")

custom_model.summary()
```

```python
# Create a fresh input
grad_input = tf.keras.Input(shape=(224, 224, 3))

# Pass the image through the existing layers
x = grad_input

for layer in custom_model.layers:
    x = layer(x)

# Create a Grad-CAM model with both outputs
grad_model = tf.keras.Model(
    inputs=grad_input,
    outputs=[
        custom_model.get_layer("conv2d_5").output
        if False else x
    ]
)
```

```python
grad_input = tf.keras.Input(shape=(224, 224, 3))

x = grad_input
feature_map = None

for layer in custom_model.layers:
    x = layer(x)

    if layer.name == "conv2d_5":
        feature_map = x

grad_model = tf.keras.Model(
    inputs=grad_input,
    outputs=[feature_map, x]
)

print("Grad-CAM connection created!")
print("Feature map shape:", grad_model.output[0].shape)
print("Prediction shape:", grad_model.output[1].shape)
```

```python
with tf.GradientTape() as tape:
    feature_maps, predictions = grad_model(input_0_1)

    predicted_class = tf.argmax(predictions[0])
    class_score = predictions[:, predicted_class]

# Calculate gradients
gradients = tape.gradient(class_score, feature_maps)

# Average gradients across spatial dimensions
pooled_gradients = tf.reduce_mean(
    gradients, axis=(0, 1, 2)
)

# Weight feature maps using the gradients
feature_maps = feature_maps[0]

heatmap = tf.reduce_sum(
    feature_maps * pooled_gradients,
    axis=-1
)

# Keep positive activations
heatmap = tf.maximum(heatmap, 0)

# Normalize
heatmap = heatmap / (tf.reduce_max(heatmap) + tf.keras.backend.epsilon())

custom_heatmap = heatmap.numpy()

print("Grad-CAM generated successfully!")
print("Heatmap shape:", custom_heatmap.shape)
print("Minimum value:", custom_heatmap.min())
print("Maximum value:", custom_heatmap.max())
```

```python
import matplotlib.pyplot as plt
import cv2

# Resize heatmap to MRI dimensions
heatmap_resized = cv2.resize(
    custom_heatmap,
    (224, 224)
)

# Display original MRI and heatmap
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original_image)
plt.title("Original MRI")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(original_image)
plt.imshow(heatmap_resized, cmap="jet", alpha=0.4)
plt.title("Custom CNN Grad-CAM")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
mobilenet_model = models["MobileNetV2"]

mobilenet_backbone = mobilenet_model.get_layer(
    "mobilenetv2_1.00_224"
)

mobilenet_target = mobilenet_backbone.get_layer("out_relu")

print("Backbone:", mobilenet_backbone.name)
print("Target layer:", mobilenet_target.name)
print("Target output:", mobilenet_backbone.output.shape)
```

```python
mobilenet_grad_input = tf.keras.Input(shape=(224, 224, 3))

# Pass input through the existing MobileNetV2 backbone
features = mobilenet_backbone(mobilenet_grad_input)

# Pass features through the original classification head
x = features

for layer in mobilenet_model.layers:
    if layer.name == mobilenet_backbone.name:
        continue
    x = layer(x)

mobilenet_grad_model = tf.keras.Model(
    inputs=mobilenet_grad_input,
    outputs=[features, x]
)

print("MobileNetV2 Grad-CAM connection created!")
print("Feature map:", mobilenet_grad_model.output[0].shape)
print("Prediction:", mobilenet_grad_model.output[1].shape)
```

```python
for i, layer in enumerate(mobilenet_model.layers):
    print(
        "Index:", i,
        "| Name:", layer.name,
        "| Type:", type(layer).__name__
    )
```

```python
mobilenet_grad_input = tf.keras.Input(shape=(224, 224, 3))

# Apply the original preprocessing layer
x = mobilenet_model.get_layer("rescaling")(mobilenet_grad_input)

# Extract features from MobileNetV2
features = mobilenet_backbone(x)

# Reconnect the original classification head
x = mobilenet_model.get_layer("global_average_pooling2d")(features)
x = mobilenet_model.get_layer("dense")(x)
x = mobilenet_model.get_layer("dropout")(x)
predictions = mobilenet_model.get_layer("dense_1")(x)

# Create Grad-CAM model
mobilenet_grad_model = tf.keras.Model(
    inputs=mobilenet_grad_input,
    outputs=[features, predictions]
)

print("MobileNetV2 Grad-CAM connection created!")
print("Feature map:", mobilenet_grad_model.output[0].shape)
print("Prediction:", mobilenet_grad_model.output[1].shape)
```

```python
with tf.GradientTape() as tape:
    feature_maps, predictions = mobilenet_grad_model(input_0_1)

    predicted_class = tf.argmax(predictions[0])
    class_score = predictions[:, predicted_class]

gradients = tape.gradient(class_score, feature_maps)

pooled_gradients = tf.reduce_mean(
    gradients, axis=(0, 1, 2)
)

feature_maps = feature_maps[0]

heatmap = tf.reduce_sum(
    feature_maps * pooled_gradients,
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)

heatmap = heatmap / (
    tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
)

mobilenet_heatmap = heatmap.numpy()

print("MobileNetV2 Grad-CAM generated!")
print("Heatmap shape:", mobilenet_heatmap.shape)
print("Minimum:", mobilenet_heatmap.min())
print("Maximum:", mobilenet_heatmap.max())
```

```python
mobilenet_heatmap_resized = cv2.resize(
    mobilenet_heatmap,
    (224, 224)
)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(original_image)
plt.title("Original MRI")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(original_image)
plt.imshow(
    mobilenet_heatmap_resized,
    cmap="jet",
    alpha=0.4
)
plt.title("MobileNetV2 Grad-CAM")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
efficientnet_model = models["EfficientNetB0"]

efficientnet_backbone = efficientnet_model.get_layer(
    "efficientnetb0"
)

efficientnet_target = efficientnet_backbone.get_layer(
    "top_conv"
)

print("Backbone:", efficientnet_backbone.name)
print("Target layer:", efficientnet_target.name)
print("Feature map:", efficientnet_backbone.output.shape)

print("\nClassification head:")
for i, layer in enumerate(efficientnet_model.layers):
    print(i, layer.name, type(layer).__name__)
```

```python
efficientnet_grad_input = tf.keras.Input(
    shape=(224, 224, 3)
)

# EfficientNetB0 backbone
features = efficientnet_backbone(
    efficientnet_grad_input
)

# Original classification head
x = efficientnet_model.get_layer(
    "global_average_pooling2d_1"
)(features)

x = efficientnet_model.get_layer("dense_2")(x)

x = efficientnet_model.get_layer("dropout_1")(x)

predictions = efficientnet_model.get_layer("dense_3")(x)

# Build Grad-CAM model
efficientnet_grad_model = tf.keras.Model(
    inputs=efficientnet_grad_input,
    outputs=[features, predictions]
)

print("EfficientNetB0 Grad-CAM connection created!")
print("Feature map:", efficientnet_grad_model.output[0].shape)
print("Prediction:", efficientnet_grad_model.output[1].shape)
```

```python
with tf.GradientTape() as tape:
    feature_maps, predictions = efficientnet_grad_model(input_0_255)

    predicted_class = tf.argmax(predictions[0])
    class_score = predictions[:, predicted_class]

gradients = tape.gradient(class_score, feature_maps)

pooled_gradients = tf.reduce_mean(
    gradients, axis=(0, 1, 2)
)

feature_maps = feature_maps[0]

heatmap = tf.reduce_sum(
    feature_maps * pooled_gradients,
    axis=-1
)

heatmap = tf.maximum(heatmap, 0)

heatmap = heatmap / (
    tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
)

efficientnet_heatmap = heatmap.numpy()

print("EfficientNetB0 Grad-CAM generated!")
print("Heatmap shape:", efficientnet_heatmap.shape)
print("Minimum:", efficientnet_heatmap.min())
print("Maximum:", efficientnet_heatmap.max())
```

```python
import matplotlib.pyplot as plt
import cv2
import os

# Resize all heatmaps to the same image dimensions
custom_resized = cv2.resize(custom_heatmap, (224, 224))
mobilenet_resized = cv2.resize(mobilenet_heatmap, (224, 224))
efficientnet_resized = cv2.resize(efficientnet_heatmap, (224, 224))
finetuned_resized = cv2.resize(finetuned_heatmap, (224, 224))

# Create comparison figure
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

titles = [
    "Original MRI",
    "Custom CNN",
    "MobileNetV2",
    "EfficientNetB0",
    "Fine-tuned MobileNetV2"
]

images = [
    None,
    custom_resized,
    mobilenet_resized,
    efficientnet_resized,
    finetuned_resized
]

# Original MRI
axes[0, 0].imshow(original_image)
axes[0, 0].set_title(titles[0])

# Four model heatmaps
for ax, title, heatmap in zip(
    [axes[0, 1], axes[0, 2], axes[1, 0], axes[1, 1]],
    titles[1:],
    images[1:]
):
    ax.imshow(original_image)
    ax.imshow(heatmap, cmap="jet", alpha=0.4)
    ax.set_title(title)

# Remove unused sixth panel
axes[1, 2].axis("off")

for ax in axes.flat:
    ax.axis("off")

plt.suptitle(
    "Grad-CAM Comparison Across Four Architectures",
    fontsize=16
)

plt.tight_layout()

# Save comparison
comparison_dir = "/content/drive/MyDrive/Brain_Tumor_CNN/results/gradcam_architecture_comparison"
os.makedirs(comparison_dir, exist_ok=True)

comparison_path = os.path.join(
    comparison_dir,
    "four_architecture_gradcam_comparison.png"
)

plt.savefig(comparison_path, dpi=300, bbox_inches="tight")
plt.show()

print("Comparison saved successfully!")
print(comparison_path)
```

```python
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Output folder
comparison_dir = "/content/drive/MyDrive/Brain_Tumor_CNN/results/gradcam_architecture_comparison"

# Store all four heatmaps
heatmaps = {
    "Custom_CNN": custom_heatmap,
    "MobileNetV2": mobilenet_heatmap,
    "EfficientNetB0": efficientnet_heatmap,
    "Fine_tuned_MobileNetV2": finetuned_heatmap
}

for name, heatmap in heatmaps.items():

    # Save original heatmap values
    np.save(
        os.path.join(comparison_dir, name + "_heatmap.npy"),
        heatmap
    )

    # Resize heatmap
    resized = cv2.resize(heatmap, (224, 224))

    # Create overlay
    colored = cv2.applyColorMap(
        np.uint8(255 * resized),
        cv2.COLORMAP_JET
    )

    colored = cv2.cvtColor(
        colored,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        original_image, 0.6,
        colored, 0.4,
        0
    )

    # Save overlay
    plt.imsave(
        os.path.join(comparison_dir, name + "_overlay.png"),
        overlay
    )

print("All four individual heatmaps and overlays saved!")

print("\nSaved files:")
for filename in sorted(os.listdir(comparison_dir)):
    print(filename)
```

```python
interpretation = """
GRAD-CAM COMPARISON ACROSS FOUR CNN ARCHITECTURES

Objective:
To compare the visual attention patterns of four CNN architectures
(Custom CNN, MobileNetV2, EfficientNetB0, and Fine-tuned MobileNetV2)
using the same glioma MRI image.

Methodology:
Grad-CAM was generated for the predicted glioma class using the
selected final convolutional feature layer of each architecture.
The same MRI image was provided to all four models.

The Custom CNN generated a 56 x 56 feature map, while the three
transfer-learning architectures generated 7 x 7 feature maps.
All heatmaps were resized to 224 x 224 pixels for visualization.

Observations:

1. Custom CNN:
The heatmap showed broad activation, including substantial activation
around the outer image boundaries and some central brain regions.

2. MobileNetV2:
The heatmap showed more localized activation, particularly on the
left and right sides of the brain image.

3. EfficientNetB0:
The heatmap showed relatively weaker activation in much of the central
brain region, with stronger activation around portions of the outer
image boundaries.

4. Fine-tuned MobileNetV2:
The heatmap showed localized activation on both sides of the brain
image, with a pattern visually similar to the original MobileNetV2.

Comparative Discussion:
All four architectures predicted the class glioma for the selected
MRI image. However, their Grad-CAM visualizations showed differences
in spatial activation patterns.

The original and fine-tuned MobileNetV2 models displayed broadly
similar attention patterns. The Custom CNN and EfficientNetB0 showed
more extensive activation around the outer image regions.

Limitations:
This comparison uses one recovered dataset MRI image and should be
considered an illustrative single-image comparison, not a general
conclusion about all images or model architectures.

The heatmaps were independently normalized. Therefore, their color
intensities should not be interpreted as directly comparable absolute
attention strengths.

The architectures also produced feature maps at different resolutions
(56 x 56 for Custom CNN and 7 x 7 for the other three models).

Grad-CAM visualizations indicate regions contributing to model
predictions. They do not establish tumour boundaries, clinical
localization, or causal explanations.

Conclusion:
The four architectures agreed on the predicted glioma class for the
selected MRI image but displayed different Grad-CAM activation
patterns. The comparison demonstrates the usefulness of Grad-CAM
for examining model attention and differences in visual explanations.
"""

interpretation_path = os.path.join(
    comparison_dir,
    "gradcam_architecture_comparison_interpretation.txt"
)

with open(interpretation_path, "w") as file:
    file.write(interpretation)

print("Research interpretation saved successfully!")
print(interpretation_path)
```

```python
# Resize the MRI image to 224 x 224
meningioma_resized = cv2.resize(
    meningioma_image,
    (224, 224)
)

# Prepare pixel values between 0 and 1
meningioma_0_1 = (
    meningioma_resized.astype("float32") / 255.0
)

# Prepare pixel values between 0 and 255
meningioma_0_255 = meningioma_resized.astype("float32")

# Add batch dimension
meningioma_input_0_1 = np.expand_dims(
    meningioma_0_1, axis=0
)

meningioma_input_0_255 = np.expand_dims(
    meningioma_0_255, axis=0
)

print("Image preparation successful!")
print("Input shape:", meningioma_input_0_1.shape)
print("0–1 input range:",
      meningioma_input_0_1.min(),
      "to",
      meningioma_input_0_1.max())

print("0–255 input range:",
      meningioma_input_0_255.min(),
      "to",
      meningioma_input_0_255.max())
```

```python
# Class names in the model's output order
class_names = ["glioma", "meningioma", "notumor", "pituitary"]

# Get predictions from all four models
predictions = {
    "Custom CNN": custom_model.predict(meningioma_input_0_1, verbose=0),
    "MobileNetV2": mobilenet_model.predict(meningioma_input_0_1, verbose=0),
    "EfficientNetB0": efficientnet_model.predict(meningioma_input_0_255, verbose=0),
    "Fine-tuned MobileNetV2": finetuned_model.predict(meningioma_input_0_1, verbose=0)
}

# Display results
for model_name, prediction in predictions.items():
    predicted_index = np.argmax(prediction[0])
    predicted_class = class_names[predicted_index]
    confidence = prediction[0][predicted_index] * 100

    print(f"{model_name}: {predicted_class} ({confidence:.2f}%)")
```

```python
import tensorflow as tf

# Rebuild the Custom CNN as a functional model
custom_input = tf.keras.Input(shape=(224, 224, 3))

x = custom_input
conv_output = None

for layer in custom_model.layers:
    x = layer(x)

    if layer.name == "conv2d_5":
        conv_output = x

custom_gradcam_model = tf.keras.Model(
    inputs=custom_input,
    outputs=[conv_output, x]
)

print("Custom CNN Grad-CAM model created!")
print("Feature map shape:", custom_gradcam_model.output[0].shape)
print("Prediction shape:", custom_gradcam_model.output[1].shape)
```

```python
# MOBILE NET V2
mobilenet_backbone = mobilenet_model.get_layer("mobilenetv2_1.00_224")

mobilenet_gradcam_model = tf.keras.Model(
    inputs=mobilenet_model.input,
    outputs=[
        mobilenet_backbone.get_layer("out_relu").output,
        mobilenet_model.output
    ]
)

# EFFICIENTNET B0
efficientnet_backbone = efficientnet_model.get_layer("efficientnetb0")

efficientnet_gradcam_model = tf.keras.Model(
    inputs=efficientnet_model.input,
    outputs=[
        efficientnet_backbone.get_layer("top_conv").output,
        efficientnet_model.output
    ]
)

# FINE-TUNED MOBILENET V2
finetuned_backbone = finetuned_model.get_layer("mobilenetv2_1.00_224")

finetuned_gradcam_model = tf.keras.Model(
    inputs=finetuned_model.input,
    outputs=[
        finetuned_backbone.get_layer("out_relu").output,
        finetuned_model.output
    ]
)

print("All three Grad-CAM models created successfully!")

print("MobileNetV2:",
      mobilenet_gradcam_model.output[0].shape)

print("EfficientNetB0:",
      efficientnet_gradcam_model.output[0].shape)

print("Fine-tuned MobileNetV2:",
      finetuned_gradcam_model.output[0].shape)
```

```python
def create_nested_gradcam_model(model, backbone_name, target_layer_name):

    inputs = tf.keras.Input(shape=(224, 224, 3))

    backbone = model.get_layer(backbone_name)

    # Create a feature extractor inside the backbone
    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer(target_layer_name).output
    )

    x = inputs
    feature_maps = None

    # Explicitly connect each outer model layer
    for layer in model.layers[1:]:

        if layer.name == backbone_name:
            feature_maps = feature_extractor(x)
            x = layer(x)

        else:
            x = layer(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, x]
    )


# Rebuild MobileNetV2
mobilenet_gradcam_model = create_nested_gradcam_model(
    mobilenet_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

# Rebuild EfficientNetB0
efficientnet_gradcam_model = create_nested_gradcam_model(
    efficientnet_model,
    "efficientnetb0",
    "top_conv"
)

# Rebuild Fine-tuned MobileNetV2
finetuned_gradcam_model = create_nested_gradcam_model(
    finetuned_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

print("All three Grad-CAM connections rebuilt!")

print("MobileNetV2:",
      mobilenet_gradcam_model.output[0].shape)

print("EfficientNetB0:",
      efficientnet_gradcam_model.output[0].shape)

print("Fine-tuned MobileNetV2:",
      finetuned_gradcam_model.output[0].shape)
```

```python
def generate_gradcam(model, image, class_index):

    with tf.GradientTape() as tape:
        feature_maps, predictions = model(image, training=False)
        target_score = predictions[:, class_index]

    gradients = tape.gradient(target_score, feature_maps)

    pooled_gradients = tf.reduce_mean(
        gradients, axis=(0, 1, 2)
    )

    feature_maps = feature_maps[0]

    heatmap = tf.reduce_sum(
        feature_maps * pooled_gradients,
        axis=-1
    )

    heatmap = tf.maximum(heatmap, 0)

    heatmap = heatmap / (
        tf.reduce_max(heatmap) + tf.keras.backend.epsilon()
    )

    return heatmap.numpy()


# Meningioma = class index 1

custom_meningioma_heatmap = generate_gradcam(
    custom_gradcam_model, meningioma_input_0_1, 1
)

mobilenet_meningioma_heatmap = generate_gradcam(
    mobilenet_gradcam_model, meningioma_input_0_1, 1
)

efficientnet_meningioma_heatmap = generate_gradcam(
    efficientnet_gradcam_model, meningioma_input_0_255, 1
)

finetuned_meningioma_heatmap = generate_gradcam(
    finetuned_gradcam_model, meningioma_input_0_1, 1
)

print("All four meningioma Grad-CAM heatmaps generated!")

print("Custom CNN:", custom_meningioma_heatmap.shape)
print("MobileNetV2:", mobilenet_meningioma_heatmap.shape)
print("EfficientNetB0:", efficientnet_meningioma_heatmap.shape)
print("Fine-tuned MobileNetV2:", finetuned_meningioma_heatmap.shape)
```

```python
def create_connected_gradcam_model(model, backbone_name, target_layer_name):

    inputs = tf.keras.Input(shape=(224, 224, 3))

    backbone = model.get_layer(backbone_name)

    # Extract the selected convolutional feature maps
    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer(target_layer_name).output
    )

    x = inputs
    feature_maps = None

    # Pass through preprocessing and the backbone
    for layer in model.layers[1:]:

        if layer.name == backbone_name:
            feature_maps = feature_extractor(x)
            x = feature_maps

        elif feature_maps is not None:
            x = layer(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, x]
    )


# Rebuild all three models
mobilenet_gradcam_model = create_connected_gradcam_model(
    mobilenet_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

efficientnet_gradcam_model = create_connected_gradcam_model(
    efficientnet_model,
    "efficientnetb0",
    "top_conv"
)

finetuned_gradcam_model = create_connected_gradcam_model(
    finetuned_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

print("Connected Grad-CAM models rebuilt!")

# Test whether gradients can flow
for name, model, image in [
    ("MobileNetV2", mobilenet_gradcam_model, meningioma_input_0_1),
    ("EfficientNetB0", efficientnet_gradcam_model, meningioma_input_0_255),
    ("Fine-tuned MobileNetV2", finetuned_gradcam_model, meningioma_input_0_1)
]:

    with tf.GradientTape() as tape:
        feature_maps, predictions = model(image, training=False)
        score = predictions[:, 1]

    gradients = tape.gradient(score, feature_maps)

    print(name, "gradient available:", gradients is not None)
```

```python
# Generate Grad-CAM for meningioma (class index 1)

custom_meningioma_heatmap = generate_gradcam(
    custom_gradcam_model, meningioma_input_0_1, 1
)

mobilenet_meningioma_heatmap = generate_gradcam(
    mobilenet_gradcam_model, meningioma_input_0_1, 1
)

efficientnet_meningioma_heatmap = generate_gradcam(
    efficientnet_gradcam_model, meningioma_input_0_255, 1
)

finetuned_meningioma_heatmap = generate_gradcam(
    finetuned_gradcam_model, meningioma_input_0_1, 1
)

print("All four meningioma Grad-CAM heatmaps generated!")

print("Custom CNN:", custom_meningioma_heatmap.shape)
print("MobileNetV2:", mobilenet_meningioma_heatmap.shape)
print("EfficientNetB0:", efficientnet_meningioma_heatmap.shape)
print("Fine-tuned MobileNetV2:", finetuned_meningioma_heatmap.shape)
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Prepare the original MRI image
base_image = meningioma_resized.copy()

# Store all four heatmaps
heatmaps = {
    "Custom CNN": custom_meningioma_heatmap,
    "MobileNetV2": mobilenet_meningioma_heatmap,
    "EfficientNetB0": efficientnet_meningioma_heatmap,
    "Fine-tuned MobileNetV2": finetuned_meningioma_heatmap
}

# Folder to save results
comparison_dir = "/content/drive/MyDrive/Brain_Tumor_CNN/results/gradcam_architecture_comparison"

os.makedirs(comparison_dir, exist_ok=True)

# Display original MRI
plt.figure(figsize=(5, 5))
plt.imshow(base_image)
plt.title("Original Meningioma MRI")
plt.axis("off")
plt.show()

# Generate and save overlays
plt.figure(figsize=(15, 10))

for i, (name, heatmap) in enumerate(heatmaps.items()):

    # Resize heatmap to MRI dimensions
    resized_heatmap = cv2.resize(
        heatmap, (224, 224)
    )

    # Convert heatmap into colour
    heatmap_uint8 = np.uint8(255 * resized_heatmap)

    colored_heatmap = cv2.applyColorMap(
        heatmap_uint8, cv2.COLORMAP_JET
    )

    colored_heatmap = cv2.cvtColor(
        colored_heatmap, cv2.COLOR_BGR2RGB
    )

    # Blend heatmap with original MRI
    overlay = cv2.addWeighted(
        base_image, 0.6,
        colored_heatmap, 0.4,
        0
    )

    # Save overlay
    safe_name = name.replace(" ", "_")
    save_path = os.path.join(
        comparison_dir,
        f"Meningioma_{safe_name}_overlay.png"
    )

    plt.imsave(save_path, overlay)

    # Display overlay
    plt.subplot(2, 2, i + 1)
    plt.imshow(overlay)
    plt.title(name)
    plt.axis("off")

plt.suptitle(
    "Four-Architecture Grad-CAM Overlays – Meningioma",
    fontsize=15
)

plt.tight_layout()
plt.show()

print("All four meningioma overlays saved successfully!")
print("Saved in:", comparison_dir)
```

```python
# Store the four no-tumor heatmaps
notumor_heatmaps = {
    "Custom CNN": custom_notumor_heatmap,
    "MobileNetV2": mobilenet_notumor_heatmap,
    "EfficientNetB0": efficientnet_notumor_heatmap,
    "Fine-tuned MobileNetV2": finetuned_notumor_heatmap
}

plt.figure(figsize=(12, 10))

for i, (name, heatmap) in enumerate(notumor_heatmaps.items()):

    resized_heatmap = cv2.resize(
        heatmap.astype("float32"),
        (224, 224)
    )

    # Avoid exaggerating nearly zero heatmaps
    if np.max(resized_heatmap) < 1e-6:
        heatmap_uint8 = np.zeros((224, 224), dtype=np.uint8)
    else:
        heatmap_uint8 = np.uint8(
            255 * np.clip(resized_heatmap, 0, 1)
        )

    colored_heatmap = cv2.applyColorMap(
        heatmap_uint8,
        cv2.COLORMAP_JET
    )

    colored_heatmap = cv2.cvtColor(
        colored_heatmap,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        notumor_resized,
        0.6,
        colored_heatmap,
        0.4,
        0
    )

    safe_name = name.replace(" ", "_").replace("-", "_")

    save_path = os.path.join(
        comparison_dir,
        f"NoTumor_{safe_name}_overlay.png"
    )

    plt.imsave(save_path, overlay)

    plt.subplot(2, 2, i + 1)
    plt.imshow(overlay)
    plt.title(name)
    plt.axis("off")

plt.suptitle(
    "Four-Architecture Grad-CAM Overlays – No-Tumor",
    fontsize=15
)

plt.tight_layout()
plt.show()

print("All four no-tumor overlays saved successfully!")
```

```python
pituitary_heatmaps = {
    "Custom CNN": custom_pituitary_heatmap,
    "MobileNetV2": mobilenet_pituitary_heatmap,
    "EfficientNetB0": efficientnet_pituitary_heatmap,
    "Fine-tuned MobileNetV2": finetuned_pituitary_heatmap
}

plt.figure(figsize=(12, 10))

for i, (name, heatmap) in enumerate(pituitary_heatmaps.items()):

    resized_heatmap = cv2.resize(
        heatmap.astype("float32"),
        (224, 224)
    )

    if np.max(resized_heatmap) < 1e-6:
        heatmap_uint8 = np.zeros((224, 224), dtype=np.uint8)
    else:
        heatmap_uint8 = np.uint8(
            255 * np.clip(resized_heatmap, 0, 1)
        )

    colored_heatmap = cv2.applyColorMap(
        heatmap_uint8,
        cv2.COLORMAP_JET
    )

    colored_heatmap = cv2.cvtColor(
        colored_heatmap,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        pituitary_resized,
        0.6,
        colored_heatmap,
        0.4,
        0
    )

    safe_name = name.replace(" ", "_").replace("-", "_")

    save_path = os.path.join(
        comparison_dir,
        f"Pituitary_{safe_name}_overlay.png"
    )

    plt.imsave(save_path, overlay)

    plt.subplot(2, 2, i + 1)
    plt.imshow(overlay)
    plt.title(name)
    plt.axis("off")

plt.suptitle(
    "Four-Architecture Grad-CAM Overlays – Pituitary",
    fontsize=15
)

plt.tight_layout()
plt.show()

print("All four pituitary overlays saved successfully!")
print("Saved in:", comparison_dir)
```

```python
architecture_summary = """
GRAD-CAM ARCHITECTURE COMPARISON
ACROSS FOUR BRAIN MRI CLASSES

Architectures:
1. Custom CNN
2. MobileNetV2
3. EfficientNetB0
4. Fine-tuned MobileNetV2

MRI classes examined:
1. Glioma
2. Meningioma
3. No-tumor
4. Pituitary

METHODOLOGY:
One representative MRI image from each class
was used for comparison across all four models.
Grad-CAM heatmaps and overlays were generated.

OBSERVATIONS:

1. CUSTOM CNN:
The Custom CNN frequently showed activation
around the outer image and brain boundaries.
This pattern was particularly noticeable in
the meningioma, no-tumor and pituitary examples.

2. MOBILENETV2:
MobileNetV2 showed more localized or broad
internal activation patterns depending on
the MRI class.

3. EFFICIENTNETB0:
EfficientNetB0 showed different spatial
activation patterns from MobileNetV2.
Very small Grad-CAM values were observed
for some examples, requiring cautious interpretation.

4. FINE-TUNED MOBILENETV2:
Its activation patterns were broadly similar
to the original MobileNetV2, although some
spatial differences were visible.

OVERALL FINDINGS:
The four architectures showed differences
in their Grad-CAM activation patterns.

MobileNetV2 and Fine-tuned MobileNetV2
often showed broadly similar patterns.

Custom CNN showed noticeable boundary
activation in several examples.

EfficientNetB0 showed different activation
distributions and very small numerical
Grad-CAM responses in some examples.

LIMITATIONS:
Only one representative image per MRI class
was used in this comparison.

The recovered image folder has not been
verified as the original independent test split.

Heatmaps were independently normalized.
Therefore, their colour intensities cannot
be treated as directly comparable strengths.

Grad-CAM does not establish tumour boundaries
or prove that a model uses clinically relevant
features.

Further multi-image validation is required.

CONCLUSION:
Grad-CAM demonstrated differences in visual
explanations across the four architectures
and four MRI classes.

These findings provide an initial qualitative
comparison rather than a general conclusion
about model interpretability or clinical validity.
"""

summary_path = os.path.join(
    comparison_dir,
    "four_architecture_four_class_comparison.txt"
)

with open(summary_path, "w") as file:
    file.write(architecture_summary)

print("Consolidated research observations saved!")
print(summary_path)
```

```python
caption = """
Figure 1. Grad-CAM visualization comparison across four brain MRI
classes and four CNN architectures.

Rows represent the four MRI classes: glioma, meningioma, no-tumor,
and pituitary. Columns represent the Custom CNN, MobileNetV2,
EfficientNetB0, and fine-tuned MobileNetV2 architectures.
Each panel displays a representative MRI image with its corresponding
Grad-CAM heatmap overlay. Warmer colors indicate regions with greater
relative contribution to the selected class prediction.

These visualizations provide a qualitative comparison of model
attention patterns. They are representative examples and should not
be interpreted as validated tumour segmentation or localization.
Heatmaps are independently normalized, so color intensity should
not be directly compared as an absolute measure across models.
"""

caption_path = os.path.join(
    paper_dir, "Figure_1_Caption.txt"
)

with open(caption_path, "w") as f:
    f.write(caption)

print("Figure 1 caption saved successfully!")
print(caption_path)
```

```python
methodology = """
Grad-CAM Methodology

Gradient-weighted Class Activation Mapping (Grad-CAM) was used
to visualize the image regions contributing to the CNN models'
class predictions. For each representative MRI image, gradients
of the selected class score were calculated with respect to the
feature maps of a target convolutional layer.

The gradients were spatially averaged to obtain channel-wise
importance weights. These weights were combined with the
corresponding feature maps, followed by a ReLU operation to
retain positive contributions. The resulting heatmap was
normalized and resized for visualization as an overlay on the
original MRI image.

For the Custom CNN, the target layer was conv2d_5. For MobileNetV2
and fine-tuned MobileNetV2, the target layer was out_relu within
the MobileNetV2 backbone. For EfficientNetB0, the target layer
was top_conv within the EfficientNetB0 backbone.

Grad-CAM visualizations were generated for four classes:
glioma, meningioma, no-tumor, and pituitary. The heatmaps were
used for qualitative comparison of attention patterns across
the four architectures.
"""

methodology_path = os.path.join(
    paper_dir, "GradCAM_Methodology.txt"
)

with open(methodology_path, "w") as f:
    f.write(methodology)

print("Grad-CAM methodology saved successfully!")
print(methodology_path)
```

```python
interpretation = """
Grad-CAM Results and Qualitative Interpretation

Grad-CAM visualizations were generated for representative MRI
images from the glioma, meningioma, no-tumor, and pituitary classes
using four CNN architectures.

The Custom CNN generally showed broader activation patterns,
including responses near the outer boundaries of some MRI images.
The transfer-learning architectures, MobileNetV2 and EfficientNetB0,
showed different spatial activation patterns. Fine-tuned MobileNetV2
often produced patterns broadly similar to MobileNetV2, although
the heatmaps were not identical.

Differences were also observed between classes. For some images,
the models emphasized central or lower brain regions, while other
heatmaps showed broader or weaker responses. In particular, some
EfficientNetB0 heatmaps had very small numerical activation values,
which resulted in weak visual responses.

These observations suggest that the architectures may rely on
different image features when producing their predictions.
However, the visualizations represent qualitative examples rather
than a comprehensive measurement of model attention.

Limitations

Only one representative MRI image per class was used in this
four-class comparison. Therefore, the observed patterns cannot
establish that an architecture consistently focuses on a particular
anatomical region.

The heatmaps were independently normalized. Consequently, their
color intensities should not be interpreted as directly comparable
absolute importance values.

No expert-annotated tumour masks were used to validate the
highlighted regions. Grad-CAM therefore does not establish tumour
boundaries or clinically accurate tumour localization.

Further evaluation across multiple images per class, together with
expert annotations where available, is needed to assess the
consistency and localization quality of the explanations.
"""

interpretation_path = os.path.join(
    paper_dir, "GradCAM_Results_and_Interpretation.txt"
)

with open(interpretation_path, "w") as f:
    f.write(interpretation)

print("Results and interpretation saved successfully!")
print(interpretation_path)
```

```python
import tensorflow as tf
import os

results_dir = os.path.join(base_dir, "results")

model_files = {
    "Custom CNN": "brain_tumor_cnn_best.keras",
    "MobileNetV2": "mobilenetv2_brain_tumor_best.keras",
    "EfficientNetB0": "efficientnetb0_brain_tumor_best.keras",
    "Fine-tuned MobileNetV2": "mobilenetv2_finetuned_best.keras"
}

print("CHECKING SAVED MODEL FILES")
print("=" * 45)

for name, filename in model_files.items():
    path = os.path.join(results_dir, filename)
    print(name, ":", "FOUND" if os.path.exists(path) else "MISSING")

print("\nLOADING MODELS")
print("=" * 45)

custom_model = tf.keras.models.load_model(
    os.path.join(results_dir, model_files["Custom CNN"]),
    compile=False
)

mobilenet_model = tf.keras.models.load_model(
    os.path.join(results_dir, model_files["MobileNetV2"]),
    compile=False
)

efficientnet_model = tf.keras.models.load_model(
    os.path.join(results_dir, model_files["EfficientNetB0"]),
    compile=False
)

finetuned_model = tf.keras.models.load_model(
    os.path.join(results_dir, model_files["Fine-tuned MobileNetV2"]),
    compile=False
)

print("\nAll four models loaded successfully!")
```

```python
import tensorflow as tf

# 1. Custom CNN Grad-CAM model
custom_input = tf.keras.Input(shape=(224, 224, 3))

x = custom_input
custom_conv_output = None

for layer in custom_model.layers:
    x = layer(x)

    if layer.name == "conv2d_5":
        custom_conv_output = x

custom_gradcam_model = tf.keras.Model(
    inputs=custom_input,
    outputs=[custom_conv_output, x]
)

# 2. Function for nested transfer-learning models
def create_connected_gradcam_model(
    model,
    backbone_name,
    target_layer_name
):
    inputs = tf.keras.Input(shape=(224, 224, 3))

    backbone = model.get_layer(backbone_name)

    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer(target_layer_name).output
    )

    x = inputs
    feature_maps = None

    for layer in model.layers[1:]:

        if layer.name == backbone_name:
            feature_maps = feature_extractor(x)
            x = feature_maps

        elif feature_maps is not None:
            x = layer(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, x]
    )


# 3. Build the three nested Grad-CAM models
mobilenet_gradcam_model = create_connected_gradcam_model(
    mobilenet_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

efficientnet_gradcam_model = create_connected_gradcam_model(
    efficientnet_model,
    "efficientnetb0",
    "top_conv"
)

finetuned_gradcam_model = create_connected_gradcam_model(
    finetuned_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

print("All four Grad-CAM models created successfully!")
```

```python
import tensorflow as tf
import numpy as np
import os
from PIL import Image

# Load one representative MRI
image_path = os.path.join(
    recovered_dir,
    "glioma",
    "glioma_00000.jpg"
)

img = Image.open(image_path).convert("RGB")
img = img.resize((224, 224))

image_array = np.array(img).astype("float32") / 255.0
image_batch = np.expand_dims(image_array, axis=0)

# Test each architecture
gradcam_models = {
    "Custom CNN": custom_gradcam_model,
    "MobileNetV2": mobilenet_gradcam_model,
    "EfficientNetB0": efficientnet_gradcam_model,
    "Fine-tuned MobileNetV2": finetuned_gradcam_model
}

print("GRADIENT CONNECTIVITY TEST")
print("=" * 55)

for name, model in gradcam_models.items():

    with tf.GradientTape() as tape:
        feature_maps, predictions = model(
            image_batch,
            training=False
        )

        predicted_class = tf.argmax(
            predictions[0]
        )

        target_score = predictions[:, predicted_class]

    gradients = tape.gradient(
        target_score,
        feature_maps
    )

    print("\nArchitecture:", name)
    print("Feature map shape:", feature_maps.shape)
    print("Prediction shape:", predictions.shape)

    if gradients is None:
        print("Gradient status: FAILED")
    else:
        gradient_array = gradients.numpy()

        print("Gradient status: AVAILABLE")
        print("Gradient contains NaN:", np.isnan(gradient_array).any())
        print("Gradient contains Inf:", np.isinf(gradient_array).any())
        print("Gradient absolute mean:", np.mean(np.abs(gradient_array)))
        print("Gradient maximum:", np.max(np.abs(gradient_array)))
```

```python
import matplotlib.pyplot as plt
import numpy as np

gradcam_models = {
    "Custom CNN": custom_gradcam_model,
    "MobileNetV2": mobilenet_gradcam_model,
    "EfficientNetB0": efficientnet_gradcam_model,
    "Fine-tuned MobileNetV2": finetuned_gradcam_model
}

def generate_gradcam(model, image, class_index):
    with tf.GradientTape() as tape:
        feature_maps, predictions = model(
            image, training=False
        )
        target_score = predictions[:, class_index]

    gradients = tape.gradient(target_score, feature_maps)

    weights = tf.reduce_mean(
        gradients, axis=(0, 1, 2)
    )

    feature_maps = feature_maps[0]

    heatmap = tf.reduce_sum(
        feature_maps * weights, axis=-1
    )

    heatmap = tf.maximum(heatmap, 0)

    maximum = tf.reduce_max(heatmap)

    heatmap = heatmap / (
        maximum + tf.keras.backend.epsilon()
    )

    return heatmap.numpy(), predictions.numpy()[0]


fig, axes = plt.subplots(1, 5, figsize=(20, 5))

axes[0].imshow(image_array)
axes[0].set_title("Original MRI")
axes[0].axis("off")

for i, (name, model) in enumerate(gradcam_models.items()):

    heatmap, predictions = generate_gradcam(
        model, image_batch, class_index=0
    )

    axes[i + 1].imshow(heatmap, cmap="jet")
    axes[i + 1].set_title(
        f"{name}\nMax: {heatmap.max():.3f}"
    )
    axes[i + 1].axis("off")

    print(
        name,
        "| Predicted class:",
        np.argmax(predictions),
        "| Confidence:",
        f"{np.max(predictions) * 100:.2f}%"
    )

plt.suptitle(
    "Fresh Grad-CAM Heatmap Comparison - Glioma",
    fontsize=15
)

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
import numpy as np

def build_graph_connected_gradcam(model, backbone_name, target_layer_name):

    backbone = model.get_layer(backbone_name)

    # Extract the target convolutional feature maps
    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer(target_layer_name).output
    )

    # Connect feature maps to the original model graph
    feature_maps = feature_extractor(backbone.input)

    gradcam_model = tf.keras.Model(
        inputs=model.inputs,
        outputs=[feature_maps, model.output]
    )

    return gradcam_model


# Build MobileNetV2 Grad-CAM model
mobilenet_gradcam_model = build_graph_connected_gradcam(
    mobilenet_model,
    "mobilenetv2_1.00_224",
    "out_relu"
)

print("MobileNetV2 Grad-CAM graph rebuilt!")
```

```python
original_prediction = mobilenet_model(
    image_batch, training=False
).numpy()[0]

_, rebuilt_prediction = mobilenet_gradcam_model(
    image_batch, training=False
)

rebuilt_prediction = rebuilt_prediction.numpy()[0]

print("Original MobileNetV2 probabilities:")
print(np.round(original_prediction, 6))

print("\nRebuilt Grad-CAM probabilities:")
print(np.round(rebuilt_prediction, 6))

difference = np.max(
    np.abs(original_prediction - rebuilt_prediction)
)

print("\nMaximum probability difference:", difference)
print("Prediction preserved:",
      np.argmax(original_prediction) == np.argmax(rebuilt_prediction))
```

```python
import tensorflow as tf
import numpy as np

def rebuild_mobilenet_gradcam(model):

    inputs = tf.keras.Input(shape=(224, 224, 3))

    # Preserve the original input preprocessing
    x = inputs
    feature_maps = None

    backbone = model.get_layer("mobilenetv2_1.00_224")

    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer("out_relu").output
    )

    # Replay the original model layers in order
    for layer in model.layers[1:]:

        if layer.name == "mobilenetv2_1.00_224":

            feature_maps = feature_extractor(x, training=False)
            x = feature_maps

        else:
            try:
                x = layer(x, training=False)
            except TypeError:
                x = layer(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, x]
    )


mobilenet_gradcam_model = rebuild_mobilenet_gradcam(
    mobilenet_model
)

print("MobileNetV2 Grad-CAM reconstruction completed!")
```

```python
# Original model prediction
original_prediction = mobilenet_model(
    image_batch,
    training=False
).numpy()[0]

# Reconstructed Grad-CAM model prediction
_, rebuilt_prediction = mobilenet_gradcam_model(
    image_batch,
    training=False
)

rebuilt_prediction = rebuilt_prediction.numpy()[0]

difference = np.max(
    np.abs(original_prediction - rebuilt_prediction)
)

print("Original MobileNetV2 probabilities:")
print(np.round(original_prediction, 8))

print("\nReconstructed Grad-CAM probabilities:")
print(np.round(rebuilt_prediction, 8))

print("\nMaximum probability difference:", difference)

if difference < 1e-6:
    print("\nSUCCESS: Predictions match!")
else:
    print("\nWARNING: Predictions still differ.")
```

```python
def rebuild_efficientnet_gradcam(model):

    inputs = tf.keras.Input(shape=(224, 224, 3))

    x = inputs
    feature_maps = None

    backbone = model.get_layer("efficientnetb0")

    feature_extractor = tf.keras.Model(
        inputs=backbone.input,
        outputs=backbone.get_layer("top_conv").output
    )

    for layer in model.layers[1:]:

        if layer.name == "efficientnetb0":

            feature_maps = feature_extractor(
                x, training=False
            )

            x = feature_maps

        else:

            try:
                x = layer(x, training=False)
            except TypeError:
                x = layer(x)

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, x]
    )


efficientnet_gradcam_model = rebuild_efficientnet_gradcam(
    efficientnet_model
)

# EfficientNetB0 requires pixel values in the 0–255 range.
efficientnet_input = image_batch * 255.0

original_predictions = efficientnet_model(
    efficientnet_input, training=False
).numpy()

gradcam_features, gradcam_predictions = efficientnet_gradcam_model(
    efficientnet_input, training=False
)

gradcam_predictions = gradcam_predictions.numpy()

print("Original EfficientNetB0:")
print(original_predictions[0])

print("\nReconstructed Grad-CAM model:")
print(gradcam_predictions[0])

difference = np.max(
    np.abs(original_predictions - gradcam_predictions)
)

print("\nMaximum probability difference:", difference)

if difference < 1e-6:
    print("SUCCESS: Predictions match!")
else:
    print("WARNING: Prediction mismatch. We need to investigate.")
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

class_names = ["glioma", "meningioma", "notumor", "pituitary"]

image_records = []

for class_name in class_names:

    class_folder = os.path.join(recovered_dir, class_name)

    image_files = sorted([
        f for f in os.listdir(class_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    for filename in image_files:

        image_path = os.path.join(class_folder, filename)

        img = Image.open(image_path).convert("RGB")
        img = img.resize((224, 224))

        image_array = np.array(img, dtype=np.float32) / 255.0

        image_records.append({
            "image_path": image_path,
            "filename": filename,
            "actual_class": class_name,
            "image_array": image_array
        })


# Prepare batches
X = np.array([record["image_array"] for record in image_records])

# Predict using Fine-tuned MobileNetV2
predictions = finetuned_model.predict(
    X,
    batch_size=32,
    verbose=1
)

predicted_indices = np.argmax(predictions, axis=1)

# Create prediction summary
for i, record in enumerate(image_records):

    predicted_class = class_names[predicted_indices[i]]

    record["predicted_class"] = predicted_class
    record["confidence_percent"] = float(
        predictions[i][predicted_indices[i]] * 100
    )
    record["correct"] = (
        record["actual_class"] == predicted_class
    )

prediction_df = pd.DataFrame([
    {
        "filename": r["filename"],
        "image_path": r["image_path"],
        "actual_class": r["actual_class"],
        "predicted_class": r["predicted_class"],
        "confidence_percent": round(r["confidence_percent"], 2),
        "correct": r["correct"]
    }
    for r in image_records
])

print("\nPREDICTION SUMMARY")
print("=" * 55)

print(prediction_df.groupby("actual_class")["correct"].agg(
    ["count", "sum"]
))

print("\nCORRECT AND INCORRECT COUNTS")
print("=" * 55)

for class_name in class_names:

    class_data = prediction_df[
        prediction_df["actual_class"] == class_name
    ]

    correct_count = int(class_data["correct"].sum())
    incorrect_count = int((~class_data["correct"]).sum())

    print(
        f"{class_name}: "
        f"{correct_count} correct | "
        f"{incorrect_count} incorrect"
    )

print("\nFIRST INCORRECT PREDICTIONS")
print("=" * 55)

display(
    prediction_df[prediction_df["correct"] == False].head(15)
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from PIL import Image
import os

# Select one informative misclassified image
test_filename = "glioma_00219.jpg"

test_path = os.path.join(
    recovered_dir,
    "glioma",
    test_filename
)

# Load MRI
test_img = Image.open(test_path).convert("RGB")
test_img = test_img.resize((224, 224))

test_array = np.array(test_img, dtype=np.float32) / 255.0
test_batch = np.expand_dims(test_array, axis=0)

# Correct preprocessing for each model
gradcam_inputs = {
    "Custom CNN": test_batch,
    "MobileNetV2": test_batch,
    "EfficientNetB0": test_batch * 255.0,
    "Fine-tuned MobileNetV2": test_batch
}

gradcam_models = {
    "Custom CNN": custom_gradcam_model,
    "MobileNetV2": mobilenet_gradcam_model,
    "EfficientNetB0": efficientnet_gradcam_model,
    "Fine-tuned MobileNetV2": finetuned_gradcam_model
}

model_objects = {
    "Custom CNN": custom_model,
    "MobileNetV2": mobilenet_model,
    "EfficientNetB0": efficientnet_model,
    "Fine-tuned MobileNetV2": finetuned_model
}

heatmaps = {}
prediction_results = {}

for model_name in gradcam_models:

    gradcam_model = gradcam_models[model_name]
    model_input = gradcam_inputs[model_name]

    # Calculate gradients for the model's predicted class
    with tf.GradientTape() as tape:

        feature_maps, predictions_output = gradcam_model(
            model_input,
            training=False
        )

        predicted_index = tf.argmax(
            predictions_output[0]
        )

        target_score = predictions_output[:, predicted_index]

    gradients = tape.gradient(
        target_score,
        feature_maps
    )

    # Grad-CAM calculation
    weights = tf.reduce_mean(
        gradients,
        axis=(1, 2),
        keepdims=True
    )

    cam = tf.reduce_sum(
        weights * feature_maps,
        axis=-1
    )

    cam = tf.nn.relu(cam)

    cam = cam[0].numpy()

    # Normalize heatmap
    if np.max(cam) > 0:
        cam = cam / np.max(cam)

    heatmaps[model_name] = cam

    prediction_results[model_name] = {
        "predicted_class": class_names[int(predicted_index.numpy())],
        "confidence": float(
            predictions_output[0, predicted_index].numpy() * 100
        ),
        "gradient_mean": float(
            tf.reduce_mean(tf.abs(gradients)).numpy()
        )
    }

# Display original MRI and all four heatmaps
fig, axes = plt.subplots(1, 5, figsize=(20, 5))

axes[0].imshow(test_array)
axes[0].set_title("Original MRI\nActual: Glioma")
axes[0].axis("off")

for i, model_name in enumerate(gradcam_models, start=1):

    axes[i].imshow(test_array)
    axes[i].imshow(
        heatmaps[model_name],
        cmap="jet",
        alpha=0.45,
        extent=(0, 224, 224, 0)
    )

    result = prediction_results[model_name]

    axes[i].set_title(
        f"{model_name}\n"
        f"Pred: {result['predicted_class']}\n"
        f"Confidence: {result['confidence']:.2f}%"
    )

    axes[i].axis("off")

plt.tight_layout()
plt.show()

print("\nGRAD-CAM NUMERICAL SUMMARY")
for name, result in prediction_results.items():
    print(
        name,
        "| Prediction:", result["predicted_class"],
        "| Confidence:", round(result["confidence"], 2),
        "| Mean gradient:", result["gradient_mean"]
    )
```

```python
true_class_index = 0

true_class_heatmaps = {}

for model_name in gradcam_models:

    gradcam_model = gradcam_models[model_name]
    model_input = gradcam_inputs[model_name]

    with tf.GradientTape() as tape:

        feature_maps, predictions_output = gradcam_model(
            model_input,
            training=False
        )

        target_score = predictions_output[:, true_class_index]

    gradients = tape.gradient(
        target_score,
        feature_maps
    )

    weights = tf.reduce_mean(
        gradients,
        axis=(1, 2),
        keepdims=True
    )

    cam = tf.reduce_sum(
        weights * feature_maps,
        axis=-1
    )

    cam = tf.nn.relu(cam)[0].numpy()

    if np.max(cam) > 0:
        cam = cam / np.max(cam)

    true_class_heatmaps[model_name] = cam

# Visualize all four explanations for the same target class
fig, axes = plt.subplots(1, 5, figsize=(20, 5))

axes[0].imshow(test_array)
axes[0].set_title("Original MRI\nTrue class: Glioma")
axes[0].axis("off")

for i, model_name in enumerate(gradcam_models, start=1):

    axes[i].imshow(test_array)

    axes[i].imshow(
        true_class_heatmaps[model_name],
        cmap="jet",
        alpha=0.45,
        interpolation="bilinear",
        extent=(0, 224, 224, 0)
    )

    axes[i].set_title(
        f"{model_name}\nTrue-class Grad-CAM"
    )

    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python
def rebuild_logit_gradcam_model(
    model,
    backbone_name=None,
    target_layer_name=None,
    custom=False
):

    inputs = tf.keras.Input(shape=(224, 224, 3))

    x = inputs
    feature_maps = None

    # Final original Dense layer (softmax classifier)
    final_dense = model.layers[-1]

    if custom:

        # Replay all layers BEFORE the final softmax layer
        for layer in model.layers[:-1]:

            try:
                x = layer(x, training=False)
            except TypeError:
                x = layer(x)

            if layer.name == "conv2d_5":
                feature_maps = x

    else:

        backbone = model.get_layer(backbone_name)

        feature_extractor = tf.keras.Model(
            inputs=backbone.input,
            outputs=backbone.get_layer(target_layer_name).output
        )

        # Replay outer layers, excluding final softmax Dense
        for layer in model.layers[1:-1]:

            if layer.name == backbone_name:

                feature_maps = feature_extractor(
                    x, training=False
                )

                x = feature_maps

            else:

                try:
                    x = layer(x, training=False)
                except TypeError:
                    x = layer(x)

    # Create a linear Dense layer for pre-softmax logits
    logits_layer = tf.keras.layers.Dense(
        units=4,
        activation=None,
        name=model.name + "_logits"
    )

    logits = logits_layer(x)

    # Transfer original classifier weights
    logits_layer.set_weights(final_dense.get_weights())

    return tf.keras.Model(
        inputs=inputs,
        outputs=[feature_maps, logits]
    )


# Rebuild all four models
custom_logit_gradcam = rebuild_logit_gradcam_model(
    custom_model,
    custom=True
)

mobilenet_logit_gradcam = rebuild_logit_gradcam_model(
    mobilenet_model,
    backbone_name="mobilenetv2_1.00_224",
    target_layer_name="out_relu"
)

efficientnet_logit_gradcam = rebuild_logit_gradcam_model(
    efficientnet_model,
    backbone_name="efficientnetb0",
    target_layer_name="top_activation"
)

finetuned_logit_gradcam = rebuild_logit_gradcam_model(
    finetuned_model,
    backbone_name="mobilenetv2_1.00_224",
    target_layer_name="out_relu"
)

print("SUCCESS: All four logit-based Grad-CAM models created!")
```

```python
logit_models = {
    "Custom CNN": custom_logit_gradcam,
    "MobileNetV2": mobilenet_logit_gradcam,
    "EfficientNetB0": efficientnet_logit_gradcam,
    "Fine-tuned MobileNetV2": finetuned_logit_gradcam
}

original_models = {
    "Custom CNN": custom_model,
    "MobileNetV2": mobilenet_model,
    "EfficientNetB0": efficientnet_model,
    "Fine-tuned MobileNetV2": finetuned_model
}

model_inputs = {
    "Custom CNN": test_batch,
    "MobileNetV2": test_batch,
    "EfficientNetB0": test_batch * 255.0,
    "Fine-tuned MobileNetV2": test_batch
}

print("LOGIT MODEL VERIFICATION")
print("=" * 60)

for name in logit_models:

    model_input = model_inputs[name]

    _, logits = logit_models[name](
        model_input,
        training=False
    )

    reconstructed_probabilities = tf.nn.softmax(
        logits, axis=-1
    ).numpy()

    original_probabilities = original_models[name](
        model_input,
        training=False
    ).numpy()

    difference = np.max(
        np.abs(
            original_probabilities -
            reconstructed_probabilities
        )
    )

    print(f"\n{name}")
    print("Original probabilities:", original_probabilities[0])
    print("Logit-derived probabilities:", reconstructed_probabilities[0])
    print("Maximum difference:", difference)

    if difference < 1e-5:
        print("PASS")
    else:
        print("WARNING: Investigate mismatch")
```

```python
print("PROBABILITY vs LOGIT GRADIENT COMPARISON")
print("=" * 65)

for name in logit_models:

    model_input = model_inputs[name]
    probability_model = gradcam_models[name]
    logit_model = logit_models[name]

    # Probability-based gradient
    with tf.GradientTape() as tape:

        features_p, probabilities = probability_model(
            model_input, training=False
        )

        target_probability = probabilities[:, 0]

    probability_gradients = tape.gradient(
        target_probability, features_p
    )

    # Logit-based gradient
    with tf.GradientTape() as tape:

        features_l, logits = logit_model(
            model_input, training=False
        )

        target_logit = logits[:, 0]

    logit_gradients = tape.gradient(
        target_logit, features_l
    )

    probability_mean = float(
        tf.reduce_mean(
            tf.abs(probability_gradients)
        ).numpy()
    )

    logit_mean = float(
        tf.reduce_mean(
            tf.abs(logit_gradients)
        ).numpy()
    )

    print(f"\n{name}")
    print("Probability gradient:", probability_mean)
    print("Logit gradient:", logit_mean)

    if probability_mean > 0:
        print(
            "Gradient magnitude ratio:",
            logit_mean / probability_mean
        )

    print(
        "Logit gradients finite:",
        np.isfinite(logit_gradients.numpy()).all()
    )
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
from PIL import Image

# Take the first selected MRI
row = selected_23_df.iloc[0]

image_path = row["image_path"]
actual_class = row["actual_class"]
filename = row["filename"]

# Load and resize the MRI
original_img = Image.open(image_path).convert("RGB").resize((224, 224))
original_array = np.array(original_img).astype("float32")

# Prepare model inputs
test_batch = np.expand_dims(original_array / 255.0, axis=0)

model_inputs_for_image = {
    "Custom CNN": test_batch,
    "MobileNetV2": test_batch,
    "EfficientNetB0": test_batch * 255.0,
    "Fine-tuned MobileNetV2": test_batch
}

# Find the actual class index
class_index = class_names.index(actual_class)

# Generate logit-based Grad-CAM heatmaps
first_image_heatmaps = {}

for name, logit_model in logit_models.items():

    model_input = model_inputs_for_image[name]

    with tf.GradientTape() as tape:
        feature_maps, logits = logit_model(
            model_input,
            training=False
        )
        target_logit = logits[:, class_index]

    gradients = tape.gradient(target_logit, feature_maps)

    weights = tf.reduce_mean(
        gradients,
        axis=(1, 2),
        keepdims=True
    )

    cam = tf.reduce_sum(weights * feature_maps, axis=-1)
    cam = tf.nn.relu(cam)[0].numpy()

    if np.max(cam) > 0:
        cam = cam / np.max(cam)

    first_image_heatmaps[name] = cam

print("MRI:", filename)
print("Actual class:", actual_class)
print("Heatmaps generated:", len(first_image_heatmaps))

# Display the results
fig, axes = plt.subplots(1, 5, figsize=(22, 5))

axes[0].imshow(original_array.astype("uint8"))
axes[0].set_title("Original MRI")

for i, (name, cam) in enumerate(first_image_heatmaps.items()):

    axes[i + 1].imshow(original_array.astype("uint8"))
    axes[i + 1].imshow(
        cv2.resize(cam, (224, 224)),
        cmap="jet",
        alpha=0.45
    )
    axes[i + 1].set_title(name)

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import cv2
from PIL import Image

total_heatmaps = 0

for index, row in selected_23_df.iterrows():

    image_path = row["image_path"]
    filename = row["filename"]
    actual_class = row["actual_class"]

    # Load original MRI
    original_img = Image.open(image_path).convert("RGB").resize((224, 224))
    original_array = np.array(original_img).astype("float32")

    # Prepare model input
    image_batch = np.expand_dims(original_array / 255.0, axis=0)

    model_inputs_for_image = {
        "Custom CNN": image_batch,
        "MobileNetV2": image_batch,
        "EfficientNetB0": image_batch * 255.0,
        "Fine-tuned MobileNetV2": image_batch
    }

    # Get actual class index
    class_index = class_names.index(actual_class)

    # Generate heatmaps for all four models
    for name, logit_model in logit_models.items():

        model_input = model_inputs_for_image[name]

        with tf.GradientTape() as tape:

            feature_maps, logits = logit_model(
                model_input,
                training=False
            )

            target_logit = logits[:, class_index]

        gradients = tape.gradient(target_logit, feature_maps)

        weights = tf.reduce_mean(
            gradients,
            axis=(1, 2),
            keepdims=True
        )

        cam = tf.reduce_sum(
            weights * feature_maps,
            axis=-1
        )

        cam = tf.nn.relu(cam)[0].numpy()

        if np.max(cam) > 0:
            cam = cam / np.max(cam)

        # Resize heatmap
        resized_cam = cv2.resize(cam, (224, 224))

        # Create overlay
        plt.figure(figsize=(5, 5))
        plt.imshow(original_array.astype("uint8"))
        plt.imshow(resized_cam, cmap="jet", alpha=0.45)
        plt.axis("off")
        plt.title(f"{filename} - {name}")

        # Save overlay
        model_folder = os.path.join(
            heatmap_output_dir,
            name.replace(" ", "_")
        )

        output_path = os.path.join(
            model_folder,
            filename.replace(".jpg", "_GradCAM.png")
        )

        plt.savefig(
            output_path,
            bbox_inches="tight",
            pad_inches=0.1,
            dpi=150
        )

        plt.close()

        total_heatmaps += 1

    print(f"Completed {index + 1}/23: {filename}")

print("\n================================")
print("HEATMAP GENERATION COMPLETED")
print("Total heatmaps generated:", total_heatmaps)
print("Expected heatmaps:", 92)
print("Saved location:", heatmap_output_dir)
print("================================")
```

```python
report_text = """
BRAIN TUMOUR CLASSIFICATION PROJECT
MULTI-IMAGE GRAD-CAM INVESTIGATION
====================================

1. OBJECTIVE

The objective of this investigation was to compare the
visual attention patterns of four deep learning architectures
when classifying brain MRI images.

The four architectures were:
- Custom CNN
- MobileNetV2
- EfficientNetB0
- Fine-tuned MobileNetV2


2. IMAGE SELECTION

A total of 23 MRI images were selected from the recovered
image folder.

The selection included:
- 12 incorrectly classified images
- 11 correctly classified images

Class-wise selection:
- Glioma: 5 images
- Meningioma: 8 images
- NoTumor: 5 images
- Pituitary: 5 images

All 12 misclassified examples were included, along with
11 correctly classified examples.


3. GRAD-CAM GENERATION

Logit-based Grad-CAM was used to generate heatmaps.

Each selected MRI was processed using all four architectures.

The heatmaps targeted the logit corresponding to the
actual class of each MRI image.

A total of 92 heatmap overlays were generated:

23 MRI images x 4 architectures = 92 overlays.

All 92 overlays were saved and verified.


4. ARCHITECTURE COMPARISON

Custom CNN:
Repeated boundary-focused activation was observed across
multiple MRI images. Some images also showed internal
activation.

MobileNetV2:
Activation was frequently observed within brain regions.
The spatial patterns varied between images.

EfficientNetB0:
Different and sometimes distributed activation patterns
were observed compared with MobileNetV2.

Fine-tuned MobileNetV2:
Several heatmaps showed similarities to the original
MobileNetV2 attention patterns.


5. CORRECT AND INCORRECT PREDICTION ANALYSIS

The selected 23-image sample contained:

- Correct predictions: 11
- Incorrect predictions: 12

Class-wise results:

Glioma:
5 images, 2 correct, 3 incorrect.

Meningioma:
8 images, 2 correct, 6 incorrect.

NoTumor:
5 images, 5 correct, 0 incorrect.

Pituitary:
5 images, 2 correct, 3 incorrect.

Some incorrect predictions had high confidence.
This demonstrates that high prediction confidence
does not guarantee a correct classification.


6. MAIN OBSERVATIONS

- The Custom CNN frequently displayed strong
  peripheral or boundary-focused activation.
- MobileNetV2 and Fine-tuned MobileNetV2 often
  displayed similar attention patterns.
- EfficientNetB0 frequently showed different
  spatial activation patterns.
- Attention patterns varied between MRI images.
- Broad attention patterns were observed in both
  correctly and incorrectly classified examples.


7. LIMITATIONS

- The 23 images were selected examples and do not
  represent a random evaluation sample.
- The recovered image folder has not been verified
  as the original independent test split.
- Grad-CAM shows regions influencing model scores.
  It does not provide confirmed tumour boundaries.
- The feature-map resolutions differ between
  architectures, which affects spatial detail.
- Heatmaps were individually normalized, so colour
  intensity should not be interpreted as a direct
  quantitative comparison between models.
- Tumour localization accuracy has not been validated
  using ground-truth segmentation masks.


8. CONCLUSION

The multi-image Grad-CAM investigation successfully
generated and compared 92 overlays across four
architectures and four MRI categories.

The architectures demonstrated different visual
attention patterns.

The investigation provides a qualitative comparison
of model behaviour and helps identify areas for
further study.

The next project phase will focus on tumour
segmentation using MATLAB annotations and a
U-Net architecture.
"""

report_path = os.path.join(
    heatmap_output_dir,
    "Multi_Image_GradCAM_Investigation_Report.txt"
)

with open(report_path, "w", encoding="utf-8") as file:
    file.write(report_text)

print("Documentation created successfully!")
print("\nSaved location:")
print(report_path)
```

```python
import zipfile

outer_path = "/content/drive/MyDrive/Brain_Tumor_CNN/Datasets_MP - 1.zip"

archive_names = [
    "brainTumorDataPublic_1-766.zip",
    "brainTumorDataPublic_767-1532.zip",
    "brainTumorDataPublic_1533-2298.zip",
    "brainTumorDataPublic_2299-3064.zip"
]

with zipfile.ZipFile(outer_path, "r") as outer_zip:

    for archive_name in archive_names:

        nested_path = "Datasets_MP - Copy/" + archive_name

        nested_data = outer_zip.read(nested_path)

        with zipfile.ZipFile(BytesIO(nested_data), "r") as inner_zip:

            count = sum(
                name.endswith(".mat")
                for name in inner_zip.namelist()
            )

        print(archive_name, ":", count, "MATLAB files")
```

```python
import pandas as pd
from io import BytesIO
import h5py
import zipfile

records = []

with zipfile.ZipFile(outer_path, "r") as outer_zip:

    for archive_name in archive_names:

        nested_path = "Datasets_MP - Copy/" + archive_name
        nested_data = outer_zip.read(nested_path)

        with zipfile.ZipFile(BytesIO(nested_data), "r") as inner_zip:

            mat_files = sorted(
                [name for name in inner_zip.namelist()
                 if name.endswith(".mat")],
                key=lambda x: int(x.split("/")[-1].replace(".mat", ""))
            )

            for filename in mat_files:

                mat_bytes = inner_zip.read(filename)

                with h5py.File(BytesIO(mat_bytes), "r") as mat_file:

                    cjdata = mat_file["cjdata"]

                    pid = bytes(
                        cjdata["PID"][:].flatten()
                    ).decode("ascii")

                    label_value = int(cjdata["label"][0, 0])

                    records.append({
                        "filename": filename,
                        "patient_id": pid,
                        "label": label_value
                    })

dataset_df = pd.DataFrame(records)

print("Total records:", len(dataset_df))
print("Unique patients:", dataset_df["patient_id"].nunique())

print("\nRecords by numerical label:")
print(dataset_df["label"].value_counts().sort_index())
```

```python
print("First 20 patient IDs:\n")
print(dataset_df["patient_id"].head(20).to_list())

print("\nNumber of unique patient IDs:")
print(dataset_df["patient_id"].nunique())

print("\nMost frequent patient IDs:")
print(dataset_df["patient_id"].value_counts().head(10))

print("\nAll unique patient IDs:")
print(sorted(dataset_df["patient_id"].unique()))
```

```python
records = []

with zipfile.ZipFile(outer_path, "r") as outer_zip:

    for archive_name in archive_names:

        nested_path = "Datasets_MP - Copy/" + archive_name
        nested_data = outer_zip.read(nested_path)

        with zipfile.ZipFile(BytesIO(nested_data), "r") as inner_zip:

            mat_files = sorted(
                [name for name in inner_zip.namelist()
                 if name.endswith(".mat")],
                key=lambda x: int(x.split("/")[-1].replace(".mat", ""))
            )

            for filename in mat_files:

                mat_bytes = inner_zip.read(filename)

                with h5py.File(BytesIO(mat_bytes), "r") as mat_file:

                    cjdata = mat_file["cjdata"]

                    pid_bytes = cjdata["PID"][:].flatten().tobytes()

                    pid = pid_bytes.decode(
                        "utf-16le"
                    ).replace("\x00", "")

                    label_value = int(cjdata["label"][0, 0])

                    records.append({
                        "filename": filename,
                        "patient_id": pid,
                        "label": label_value
                    })

dataset_df = pd.DataFrame(records)

print("Total records:", len(dataset_df))
print("Unique patients:", dataset_df["patient_id"].nunique())

print("\nFirst 15 decoded patient IDs:")
print(dataset_df["patient_id"].head(15).to_list())

print("\nMost frequent patient IDs:")
print(dataset_df["patient_id"].value_counts().head(10))
```

```python
patient_df = (
    dataset_df.groupby("patient_id")
    .agg(
        label=("label", "first"),
        number_of_slices=("filename", "count")
    )
    .reset_index()
)

patient_df["tumour_category"] = patient_df["label"].map({
    1: "Meningioma",
    2: "Glioma",
    3: "Pituitary Tumour"
})

print("Patient-level dataset shape:", patient_df.shape)

print("\nPatients by tumour category:")
print(
    patient_df.groupby("tumour_category")
    .size()
)

print("\nFirst 10 patient records:")
print(patient_df.head(10))
```

```python
from sklearn.model_selection import train_test_split

# First: separate 70% training patients from the remaining 30%
train_patients, remaining_patients = train_test_split(
    patient_df,
    test_size=0.30,
    random_state=42,
    stratify=patient_df["label"]
)

# Second: divide the remaining 30% equally
val_patients, test_patients = train_test_split(
    remaining_patients,
    test_size=0.50,
    random_state=42,
    stratify=remaining_patients["label"]
)

# Display patient counts
print("TRAINING PATIENTS:", len(train_patients))
print("VALIDATION PATIENTS:", len(val_patients))
print("TESTING PATIENTS:", len(test_patients))

print("\nTumour category distribution:")

for name, group in [
    ("Training", train_patients),
    ("Validation", val_patients),
    ("Testing", test_patients)
]:
    print(f"\n{name}:")
    print(group["tumour_category"].value_counts())

# Check that no patient appears in multiple groups
train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

print("\nPatient overlap checks:")
print("Training vs Validation:", len(train_ids & val_ids))
print("Training vs Testing:", len(train_ids & test_ids))
print("Validation vs Testing:", len(val_ids & test_ids))
```

```python
# Get the patient IDs for each group
train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

# Assign every MRI record to its patient's group
dataset_df["split"] = dataset_df["patient_id"].map(
    lambda pid:
        "train" if pid in train_ids else
        "validation" if pid in val_ids else
        "test" if pid in test_ids else
        "unknown"
)

# Check the number of MRI slices in each group
print("MRI slices assigned to each group:\n")
print(dataset_df["split"].value_counts())

print("\nPatient counts in each group:\n")
print(
    dataset_df.groupby("split")["patient_id"]
    .nunique()
)

print("\nUnassigned MRI records:",
      (dataset_df["split"] == "unknown").sum())
```

```python
from PIL import Image
import numpy as np
import os
import zipfile
import h5py
from io import BytesIO

# Temporary test folder
test_dir = os.path.join(segmentation_dir, "extraction_test")
os.makedirs(test_dir, exist_ok=True)

# Read the first MATLAB file
with zipfile.ZipFile(outer_path, "r") as outer_zip:

    nested_data = outer_zip.read(
        "Datasets_MP - Copy/brainTumorDataPublic_1-766.zip"
    )

    with zipfile.ZipFile(BytesIO(nested_data), "r") as inner_zip:

        mat_bytes = inner_zip.read("1.mat")

        with h5py.File(BytesIO(mat_bytes), "r") as mat_file:

            cjdata = mat_file["cjdata"]

            image = cjdata["image"][:]
            tumor_mask = cjdata["tumorMask"][:]

# Convert MRI intensity values to 0–255
image_8bit = (
    255 * (image - image.min()) /
    (image.max() - image.min())
).astype(np.uint8)

# Save MRI and mask
Image.fromarray(image_8bit).save(
    os.path.join(test_dir, "1_image.png")
)

Image.fromarray(tumor_mask).save(
    os.path.join(test_dir, "1_mask.png")
)

print("MRI saved:", os.path.exists(
    os.path.join(test_dir, "1_image.png")
))

print("Mask saved:", os.path.exists(
    os.path.join(test_dir, "1_mask.png")
))

print("MRI dimensions:", image_8bit.shape)
print("Mask dimensions:", tumor_mask.shape)
print("Mask values:", np.unique(tumor_mask))
```

```python
from PIL import Image
from io import BytesIO
import zipfile
import h5py
import numpy as np
import os

# Create a lookup table for the patient-separated splits
split_lookup = dataset_df.set_index("filename")["split"].to_dict()

extracted_counts = {
    "train": 0,
    "validation": 0,
    "test": 0
}

with zipfile.ZipFile(outer_path, "r") as outer_zip:

    for archive_name in archive_names:

        nested_path = "Datasets_MP - Copy/" + archive_name
        nested_data = outer_zip.read(nested_path)

        with zipfile.ZipFile(BytesIO(nested_data), "r") as inner_zip:

            mat_files = [
                name for name in inner_zip.namelist()
                if name.endswith(".mat")
            ]

            for filename in mat_files:

                split = split_lookup[filename]

                mat_bytes = inner_zip.read(filename)

                with h5py.File(BytesIO(mat_bytes), "r") as mat_file:

                    cjdata = mat_file["cjdata"]

                    image = cjdata["image"][:]
                    mask = cjdata["tumorMask"][:]

                # Apply percentile-based MRI contrast scaling
                lower = np.percentile(image, 1)
                upper = np.percentile(image, 99)

                clipped = np.clip(image, lower, upper)

                image_8bit = (
                    255 * (clipped - lower) / (upper - lower)
                ).astype(np.uint8)

                # Preserve the original binary mask
                mask_8bit = mask.astype(np.uint8)

                # Prepare matching filenames
                file_id = os.path.splitext(
                    os.path.basename(filename)
                )[0]

                image_path = os.path.join(
                    segmentation_dir, split,
                    "images", file_id + ".png"
                )

                mask_path = os.path.join(
                    segmentation_dir, split,
                    "masks", file_id + ".png"
                )

                # Save MRI and mask
                Image.fromarray(image_8bit).save(image_path)
                Image.fromarray(mask_8bit).save(mask_path)

                extracted_counts[split] += 1

        print("Finished:", archive_name)

print("\nEXTRACTION SUMMARY")

for split, count in extracted_counts.items():
    print(split, ":", count, "MRI-mask pairs")

print("\nTotal extracted:", sum(extracted_counts.values()))
```

```python
from PIL import Image
import numpy as np
import os

total_checked = 0
problems = []

for split in ["train", "validation", "test"]:

    image_dir = os.path.join(
        segmentation_dir, split, "images"
    )

    mask_dir = os.path.join(
        segmentation_dir, split, "masks"
    )

    image_files = set(os.listdir(image_dir))
    mask_files = set(os.listdir(mask_dir))

    print("\nChecking:", split)

    print("MRI files:", len(image_files))
    print("Mask files:", len(mask_files))

    if image_files != mask_files:
        problems.append(f"{split}: MRI-mask filenames do not match")

    for filename in sorted(image_files & mask_files):

        image = np.array(
            Image.open(os.path.join(image_dir, filename))
        )

        mask = np.array(
            Image.open(os.path.join(mask_dir, filename))
        )

        if image.shape != (512, 512):
            problems.append(f"{filename}: incorrect MRI dimensions")

        if mask.shape != (512, 512):
            problems.append(f"{filename}: incorrect mask dimensions")

        if not np.isin(mask, [0, 1]).all():
            problems.append(f"{filename}: mask is not binary")

        total_checked += 1

print("\nTOTAL PAIRS CHECKED:", total_checked)
print("TOTAL PROBLEMS:", len(problems))

if problems:
    print("\nProblems found:")
    print(problems[:20])
else:
    print("\nAll MRI-mask pairs passed the integrity checks!")
```

```python
from collections import Counter
import os
import numpy as np
from PIL import Image

dimension_counts = Counter()
unusual_files = []

for split in ["train", "validation", "test"]:

    image_dir = os.path.join(
        segmentation_dir, split, "images"
    )

    mask_dir = os.path.join(
        segmentation_dir, split, "masks"
    )

    for filename in sorted(os.listdir(image_dir)):

        image = np.array(
            Image.open(os.path.join(image_dir, filename))
        )

        mask = np.array(
            Image.open(os.path.join(mask_dir, filename))
        )

        dimension_counts[image.shape] += 1

        if image.shape != (512, 512) or mask.shape != (512, 512):

            unusual_files.append({
                "split": split,
                "filename": filename,
                "MRI_shape": image.shape,
                "mask_shape": mask.shape
            })

print("IMAGE DIMENSION DISTRIBUTION:")
for shape, count in dimension_counts.items():
    print(shape, ":", count, "files")

print("\nUNUSUAL MRI-MASK PAIRS:")
for item in unusual_files:
    print(item)

print("\nTotal unusual pairs:", len(unusual_files))
```

```python
from PIL import Image
import os
import shutil

# Create a separate standardised dataset
standardized_dir = os.path.join(
    "/content/drive/MyDrive/Brain_Tumor_CNN",
    "Brain_Tumor_Segmentation_512"
)

for split in ["train", "validation", "test"]:

    source_images = os.path.join(
        segmentation_dir, split, "images"
    )

    source_masks = os.path.join(
        segmentation_dir, split, "masks"
    )

    target_images = os.path.join(
        standardized_dir, split, "images"
    )

    target_masks = os.path.join(
        standardized_dir, split, "masks"
    )

    os.makedirs(target_images, exist_ok=True)
    os.makedirs(target_masks, exist_ok=True)

    for filename in os.listdir(source_images):

        # Open original MRI
        image = Image.open(
            os.path.join(source_images, filename)
        )

        # Open matching tumour mask
        mask = Image.open(
            os.path.join(source_masks, filename)
        )

        # Resize only when dimensions differ
        if image.size != (512, 512):
            image = image.resize(
                (512, 512),
                Image.Resampling.BILINEAR
            )

            mask = mask.resize(
                (512, 512),
                Image.Resampling.NEAREST
            )

        # Save both files in the new dataset
        image.save(
            os.path.join(target_images, filename)
        )

        mask.save(
            os.path.join(target_masks, filename)
        )

    print("Finished:", split)

print("\nStandardised dataset created!")
print("Location:", standardized_dir)
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

print("TensorFlow:", tf.__version__)
print("NumPy:", np.__version__)

print("\nGPU available:",
      len(tf.config.list_physical_devices("GPU")) > 0)

print("\nSegmentation dataset available:",
      os.path.exists(standardized_dir))
```

```python
from PIL import Image

# Training image and mask folders
train_image_dir = os.path.join(
    standardized_dir, "train", "images"
)

train_mask_dir = os.path.join(
    standardized_dir, "train", "masks"
)

# Select one MRI file
sample_filename = sorted(
    os.listdir(train_image_dir),
    key=lambda x: int(x.split(".")[0])
)[0]

# Load MRI
sample_image = np.array(
    Image.open(
        os.path.join(train_image_dir, sample_filename)
    )
)

# Load matching tumour mask
sample_mask = np.array(
    Image.open(
        os.path.join(train_mask_dir, sample_filename)
    )
)

print("Filename:", sample_filename)
print("MRI shape:", sample_image.shape)
print("Mask shape:", sample_mask.shape)

print("MRI pixel range:",
      sample_image.min(), "to", sample_image.max())

print("Mask values:", np.unique(sample_mask))
```

```python
# Normalise MRI image to the range 0–1
sample_image_normalized = (
    sample_image.astype(np.float32) / 255.0
)

# Convert tumour mask to floating-point format
sample_mask_float = sample_mask.astype(np.float32)

print("Normalised MRI:")
print("Minimum:", sample_image_normalized.min())
print("Maximum:", sample_image_normalized.max())

print("\nTumour mask:")
print("Unique values:", np.unique(sample_mask_float))

print("\nMRI data type:", sample_image_normalized.dtype)
print("Mask data type:", sample_mask_float.dtype)
```

```python
# Add one channel dimension to the MRI
sample_input = np.expand_dims(
    sample_image_normalized, axis=-1
)

# Add one channel dimension to the tumour mask
sample_target = np.expand_dims(
    sample_mask_float, axis=-1
)

print("MRI input shape:", sample_input.shape)
print("Tumour mask shape:", sample_target.shape)

print("\nInput data type:", sample_input.dtype)
print("Target data type:", sample_target.dtype)
```

```python
from tensorflow.keras.layers import (
    Conv2D,
    BatchNormalization,
    Activation
)

def convolution_block(inputs, filters):

    x = Conv2D(
        filters,
        kernel_size=3,
        padding="same",
        kernel_initializer="he_normal"
    )(inputs)

    x = BatchNormalization()(x)
    x = Activation("relu")(x)

    x = Conv2D(
        filters,
        kernel_size=3,
        padding="same",
        kernel_initializer="he_normal"
    )(x)

    x = BatchNormalization()(x)
    x = Activation("relu")(x)

    return x

print("Convolution block function created successfully!")
```

```python
from tensorflow.keras.layers import Input, MaxPooling2D

# Define the MRI input shape
inputs = Input(shape=(512, 512, 1))

# First encoder level
encoder1 = convolution_block(inputs, 16)

# Reduce spatial dimensions
pool1 = MaxPooling2D(pool_size=(2, 2))(encoder1)

print("First encoder level shape:", encoder1.shape)
print("After max pooling:", pool1.shape)
```

```python
# Second encoder level
encoder2 = convolution_block(pool1, 32)

# Reduce spatial dimensions again
pool2 = MaxPooling2D(pool_size=(2, 2))(encoder2)

print("Second encoder level shape:", encoder2.shape)
print("After second max pooling:", pool2.shape)
```

```python
# Third encoder level
encoder3 = convolution_block(pool2, 64)

# Reduce spatial dimensions again
pool3 = MaxPooling2D(pool_size=(2, 2))(encoder3)

print("Third encoder level shape:", encoder3.shape)
print("After third max pooling:", pool3.shape)
```

```python
# Fourth encoder level
encoder4 = convolution_block(pool3, 128)

# Reduce spatial dimensions again
pool4 = MaxPooling2D(pool_size=(2, 2))(encoder4)

print("Fourth encoder level shape:", encoder4.shape)
print("After fourth max pooling:", pool4.shape)
```

```python
# U-Net bottleneck
bottleneck = convolution_block(pool4, 256)

print("Bottleneck shape:", bottleneck.shape)
```

```python
from tensorflow.keras.layers import UpSampling2D, Concatenate

# Upsample the bottleneck
up1 = UpSampling2D(size=(2, 2))(bottleneck)

# Connect with the fourth encoder level
merge1 = Concatenate()([up1, encoder4])

# Apply convolution block
decoder1 = convolution_block(merge1, 128)

print("Upsampled shape:", up1.shape)
print("After skip connection:", merge1.shape)
print("First decoder level shape:", decoder1.shape)
```

```python
# Upsample the first decoder level
up2 = UpSampling2D(size=(2, 2))(decoder1)

# Connect with the third encoder level
merge2 = Concatenate()([up2, encoder3])

# Apply convolution block
decoder2 = convolution_block(merge2, 64)

print("Upsampled shape:", up2.shape)
print("After skip connection:", merge2.shape)
print("Second decoder level shape:", decoder2.shape)
```

```python
# Upsample the second decoder level
up3 = UpSampling2D(size=(2, 2))(decoder2)

# Connect with the second encoder level
merge3 = Concatenate()([up3, encoder2])

# Apply convolution block
decoder3 = convolution_block(merge3, 32)

print("Upsampled shape:", up3.shape)
print("After skip connection:", merge3.shape)
print("Third decoder level shape:", decoder3.shape)
```

```python
# Upsample the third decoder level
up4 = UpSampling2D(size=(2, 2))(decoder3)

# Connect with the first encoder level
merge4 = Concatenate()([up4, encoder1])

# Apply convolution block
decoder4 = convolution_block(merge4, 16)

print("Upsampled shape:", up4.shape)
print("After skip connection:", merge4.shape)
print("Fourth decoder level shape:", decoder4.shape)
```

```python
from tensorflow.keras.layers import Conv2D

# Final segmentation output layer
outputs = Conv2D(
    1,
    kernel_size=1,
    activation="sigmoid",
    name="segmentation_output"
)(decoder4)

print("Final output shape:", outputs.shape)
```

```python
from tensorflow.keras.models import Model

# Assemble the complete U-Net
unet_model = Model(
    inputs=inputs,
    outputs=outputs,
    name="Brain_Tumor_U_Net"
)

# Display model information
print("U-Net model created successfully!")
print("Input shape:", unet_model.input_shape)
print("Output shape:", unet_model.output_shape)
print("Total parameters:", unet_model.count_params())
```

```python
# Define the image and mask directories

train_image_dir = os.path.join(standardized_dir, "train", "images")
train_mask_dir = os.path.join(standardized_dir, "train", "masks")

val_image_dir = os.path.join(standardized_dir, "validation", "images")
val_mask_dir = os.path.join(standardized_dir, "validation", "masks")

test_image_dir = os.path.join(standardized_dir, "test", "images")
test_mask_dir = os.path.join(standardized_dir, "test", "masks")


# Count PNG files in each directory
def count_png_files(folder):
    return len([
        filename for filename in os.listdir(folder)
        if filename.lower().endswith(".png")
    ])


print("TRAINING DATA")
print("MRI images:", count_png_files(train_image_dir))
print("Tumour masks:", count_png_files(train_mask_dir))

print("\nVALIDATION DATA")
print("MRI images:", count_png_files(val_image_dir))
print("Tumour masks:", count_png_files(val_mask_dir))

print("\nTESTING DATA")
print("MRI images:", count_png_files(test_image_dir))
print("Tumour masks:", count_png_files(test_mask_dir))
```

```python
def load_image_and_mask(image_path, mask_path):

    # Load MRI image
    image = tf.io.read_file(image_path)
    image = tf.io.decode_png(image, channels=1)
    image = tf.cast(image, tf.float32) / 255.0

    # Load tumour mask
    mask = tf.io.read_file(mask_path)
    mask = tf.io.decode_png(mask, channels=1)
    mask = tf.cast(mask, tf.float32)

    return image, mask


print("Image and mask loading function created successfully!")
```

```python
# Collect training image filenames
train_filenames = sorted([
    f for f in os.listdir(train_image_dir)
    if f.lower().endswith(".png")
])

# Create corresponding image and mask paths
train_image_paths = [
    os.path.join(train_image_dir, f)
    for f in train_filenames
]

train_mask_paths = [
    os.path.join(train_mask_dir, f)
    for f in train_filenames
]

# Verify matching files
print("Number of training images:", len(train_image_paths))
print("Number of training masks:", len(train_mask_paths))

print("\nFirst 5 image-mask pairs:")

for image_path, mask_path in zip(
    train_image_paths[:5],
    train_mask_paths[:5]
):
    print(os.path.basename(image_path), "<-->", os.path.basename(mask_path))

# Check whether all corresponding mask files exist
all_masks_exist = all(os.path.exists(path) for path in train_mask_paths)

print("\nAll corresponding masks exist:", all_masks_exist)
```

```python
# Collect validation filenames
val_filenames = sorted([
    f for f in os.listdir(val_image_dir)
    if f.lower().endswith(".png")
])

val_image_paths = [
    os.path.join(val_image_dir, f)
    for f in val_filenames
]

val_mask_paths = [
    os.path.join(val_mask_dir, f)
    for f in val_filenames
]


# Collect testing filenames
test_filenames = sorted([
    f for f in os.listdir(test_image_dir)
    if f.lower().endswith(".png")
])

test_image_paths = [
    os.path.join(test_image_dir, f)
    for f in test_filenames
]

test_mask_paths = [
    os.path.join(test_mask_dir, f)
    for f in test_filenames
]


# Verify the datasets
print("VALIDATION")
print("Images:", len(val_image_paths))
print("Masks:", len(val_mask_paths))
print("All masks exist:", all(os.path.exists(p) for p in val_mask_paths))

print("\nTESTING")
print("Images:", len(test_image_paths))
print("Masks:", len(test_mask_paths))
print("All masks exist:", all(os.path.exists(p) for p in test_mask_paths))
```

```python
# Set batch size
BATCH_SIZE = 2

# Create training dataset
train_dataset = tf.data.Dataset.from_tensor_slices(
    (train_image_paths, train_mask_paths)
)

train_dataset = train_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_dataset = train_dataset.shuffle(
    buffer_size=500,
    seed=42
)

train_dataset = train_dataset.batch(BATCH_SIZE)

train_dataset = train_dataset.prefetch(tf.data.AUTOTUNE)


print("Training TensorFlow dataset created successfully!")
print("Batch size:", BATCH_SIZE)
```

```python
# Create validation dataset
val_dataset = tf.data.Dataset.from_tensor_slices(
    (val_image_paths, val_mask_paths)
)

val_dataset = val_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_dataset = val_dataset.batch(BATCH_SIZE)
val_dataset = val_dataset.prefetch(tf.data.AUTOTUNE)


# Create testing dataset
test_dataset = tf.data.Dataset.from_tensor_slices(
    (test_image_paths, test_mask_paths)
)

test_dataset = test_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_dataset = test_dataset.batch(BATCH_SIZE)
test_dataset = test_dataset.prefetch(tf.data.AUTOTUNE)


print("Validation and testing datasets created successfully!")
```

```python
# Retrieve one batch from the training dataset
sample_batch_images, sample_batch_masks = next(iter(train_dataset))

# Display batch information
print("MRI batch shape:", sample_batch_images.shape)
print("Mask batch shape:", sample_batch_masks.shape)

print("\nMRI data type:", sample_batch_images.dtype)
print("Mask data type:", sample_batch_masks.dtype)

print("\nMRI minimum value:", tf.reduce_min(sample_batch_images).numpy())
print("MRI maximum value:", tf.reduce_max(sample_batch_images).numpy())

print("\nUnique mask values:", np.unique(sample_batch_masks.numpy()))
```

```python
def dice_coefficient(y_true, y_pred):

    smooth = 1e-6

    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)

    y_true = tf.reshape(y_true, [-1])
    y_pred = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true * y_pred)

    dice = (
        (2.0 * intersection + smooth) /
        (tf.reduce_sum(y_true) + tf.reduce_sum(y_pred) + smooth)
    )

    return dice


print("Dice coefficient function created successfully!")
```

```python
def dice_loss(y_true, y_pred):

    return 1.0 - dice_coefficient(y_true, y_pred)


print("Dice loss function created successfully!")
```

```python
def iou_metric(y_true, y_pred):

    smooth = 1e-6

    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)

    y_true = tf.reshape(y_true, [-1])
    y_pred = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true * y_pred)

    union = (
        tf.reduce_sum(y_true)
        + tf.reduce_sum(y_pred)
        - intersection
    )

    iou = (intersection + smooth) / (union + smooth)

    return iou


print("IoU metric function created successfully!")
```

```python
# Configure the Adam optimizer
optimizer = tf.keras.optimizers.Adam(
    learning_rate=0.0001
)

# Compile the U-Net model
unet_model.compile(
    optimizer=optimizer,
    loss=dice_loss,
    metrics=[
        dice_coefficient,
        iou_metric
    ]
)

print("U-Net model compiled successfully!")
print("Optimizer: Adam")
print("Learning rate: 0.0001")
print("Loss function: Dice Loss")
print("Metrics: Dice Coefficient and IoU")
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

# Number of training epochs
EPOCHS = 30

# Save the best model weights
checkpoint = ModelCheckpoint(
    filepath=os.path.join(standardized_dir, "best_unet.weights.h5"),
    monitor="val_loss",
    save_best_only=True,
    save_weights_only=True,
    verbose=1
)

# Stop training if validation loss stops improving
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

callbacks = [checkpoint, early_stopping]

print("Training configuration prepared successfully!")
print("Maximum epochs:", EPOCHS)
print("Early stopping patience: 5")
print("Best model weights will be saved.")
```

```python
# Train the U-Net model
history = unet_model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

print("\nU-Net training completed!")
```

```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, Activation,
    MaxPooling2D, UpSampling2D, Concatenate
)
from tensorflow.keras.models import Model
import os

# Dice and IoU metrics
def dice_coefficient(y_true, y_pred):
    y_true = tf.reshape(tf.cast(y_true, tf.float32), [-1])
    y_pred = tf.reshape(tf.cast(y_pred, tf.float32), [-1])

    intersection = tf.reduce_sum(y_true * y_pred)
    return (2.0 * intersection + 1e-6) / (
        tf.reduce_sum(y_true) + tf.reduce_sum(y_pred) + 1e-6
    )

def dice_loss(y_true, y_pred):
    return 1.0 - dice_coefficient(y_true, y_pred)

def iou_metric(y_true, y_pred):
    y_true = tf.reshape(tf.cast(y_true, tf.float32), [-1])
    y_pred = tf.reshape(tf.cast(y_pred, tf.float32), [-1])

    intersection = tf.reduce_sum(y_true * y_pred)
    union = (
        tf.reduce_sum(y_true)
        + tf.reduce_sum(y_pred)
        - intersection
    )
    return (intersection + 1e-6) / (union + 1e-6)

# Same convolution block used in the original U-Net
def convolution_block(inputs, filters):
    x = Conv2D(
        filters, 3, padding="same",
        kernel_initializer="he_normal"
    )(inputs)
    x = BatchNormalization()(x)
    x = Activation("relu")(x)

    x = Conv2D(
        filters, 3, padding="same",
        kernel_initializer="he_normal"
    )(x)
    x = BatchNormalization()(x)
    x = Activation("relu")(x)

    return x

# Rebuild the original U-Net architecture
inputs = Input(shape=(512, 512, 1))

encoder1 = convolution_block(inputs, 16)
pool1 = MaxPooling2D(pool_size=(2, 2))(encoder1)

encoder2 = convolution_block(pool1, 32)
pool2 = MaxPooling2D(pool_size=(2, 2))(encoder2)

encoder3 = convolution_block(pool2, 64)
pool3 = MaxPooling2D(pool_size=(2, 2))(encoder3)

encoder4 = convolution_block(pool3, 128)
pool4 = MaxPooling2D(pool_size=(2, 2))(encoder4)

bottleneck = convolution_block(pool4, 256)

up1 = UpSampling2D(size=(2, 2))(bottleneck)
merge1 = Concatenate()([up1, encoder4])
decoder1 = convolution_block(merge1, 128)

up2 = UpSampling2D(size=(2, 2))(decoder1)
merge2 = Concatenate()([up2, encoder3])
decoder2 = convolution_block(merge2, 64)

up3 = UpSampling2D(size=(2, 2))(decoder2)
merge3 = Concatenate()([up3, encoder2])
decoder3 = convolution_block(merge3, 32)

up4 = UpSampling2D(size=(2, 2))(decoder3)
merge4 = Concatenate()([up4, encoder1])
decoder4 = convolution_block(merge4, 16)

outputs = Conv2D(
    1, kernel_size=1,
    activation="sigmoid",
    name="segmentation_output"
)(decoder4)

unet_model = Model(
    inputs=inputs,
    outputs=outputs,
    name="Brain_Tumor_U_Net"
)

# Compile the model
unet_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss=dice_loss,
    metrics=[dice_coefficient, iou_metric]
)

# Load the previously saved best weights
weights_path = os.path.join(
    standardized_dir,
    "best_unet.weights.h5"
)

unet_model.load_weights(weights_path)

print("Saved weights loaded successfully!")
print("Model name:", unet_model.name)
print("Input shape:", unet_model.input_shape)
print("Output shape:", unet_model.output_shape)
print("Total parameters:", unet_model.count_params())
```

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

# Test image and mask folders
test_image_dir = os.path.join(standardized_dir, "test", "images")
test_mask_dir = os.path.join(standardized_dir, "test", "masks")

# Collect matching image and mask paths
test_image_paths = sorted(glob.glob(os.path.join(test_image_dir, "*.png")))
test_mask_paths = sorted(glob.glob(os.path.join(test_mask_dir, "*.png")))

print("Test images:", len(test_image_paths))
print("Test masks:", len(test_mask_paths))

# Load an MRI image and its corresponding binary mask
def load_image_and_mask(image_path, mask_path):
    image = tf.io.read_file(image_path)
    image = tf.image.decode_png(image, channels=1)
    image = tf.cast(image, tf.float32) / 255.0

    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = tf.cast(mask, tf.float32)

    return image, mask

# Create the test dataset
test_dataset = tf.data.Dataset.from_tensor_slices(
    (test_image_paths, test_mask_paths)
)

test_dataset = test_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_dataset = test_dataset.batch(2).prefetch(tf.data.AUTOTUNE)

print("Test dataset restored successfully!")
```

```python
# Calculate Dice and IoU for every test image

all_dice_scores = []
all_iou_scores = []

for images, masks in test_dataset:

    # Generate model predictions
    predictions = unet_model.predict(images, verbose=0)

    # Convert probabilities into binary masks using threshold 0.5
    predicted_masks = (predictions >= 0.5).astype(np.uint8)
    true_masks = masks.numpy().astype(np.uint8)

    # Calculate scores separately for each image in the batch
    for i in range(len(predicted_masks)):

        y_true = true_masks[i].flatten()
        y_pred = predicted_masks[i].flatten()

        intersection = np.sum(y_true * y_pred)
        total_true = np.sum(y_true)
        total_pred = np.sum(y_pred)
        union = total_true + total_pred - intersection

        dice = (2 * intersection + 1e-6) / (
            total_true + total_pred + 1e-6
        )

        iou = (intersection + 1e-6) / (union + 1e-6)

        all_dice_scores.append(dice)
        all_iou_scores.append(iou)

# Convert results into NumPy arrays
all_dice_scores = np.array(all_dice_scores)
all_iou_scores = np.array(all_iou_scores)

# Display summary statistics
print("Number of test images:", len(all_dice_scores))

print("\nDICE SCORE")
print("Mean:", np.mean(all_dice_scores))
print("Median:", np.median(all_dice_scores))
print("Standard deviation:", np.std(all_dice_scores))

print("\nIoU SCORE")
print("Mean:", np.mean(all_iou_scores))
print("Median:", np.median(all_iou_scores))
print("Standard deviation:", np.std(all_iou_scores))
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(14, 5))

# Dice score distribution
plt.subplot(1, 2, 1)
plt.hist(all_dice_scores, bins=20, edgecolor="black")
plt.axvline(
    np.mean(all_dice_scores),
    linestyle="--",
    linewidth=2,
    label=f"Mean = {np.mean(all_dice_scores):.3f}"
)
plt.title("Distribution of Dice Scores")
plt.xlabel("Dice Score")
plt.ylabel("Number of MRI Slices")
plt.xlim(0, 1)
plt.legend()

# IoU score distribution
plt.subplot(1, 2, 2)
plt.hist(all_iou_scores, bins=20, edgecolor="black")
plt.axvline(
    np.mean(all_iou_scores),
    linestyle="--",
    linewidth=2,
    label=f"Mean = {np.mean(all_iou_scores):.3f}"
)
plt.title("Distribution of IoU Scores")
plt.xlabel("IoU Score")
plt.ylabel("Number of MRI Slices")
plt.xlim(0, 1)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Inspect the five worst-performing MRI slices

plt.figure(figsize=(16, 18))

for row, index in enumerate(worst_indices):

    image, mask = load_image_and_mask(
        test_image_paths[index],
        test_mask_paths[index]
    )

    # Generate prediction
    prediction = unet_model.predict(
        tf.expand_dims(image, axis=0),
        verbose=0
    )[0, :, :, 0]

    true_mask = mask.numpy().squeeze() > 0.5
    predicted_mask = prediction >= 0.5

    # Calculate pixel counts
    intersection = np.logical_and(
        true_mask, predicted_mask
    ).sum()

    print(
        f"{os.path.basename(test_image_paths[index])}: "
        f"Ground-truth pixels = {true_mask.sum()}, "
        f"Predicted pixels = {predicted_mask.sum()}, "
        f"Overlap pixels = {intersection}"
    )

    # Create RGB overlay
    overlay = np.zeros((*true_mask.shape, 3), dtype=np.float32)
    overlay[true_mask] = [0, 1, 0]          # Green
    overlay[predicted_mask] += [1, 0, 0]    # Red
    overlay[np.logical_and(true_mask, predicted_mask)] = [1, 1, 0]

    # Display four views
    plt.subplot(5, 4, row * 4 + 1)
    plt.imshow(image.numpy().squeeze(), cmap="gray")
    plt.title(f"MRI: {os.path.basename(test_image_paths[index])}")
    plt.axis("off")

    plt.subplot(5, 4, row * 4 + 2)
    plt.imshow(true_mask, cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")

    plt.subplot(5, 4, row * 4 + 3)
    plt.imshow(predicted_mask, cmap="gray")
    plt.title("U-Net Prediction")
    plt.axis("off")

    plt.subplot(5, 4, row * 4 + 4)
    plt.imshow(overlay)
    plt.title("Overlay")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("DETAILED ANALYSIS OF WORST-PERFORMING CASES")
print("=" * 60)

for index in worst_indices:

    image, mask = load_image_and_mask(
        test_image_paths[index],
        test_mask_paths[index]
    )

    prediction = unet_model.predict(
        tf.expand_dims(image, axis=0),
        verbose=0
    )[0, :, :, 0]

    true_mask = mask.numpy().squeeze() > 0.5
    predicted_mask = prediction >= 0.5

    gt_pixels = np.sum(true_mask)
    pred_pixels = np.sum(predicted_mask)

    overlap = np.sum(
        np.logical_and(true_mask, predicted_mask)
    )

    print(f"\nImage: {os.path.basename(test_image_paths[index])}")
    print("Ground-truth tumour pixels:", gt_pixels)
    print("Predicted tumour pixels:", pred_pixels)
    print("Overlapping pixels:", overlap)
    print("Dice score:", f"{all_dice_scores[index]:.4f}")

    if pred_pixels == 0:
        print("Observation: Model predicted an empty mask.")
    elif overlap == 0:
        print("Observation: Prediction exists but has no overlap.")
    else:
        print("Observation: Partial overlap exists.")
```

```python
print("LOCATION ANALYSIS OF FIVE WORST CASES")
print("=" * 60)

for index in worst_indices:

    image, mask = load_image_and_mask(
        test_image_paths[index],
        test_mask_paths[index]
    )

    prediction = unet_model.predict(
        tf.expand_dims(image, axis=0),
        verbose=0
    )[0, :, :, 0]

    gt_mask = mask.numpy().squeeze() > 0.5
    pred_mask = prediction >= 0.5

    gt_y, gt_x = np.where(gt_mask)
    pred_y, pred_x = np.where(pred_mask)

    print(f"\nImage: {os.path.basename(test_image_paths[index])}")

    print(
        "Ground-truth centre (x, y):",
        (round(np.mean(gt_x), 1), round(np.mean(gt_y), 1))
    )

    print(
        "Prediction centre (x, y):",
        (round(np.mean(pred_x), 1), round(np.mean(pred_y), 1))
    )

    distance = np.sqrt(
        (np.mean(gt_x) - np.mean(pred_x)) ** 2
        + (np.mean(gt_y) - np.mean(pred_y)) ** 2
    )

    print("Distance between centres (pixels):", round(distance, 2))
```

```python
plt.figure(figsize=(15, 15))

for row, index in enumerate(worst_indices):

    image, mask = load_image_and_mask(
        test_image_paths[index],
        test_mask_paths[index]
    )

    prediction = unet_model.predict(
        tf.expand_dims(image, axis=0),
        verbose=0
    )[0, :, :, 0]

    mri = image.numpy().squeeze()
    gt_mask = mask.numpy().squeeze() > 0.5
    pred_mask = prediction >= 0.5

    # Convert grayscale MRI to RGB
    background = np.stack([mri] * 3, axis=-1)

    # Create separate MRI overlays
    gt_overlay = background.copy()
    pred_overlay = background.copy()

    # Green ground-truth region
    gt_overlay[gt_mask] = (
        0.5 * gt_overlay[gt_mask] + 0.5 * np.array([0, 1, 0])
    )

    # Red predicted region
    pred_overlay[pred_mask] = (
        0.5 * pred_overlay[pred_mask] + 0.5 * np.array([1, 0, 0])
    )

    plt.subplot(5, 3, row * 3 + 1)
    plt.imshow(mri, cmap="gray")
    plt.title(os.path.basename(test_image_paths[index]))
    plt.axis("off")

    plt.subplot(5, 3, row * 3 + 2)
    plt.imshow(gt_overlay)
    plt.title("MRI + Ground Truth (Green)")
    plt.axis("off")

    plt.subplot(5, 3, row * 3 + 3)
    plt.imshow(pred_overlay)
    plt.title("MRI + Prediction (Red)")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import h5py
import numpy as np
import os
import glob
from collections import Counter

annotation_records = []

for mat_path in mat_files:

    with h5py.File(mat_path, "r") as mat_file:

        cjdata = mat_file["cjdata"]

        label = int(np.array(cjdata["label"]).squeeze())

        pid_raw = np.array(cjdata["PID"]).squeeze()

        # Decode MATLAB character codes
        patient_id = "".join(
            chr(int(value)) for value in pid_raw
        )

        annotation_records.append({
            "filename": os.path.basename(mat_path),
            "label": label,
            "patient_id": patient_id
        })

# Count samples in each tumour class
class_counts = Counter(
    record["label"] for record in annotation_records
)

print("Total annotation records:", len(annotation_records))

print("\nTumour class distribution:")
print("Meningioma (1):", class_counts[1])
print("Glioma (2):", class_counts[2])
print("Pituitary tumour (3):", class_counts[3])

print("\nUnique patients:", len(set(
    record["patient_id"] for record in annotation_records
)))
```

```python
# Create a lookup table using original MATLAB filenames
annotation_lookup = {
    os.path.splitext(record["filename"])[0]: record
    for record in annotation_records
}

# Match each test image with its MATLAB annotation
test_records = []
unmatched_files = []

for image_path in test_image_paths:

    image_name = os.path.splitext(
        os.path.basename(image_path)
    )[0]

    if image_name in annotation_lookup:
        record = annotation_lookup[image_name]

        test_records.append({
            "filename": image_name + ".png",
            "label": record["label"],
            "patient_id": record["patient_id"]
        })
    else:
        unmatched_files.append(image_name)

print("Total test images:", len(test_image_paths))
print("Successfully matched:", len(test_records))
print("Unmatched images:", len(unmatched_files))

# Count test images by tumour type
test_class_counts = Counter(
    record["label"] for record in test_records
)

print("\nTest-set tumour distribution:")
print("Meningioma:", test_class_counts[1])
print("Glioma:", test_class_counts[2])
print("Pituitary tumour:", test_class_counts[3])

print(
    "\nUnique test patients:",
    len(set(record["patient_id"] for record in test_records))
)
```

```python
# Group individual test scores by tumour class

class_scores = {
    1: {"dice": [], "iou": []},
    2: {"dice": [], "iou": []},
    3: {"dice": [], "iou": []}
}

test_index = 0

for images, masks in test_dataset:

    predictions = unet_model.predict(images, verbose=0)

    predicted_masks = (predictions >= 0.5).astype(np.uint8)
    true_masks = masks.numpy().astype(np.uint8)

    for i in range(len(predicted_masks)):

        y_true = true_masks[i].flatten()
        y_pred = predicted_masks[i].flatten()

        intersection = np.sum(y_true * y_pred)
        total_true = np.sum(y_true)
        total_pred = np.sum(y_pred)
        union = total_true + total_pred - intersection

        dice = (2 * intersection + 1e-6) / (
            total_true + total_pred + 1e-6
        )

        iou = (intersection + 1e-6) / (union + 1e-6)

        # Assign scores using the matching MATLAB label
        label = test_records[test_index]["label"]

        class_scores[label]["dice"].append(dice)
        class_scores[label]["iou"].append(iou)

        test_index += 1

# Display class-wise results
class_names = {
    1: "Meningioma",
    2: "Glioma",
    3: "Pituitary tumour"
}

print("TUMOUR-TYPE-WISE SEGMENTATION PERFORMANCE")
print("=" * 60)

for label in [1, 2, 3]:

    dice_values = np.array(class_scores[label]["dice"])
    iou_values = np.array(class_scores[label]["iou"])

    print(f"\n{class_names[label]}")
    print("Number of images:", len(dice_values))
    print("Mean Dice:", round(np.mean(dice_values), 4))
    print("Mean IoU:", round(np.mean(iou_values), 4))
```

```python
print("DETAILED CLASS-WISE PERFORMANCE ANALYSIS")
print("=" * 60)

for label in [1, 2, 3]:

    dice_values = np.array(class_scores[label]["dice"])
    iou_values = np.array(class_scores[label]["iou"])

    print(f"\n{class_names[label]}")

    print("Number of slices:", len(dice_values))
    print("Mean Dice:", round(np.mean(dice_values), 4))
    print("Median Dice:", round(np.median(dice_values), 4))
    print("Dice standard deviation:", round(np.std(dice_values), 4))

    print("Mean IoU:", round(np.mean(iou_values), 4))
    print("Median IoU:", round(np.median(iou_values), 4))

    print(
        "Zero-Dice slices:",
        np.sum(dice_values < 1e-6)
    )

    print(
        "Dice below 0.5:",
        np.sum(dice_values < 0.5)
    )
```

```python
from skimage.measure import perimeter

tumour_shape_data = {
    1: {"perimeter": [], "circularity": []},
    2: {"perimeter": [], "circularity": []},
    3: {"perimeter": [], "circularity": []}
}

for index, mask_path in enumerate(test_mask_paths):

    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = mask.numpy().squeeze() > 0

    area = np.sum(mask)
    boundary = perimeter(mask, neighborhood=8)

    circularity = (
        4 * np.pi * area / (boundary ** 2)
        if boundary > 0 else 0
    )

    label = test_records[index]["label"]

    tumour_shape_data[label]["perimeter"].append(boundary)
    tumour_shape_data[label]["circularity"].append(circularity)

print("TUMOUR SHAPE ANALYSIS")
print("=" * 55)

for label in [1, 2, 3]:

    p = np.array(tumour_shape_data[label]["perimeter"])
    c = np.array(tumour_shape_data[label]["circularity"])

    print(f"\n{class_names[label]}")
    print("Mean perimeter:", round(np.mean(p), 2), "pixels")
    print("Mean circularity:", round(np.mean(c), 4))
```

```python
from skimage.measure import perimeter
from scipy.stats import spearmanr

# Store measurements and Dice scores by tumour class
shape_dice_analysis = {
    1: {"area": [], "circularity": [], "dice": []},
    2: {"area": [], "circularity": [], "dice": []},
    3: {"area": [], "circularity": [], "dice": []}
}

for index, mask_path in enumerate(test_mask_paths):

    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = mask.numpy().squeeze() > 0

    area = np.sum(mask)
    boundary = perimeter(mask, neighborhood=8)

    circularity = (
        4 * np.pi * area / (boundary ** 2)
        if boundary > 0 else 0
    )

    label = test_records[index]["label"]

    shape_dice_analysis[label]["area"].append(area)
    shape_dice_analysis[label]["circularity"].append(circularity)
    shape_dice_analysis[label]["dice"].append(
        all_dice_scores[index]
    )

print("RELATIONSHIP BETWEEN TUMOUR FEATURES AND DICE")
print("=" * 65)

for label in [1, 2, 3]:

    data = shape_dice_analysis[label]

    area_corr, area_p = spearmanr(
        data["area"], data["dice"]
    )

    circ_corr, circ_p = spearmanr(
        data["circularity"], data["dice"]
    )

    print(f"\n{class_names[label]}")

    print(
        f"Area vs Dice: correlation = {area_corr:.4f}, "
        f"p-value = {area_p:.4f}"
    )

    print(
        f"Circularity vs Dice: correlation = {circ_corr:.4f}, "
        f"p-value = {circ_p:.4f}"
    )
```

```python
# Compare predicted and ground-truth tumour areas

area_comparison = {
    1: {"ground_truth": [], "predicted": []},
    2: {"ground_truth": [], "predicted": []},
    3: {"ground_truth": [], "predicted": []}
}

test_index = 0

for images, masks in test_dataset:

    predictions = unet_model.predict(images, verbose=0)

    predicted_masks = predictions >= 0.5
    true_masks = masks.numpy() >= 0.5

    for i in range(len(predicted_masks)):

        gt_area = np.sum(true_masks[i])
        pred_area = np.sum(predicted_masks[i])

        label = test_records[test_index]["label"]

        area_comparison[label]["ground_truth"].append(gt_area)
        area_comparison[label]["predicted"].append(pred_area)

        test_index += 1

print("PREDICTED VS GROUND-TRUTH TUMOUR AREA")
print("=" * 60)

for label in [1, 2, 3]:

    gt = np.array(area_comparison[label]["ground_truth"])
    pred = np.array(area_comparison[label]["predicted"])

    print(f"\n{class_names[label]}")

    print("Mean ground-truth area:", round(np.mean(gt), 2))
    print("Mean predicted area:", round(np.mean(pred), 2))

    print(
        "Mean area difference (prediction - ground truth):",
        round(np.mean(pred - gt), 2)
    )
```

```python
print("OVERSEGMENTATION AND UNDERSEGMENTATION ANALYSIS")
print("=" * 60)

for label in [1, 2, 3]:

    gt = np.array(area_comparison[label]["ground_truth"])
    pred = np.array(area_comparison[label]["predicted"])

    over = np.sum(pred > gt)
    under = np.sum(pred < gt)
    equal = np.sum(pred == gt)

    print(f"\n{class_names[label]}")
    print("Total slices:", len(gt))

    print(
        f"Oversegmentation: {over} "
        f"({over / len(gt) * 100:.1f}%)"
    )

    print(
        f"Undersegmentation: {under} "
        f"({under / len(gt) * 100:.1f}%)"
    )

    print(
        f"Equal area: {equal} "
        f"({equal / len(gt) * 100:.1f}%)"
    )
```

```python
# Pixel-level precision and recall by tumour class

pixel_metrics = {
    1: {"precision": [], "recall": []},
    2: {"precision": [], "recall": []},
    3: {"precision": [], "recall": []}
}

test_index = 0

for images, masks in test_dataset:

    predictions = unet_model.predict(images, verbose=0)

    predicted_masks = predictions >= 0.5
    true_masks = masks.numpy() >= 0.5

    for i in range(len(predicted_masks)):

        pred = predicted_masks[i].flatten()
        true = true_masks[i].flatten()

        tp = np.sum(pred & true)
        fp = np.sum(pred & ~true)
        fn = np.sum(~pred & true)

        precision = tp / (tp + fp + 1e-6)
        recall = tp / (tp + fn + 1e-6)

        label = test_records[test_index]["label"]

        pixel_metrics[label]["precision"].append(precision)
        pixel_metrics[label]["recall"].append(recall)

        test_index += 1

print("PIXEL-LEVEL PRECISION AND RECALL")
print("=" * 55)

for label in [1, 2, 3]:

    precision = np.mean(pixel_metrics[label]["precision"])
    recall = np.mean(pixel_metrics[label]["recall"])

    print(f"\n{class_names[label]}")
    print("Mean precision:", round(precision, 4))
    print("Mean recall:", round(recall, 4))
```

```python
import h5py
import numpy as np
import os

# Check three glioma examples
glioma_indices = [
    i for i, record in enumerate(test_records)
    if record["label"] == 2
][:3]

def calculate_mask_iou(mask1, mask2):
    intersection = np.logical_and(mask1, mask2).sum()
    union = np.logical_or(mask1, mask2).sum()

    return intersection / union if union > 0 else 1.0

for index in glioma_indices:

    filename = test_records[index]["filename"].replace(".png", ".mat")
    mat_path = os.path.join(annotation_dir, filename)

    # Read original MATLAB mask
    with h5py.File(mat_path, "r") as mat_file:
        original_mask = np.array(
            mat_file["cjdata"]["tumorMask"]
        ).T

    original_mask = original_mask > 0

    # Read standardized PNG mask
    standardized_mask = tf.io.read_file(test_mask_paths[index])
    standardized_mask = tf.image.decode_png(
        standardized_mask, channels=1
    )
    standardized_mask = standardized_mask.numpy().squeeze() > 0

    # Resize original mask if necessary
    if original_mask.shape != standardized_mask.shape:
        original_mask = tf.image.resize(
            original_mask.astype(np.float32)[..., None],
            standardized_mask.shape,
            method="nearest"
        ).numpy().squeeze() > 0.5

    # Compare rotations and flips
    transformations = {
        "Original": original_mask,
        "Rotate 90": np.rot90(original_mask, 1),
        "Rotate 180": np.rot90(original_mask, 2),
        "Rotate 270": np.rot90(original_mask, 3),
        "Flip horizontal": np.fliplr(original_mask),
        "Flip vertical": np.flipud(original_mask)
    }

    scores = {
        name: calculate_mask_iou(transformed, standardized_mask)
        for name, transformed in transformations.items()
    }

    best_transform = max(scores, key=scores.get)

    print("\nImage:", filename)
    print("Best matching transformation:", best_transform)
    print("Best mask IoU:", round(scores[best_transform], 4))
    print("Direct mask IoU:", round(scores["Original"], 4))
```

```python
import h5py
import numpy as np
import os
import tensorflow as tf

check_indices = glioma_indices[:3]

for index in check_indices:

    filename = test_records[index]["filename"].replace(".png", ".mat")
    mat_path = os.path.join(annotation_dir, filename)

    with h5py.File(mat_path, "r") as mat_file:
        raw_mask = np.array(
            mat_file["cjdata"]["tumorMask"]
        )

    raw_mask = raw_mask > 0

    std_mask = tf.io.read_file(test_mask_paths[index])
    std_mask = tf.image.decode_png(std_mask, channels=1)
    std_mask = std_mask.numpy().squeeze() > 0

    # Test all eight rotations/reflections
    candidates = {
        "Original": raw_mask,
        "Rotate 90": np.rot90(raw_mask, 1),
        "Rotate 180": np.rot90(raw_mask, 2),
        "Rotate 270": np.rot90(raw_mask, 3),
        "Flip horizontal": np.fliplr(raw_mask),
        "Flip vertical": np.flipud(raw_mask),
        "Transpose": raw_mask.T,
        "Transpose + flip": np.fliplr(raw_mask.T)
    }

    results = {}

    for name, candidate in candidates.items():

        if candidate.shape != std_mask.shape:
            candidate = tf.image.resize(
                candidate.astype(np.float32)[..., None],
                std_mask.shape,
                method="nearest"
            ).numpy().squeeze() > 0.5

        intersection = np.logical_and(
            candidate, std_mask
        ).sum()

        union = np.logical_or(
            candidate, std_mask
        ).sum()

        iou = intersection / union if union > 0 else 1.0
        results[name] = iou

    best = max(results, key=results.get)

    print("\nImage:", filename)
    print("Original mask shape:", raw_mask.shape)
    print("Standardized mask shape:", std_mask.shape)
    print("Original tumour pixels:", raw_mask.sum())
    print("Standardized tumour pixels:", std_mask.sum())
    print("Best orientation:", best)
    print("Best IoU:", round(results[best], 4))
    print("All orientation scores:", results)
```

```python
import h5py
import numpy as np
import os
from scipy.stats import pearsonr

print("MRI IMAGE ORIENTATION VERIFICATION")
print("=" * 55)

for index in glioma_indices[:3]:

    filename = test_records[index]["filename"].replace(".png", ".mat")
    mat_path = os.path.join(annotation_dir, filename)

    with h5py.File(mat_path, "r") as mat_file:
        original_image = np.array(
            mat_file["cjdata"]["image"]
        )

    standardized = tf.io.read_file(test_image_paths[index])
    standardized = tf.image.decode_png(
        standardized, channels=1
    )
    standardized = standardized.numpy().squeeze().astype(np.float32)

    # Compare possible orientations
    candidates = {
        "Original": original_image,
        "Transpose": original_image.T,
        "Rotate 90": np.rot90(original_image, 1),
        "Rotate 180": np.rot90(original_image, 2),
        "Rotate 270": np.rot90(original_image, 3),
        "Flip horizontal": np.fliplr(original_image),
        "Flip vertical": np.flipud(original_image)
    }

    scores = {}

    for name, candidate in candidates.items():

        if candidate.shape != standardized.shape:
            continue

        candidate = candidate.astype(np.float32)

        # Apply the same percentile-based normalization
        low, high = np.percentile(candidate, [1, 99])

        candidate = np.clip(candidate, low, high)

        if high > low:
            candidate = (candidate - low) / (high - low)
        else:
            continue

        correlation = pearsonr(
            candidate.flatten(),
            standardized.flatten()
        )[0]

        scores[name] = correlation

    best = max(scores, key=scores.get)

    print(f"\nImage: {filename}")
    print("Best matching orientation:", best)
    print("Pixel correlation:", round(scores[best], 4))
```

```python
import os
import glob
import tensorflow as tf
import numpy as np

# Restore dataset folder paths
train_image_dir = os.path.join(
    standardized_dir, "train", "images"
)
train_mask_dir = os.path.join(
    standardized_dir, "train", "masks"
)

val_image_dir = os.path.join(
    standardized_dir, "validation", "images"
)
val_mask_dir = os.path.join(
    standardized_dir, "validation", "masks"
)

# Collect image and mask paths
train_image_paths = sorted(
    glob.glob(os.path.join(train_image_dir, "*.png"))
)
train_mask_paths = sorted(
    glob.glob(os.path.join(train_mask_dir, "*.png"))
)

val_image_paths = sorted(
    glob.glob(os.path.join(val_image_dir, "*.png"))
)
val_mask_paths = sorted(
    glob.glob(os.path.join(val_mask_dir, "*.png"))
)

# Load MRI images and masks
def load_image_and_mask(image_path, mask_path):

    image = tf.io.read_file(image_path)
    image = tf.image.decode_png(image, channels=1)
    image = tf.cast(image, tf.float32) / 255.0

    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = tf.cast(mask, tf.float32)

    return image, mask

# Create training dataset
train_dataset = tf.data.Dataset.from_tensor_slices(
    (train_image_paths, train_mask_paths)
)

train_dataset = train_dataset.shuffle(
    buffer_size=len(train_image_paths),
    seed=42
)

train_dataset = train_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_dataset = train_dataset.batch(2).prefetch(
    tf.data.AUTOTUNE
)

# Create validation dataset
val_dataset = tf.data.Dataset.from_tensor_slices(
    (val_image_paths, val_mask_paths)
)

val_dataset = val_dataset.map(
    load_image_and_mask,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_dataset = val_dataset.batch(2).prefetch(
    tf.data.AUTOTUNE
)

print("TRAINING DATASET:", len(train_image_paths), "images")
print("VALIDATION DATASET:", len(val_image_paths), "images")

print("Training and validation pipelines restored successfully!")
```

```python
import os
import tensorflow as tf

baseline_checkpoint = os.path.join(
    standardized_dir,
    "best_unet.weights.h5"
)

experimental_checkpoint = os.path.join(
    standardized_dir,
    "experimental_unet_dice_bce.weights.h5"
)

# Clone only the architecture, without compilation settings
experimental_model = tf.keras.models.clone_model(
    unet_model,
    compile=False
)

# Load the original trained weights
experimental_model.load_weights(baseline_checkpoint)

# Combined Dice + Binary Cross-Entropy loss
def dice_bce_loss(y_true, y_pred):
    bce = tf.reduce_mean(
        tf.keras.losses.binary_crossentropy(y_true, y_pred)
    )

    y_true_flat = tf.reshape(y_true, [tf.shape(y_true)[0], -1])
    y_pred_flat = tf.reshape(y_pred, [tf.shape(y_pred)[0], -1])

    intersection = tf.reduce_sum(
        y_true_flat * y_pred_flat, axis=1
    )

    dice = (
        2.0 * intersection + 1e-6
    ) / (
        tf.reduce_sum(y_true_flat, axis=1)
        + tf.reduce_sum(y_pred_flat, axis=1)
        + 1e-6
    )

    return bce + (1.0 - tf.reduce_mean(dice))

# Compile experimental model
experimental_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.00001
    ),
    loss=dice_bce_loss,
    metrics=[dice_coefficient, iou_metric]
)

print("Experimental model created successfully!")
print("Original checkpoint preserved:", os.path.exists(baseline_checkpoint))
print("Experimental model parameters:", experimental_model.count_params())
```

```python
import os
import tensorflow as tf

baseline_checkpoint = os.path.join(
    standardized_dir,
    "best_unet.weights.h5"
)

experimental_checkpoint = os.path.join(
    standardized_dir,
    "experimental_unet_dice_bce.weights.h5"
)

# Rebuild the architecture without copying compilation settings
experimental_model = unet_model.__class__.from_config(
    unet_model.get_config()
)

# Load the original trained weights
experimental_model.load_weights(baseline_checkpoint)

# Define combined Dice + Binary Cross-Entropy loss
def dice_bce_loss(y_true, y_pred):
    bce = tf.reduce_mean(
        tf.keras.losses.binary_crossentropy(y_true, y_pred)
    )

    y_true_flat = tf.reshape(y_true, [tf.shape(y_true)[0], -1])
    y_pred_flat = tf.reshape(y_pred, [tf.shape(y_pred)[0], -1])

    intersection = tf.reduce_sum(
        y_true_flat * y_pred_flat, axis=1
    )

    dice = (
        2.0 * intersection + 1e-6
    ) / (
        tf.reduce_sum(y_true_flat, axis=1)
        + tf.reduce_sum(y_pred_flat, axis=1)
        + 1e-6
    )

    return bce + (1.0 - tf.reduce_mean(dice))

# Compile the experimental model
experimental_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.00001),
    loss=dice_bce_loss,
    metrics=[dice_coefficient, iou_metric]
)

print("Experimental model created successfully!")
print("Original checkpoint preserved:", os.path.exists(baseline_checkpoint))
print("Model parameters:", experimental_model.count_params())
```

```python
import numpy as np
import os

# Restore the original baseline weights
unet_model.load_weights(baseline_checkpoint)

# Restore the best experimental weights (Epoch 2)
experimental_model.load_weights(experimental_checkpoint)

# Collect validation ground-truth masks
y_true_list = []

for images, masks in val_dataset:
    y_true_list.append(masks.numpy())

y_true = np.concatenate(y_true_list, axis=0)

# Generate predictions using both models
print("Generating baseline predictions...")
baseline_predictions = unet_model.predict(
    val_dataset, verbose=1
)

print("\nGenerating experimental predictions...")
experimental_predictions = experimental_model.predict(
    val_dataset, verbose=1
)

# Calculate per-image Dice and IoU consistently
def calculate_per_image_metrics(y_true, predictions):

    true_masks = y_true > 0.5
    predicted_masks = predictions > 0.5

    dice_scores = []
    iou_scores = []

    for true_mask, predicted_mask in zip(
        true_masks, predicted_masks
    ):
        intersection = np.logical_and(
            true_mask, predicted_mask
        ).sum()

        union = np.logical_or(
            true_mask, predicted_mask
        ).sum()

        dice = (
            2 * intersection /
            (true_mask.sum() + predicted_mask.sum() + 1e-7)
        )

        iou = intersection / (union + 1e-7)

        dice_scores.append(dice)
        iou_scores.append(iou)

    return np.array(dice_scores), np.array(iou_scores)


baseline_dice, baseline_iou = calculate_per_image_metrics(
    y_true, baseline_predictions
)

experimental_dice, experimental_iou = calculate_per_image_metrics(
    y_true, experimental_predictions
)

print("\n===== FAIR VALIDATION COMPARISON =====")

print("\nBASELINE U-NET")
print("Mean Dice:", round(baseline_dice.mean(), 4))
print("Mean IoU:", round(baseline_iou.mean(), 4))

print("\nEXPERIMENTAL U-NET")
print("Mean Dice:", round(experimental_dice.mean(), 4))
print("Mean IoU:", round(experimental_iou.mean(), 4))

print("\n===== CHANGE =====")
print("Dice improvement:",
      round(experimental_dice.mean() - baseline_dice.mean(), 4))
print("IoU improvement:",
      round(experimental_iou.mean() - baseline_iou.mean(), 4))
```

```python
import pandas as pd
import numpy as np

# Build a table containing image-level results and patient IDs
patient_df = pd.DataFrame({
    "patient_id": [r["patient_id"] for r in validation_records],
    "label": validation_labels,
    "baseline_dice": baseline_dice,
    "experimental_dice": experimental_dice,
    "baseline_iou": baseline_iou,
    "experimental_iou": experimental_iou
})

# Average slice scores within each patient
patient_results_validation = patient_df.groupby(
    ["patient_id", "label"]
).agg(
    slices=("baseline_dice", "count"),
    baseline_dice=("baseline_dice", "mean"),
    experimental_dice=("experimental_dice", "mean"),
    baseline_iou=("baseline_iou", "mean"),
    experimental_iou=("experimental_iou", "mean")
).reset_index()

# Calculate patient-level differences
patient_results_validation["dice_change"] = (
    patient_results_validation["experimental_dice"]
    - patient_results_validation["baseline_dice"]
)

patient_results_validation["iou_change"] = (
    patient_results_validation["experimental_iou"]
    - patient_results_validation["baseline_iou"]
)

print("PATIENT-LEVEL VALIDATION COMPARISON")
print("=" * 55)

for label, class_name in class_names.items():

    subset = patient_results_validation[
        patient_results_validation["label"] == label
    ]

    print(f"\n{class_name}")
    print("Patients:", len(subset))
    print("Mean patient Dice change:",
          round(subset["dice_change"].mean(), 4))
    print("Mean patient IoU change:",
          round(subset["iou_change"].mean(), 4))
    print("Patients improved in Dice:",
          int((subset["dice_change"] > 0).sum()),
          "of", len(subset))
    print("Patients improved in IoU:",
          int((subset["iou_change"] > 0).sum()),
          "of", len(subset))

print("\nOVERALL")
print("Validation patients:",
      patient_results_validation["patient_id"].nunique())

print("Patients improved in Dice:",
      int((patient_results_validation["dice_change"] > 0).sum()),
      "of", len(patient_results_validation))

print("Patients improved in IoU:",
      int((patient_results_validation["iou_change"] > 0).sum()),
      "of", len(patient_results_validation))
```

```python
import numpy as np

# Patient-level paired Dice differences
patient_differences = (
    patient_results_validation["experimental_dice"].to_numpy()
    - patient_results_validation["baseline_dice"].to_numpy()
)

# Patient-level bootstrap
rng = np.random.default_rng(42)

bootstrap_means = []

for _ in range(10000):

    sampled = rng.choice(
        patient_differences,
        size=len(patient_differences),
        replace=True
    )

    bootstrap_means.append(sampled.mean())

bootstrap_means = np.array(bootstrap_means)

# Mean difference and 95% confidence interval
mean_difference = patient_differences.mean()

lower_ci, upper_ci = np.percentile(
    bootstrap_means, [2.5, 97.5]
)

print("PATIENT-LEVEL BOOTSTRAP ANALYSIS")
print("-----------------------------------")
print("Number of patients:", len(patient_differences))
print("Mean Dice difference:", round(mean_difference, 4))
print("95% confidence interval:",
      (round(lower_ci, 4), round(upper_ci, 4)))

if lower_ci > 0:
    print("Interpretation: Experimental model shows a positive difference.")
elif upper_ci < 0:
    print("Interpretation: Experimental model shows a negative difference.")
else:
    print("Interpretation: The confidence interval includes zero.")
    print("No clear overall improvement is established.")
```

```python
import os
import numpy as np
from PIL import Image

dataset_splits = {
    "Training": (train_image_dir, train_mask_dir),
    "Validation": (val_image_dir, val_mask_dir),
    "Testing": (test_image_dir, test_mask_dir)
}

print("IMAGE-MASK INTEGRITY AUDIT")
print("=" * 55)

for split_name, (image_dir, mask_dir) in dataset_splits.items():

    image_files = set(os.listdir(image_dir))
    mask_files = set(os.listdir(mask_dir))

    missing_masks = image_files - mask_files
    missing_images = mask_files - image_files

    dimension_errors = []
    non_binary_masks = []
    unreadable_files = []

    for filename in sorted(image_files & mask_files):

        try:
            image = np.array(
                Image.open(os.path.join(image_dir, filename))
            )

            mask = np.array(
                Image.open(os.path.join(mask_dir, filename))
            )

            if image.shape != mask.shape:
                dimension_errors.append(filename)

            if not np.isin(np.unique(mask), [0, 1]).all():
                non_binary_masks.append(filename)

        except Exception:
            unreadable_files.append(filename)

    print(f"\n{split_name}")
    print("MRI files:", len(image_files))
    print("Mask files:", len(mask_files))
    print("Missing masks:", len(missing_masks))
    print("Missing images:", len(missing_images))
    print("Dimension mismatches:", len(dimension_errors))
    print("Non-binary masks:", len(non_binary_masks))
    print("Unreadable pairs:", len(unreadable_files))

    if not any([
        missing_masks,
        missing_images,
        dimension_errors,
        non_binary_masks,
        unreadable_files
    ]):
        print("STATUS: PASS")
    else:
        print("STATUS: CHECK REQUIRED")
```

```python
import pandas as pd

split_records = {
    "Training": train_records,
    "Validation": validation_records,
    "Testing": test_records
}

class_distribution = []

for split_name, records in split_records.items():

    counts = pd.Series(
        [r["label"] for r in records]
    ).value_counts()

    class_distribution.append({
        "Dataset": split_name,
        "Meningioma": counts.get(1, 0),
        "Glioma": counts.get(2, 0),
        "Pituitary tumour": counts.get(3, 0),
        "Total": len(records)
    })

distribution_df = pd.DataFrame(class_distribution)

display(distribution_df)

print("\nPATIENT-LEVEL CLASS DISTRIBUTION")

for split_name, records in split_records.items():

    patient_classes = {}

    for record in records:
        patient_classes.setdefault(
            record["patient_id"], set()
        ).add(record["label"])

    mixed_patients = {
        patient: labels
        for patient, labels in patient_classes.items()
        if len(labels) > 1
    }

    print(
        split_name,
        "- Patients:", len(patient_classes),
        "| Patients with multiple tumour labels:",
        len(mixed_patients)
    )
```

```python
limitations = """
PROJECT LIMITATIONS

1. Dataset scope:
The dataset contains 3,064 T1-weighted contrast-enhanced MRI slices
from 233 patients and includes three tumour classes: meningioma,
glioma, and pituitary tumour. It does not include a no-tumour class.

2. Class imbalance:
The number of slices differs across classes. Glioma has 1,426 slices,
meningioma 708, and pituitary tumour 930.

3. Test-set usage:
The test set was used for exploratory evaluation, visual inspection,
class-wise analysis, and error analysis. Therefore, it should not be
described as a completely untouched independent final test set.
Further model selection was based on validation results.

4. Class-wise segmentation performance:
The baseline model performed better on meningioma and pituitary
tumours than on glioma. Glioma segmentation remains a major challenge.

5. Experimental model:
The BCE + Dice fine-tuning experiment did not establish a clear
overall improvement in validation Dice. Performance improved for
meningioma and pituitary tumour but declined for glioma.

6. Patient-level uncertainty:
The exploratory patient-level bootstrap confidence interval for
the mean Dice difference included zero. Therefore, an overall
improvement cannot be established from this experiment.

7. Generalisation:
The dataset contains 233 patients. Evaluation on additional
independent datasets and institutions would be needed to assess
broader generalisation.

8. Scope:
The model performs tumour-region segmentation on individual MRI
slices. The results do not establish clinical diagnostic accuracy
or clinical readiness.
"""

limitations_path = os.path.join(final_dir, "project_limitations.txt")

with open(limitations_path, "w") as f:
    f.write(limitations.strip())

print("Limitations file saved successfully:")
print(limitations_path)
```
